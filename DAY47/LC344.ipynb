# LC 344 - Reverse String

## 1. Problem

Given an array of characters `s`, reverse the array **in-place**.

Example:

Input:
`s = ["h","e","l","l","o"]`

Output:
`["o","l","l","e","h"]`

Important requirement:

**Modify the original array directly. Do not create another array.**

---

## 2. Core Idea

Use the **Two Pointer** approach.

- `left` starts at the beginning.
- `right` starts at the end.
- Swap `s[left]` and `s[right]`.
- Move both pointers toward the center.
- Stop when `left >= right`.

Pattern:

`Left → Swap ← Right`

---

## 3. Pointer Initialization

For:

`s = ["h","e","l","l","o"]`

Indexes:

`0   1   2   3   4`

Initialize:

`left = 0`

`right = n - 1`

Therefore:

`left = 0`

`right = 4`

---

## 4. Why Two Pointers?

The first element needs to move to the last position.

The last element needs to move to the first position.

So instead of moving elements one by one, directly swap the two ends.

Example:

`["h","e","l","l","o"]`

Swap:

`h ↔ o`

Result:

`["o","e","l","l","h"]`

Then move inward and repeat.

---

## 5. Main Algorithm

At every step:

1. Swap `s[left]` and `s[right]`.
2. Increment `left`.
3. Decrement `right`.
4. Continue while `left < right`.

Mental model:

`Swap → Move Inward → Swap → Move Inward`

---

## 6. Complete Dry Run

Input:

`s = ["h","e","l","l","o"]`

Initial:

`left = 0`

`right = 4`

### Step 1

`left → "h"`

`right → "o"`

Swap:

`"h" ↔ "o"`

Array:

`["o","e","l","l","h"]`

Move:

`left = 1`

`right = 3`

---

### Step 2

`left → "e"`

`right → "l"`

Swap:

`"e" ↔ "l"`

Array:

`["o","l","l","e","h"]`

Move:

`left = 2`

`right = 2`

---

### Step 3

Now:

`left == right`

Both pointers are at the middle.

No swap is needed.

Loop stops.

Final:

`["o","l","l","e","h"]`

---

## 7. Why `while left < right`?

We only need to swap two different positions.

When:

`left < right`

there are still two elements that need to be swapped.

When:

`left == right`

we are at the middle element.

A middle element stays where it is.

Therefore:

`while left < right`

is sufficient.

---

## 8. Odd-Length Array

Example:

`["a","b","c","d","e"]`

Swaps:

`a ↔ e`

`b ↔ d`

The middle element:

`c`

does not move.

Final:

`["e","d","c","b","a"]`

---

## 9. Even-Length Array

Example:

`["a","b","c","d"]`

Swaps:

`a ↔ d`

`b ↔ c`

Final:

`["d","c","b","a"]`

There is no single middle element.

---

## 10. In-Place Requirement

The problem requires modifying `s` directly.

We do not create:

`new_array = []`

or:

`reversed_s = ...`

We simply swap elements inside the original array.

Therefore the solution uses:

**O(1) auxiliary space**

---

## 11. Code

class Solution(object):
    def reverseString(self, s):
        """
        :type s: List[str]
        :rtype: None Do not return anything, modify s in-place instead.
        """
        n = len(s)
        left = 0
        right = n - 1

        while left < right:
            s[left], s[right] = s[right], s[left]
            left += 1
            right -= 1

---

## 12. Code Breakdown

`n = len(s)`

Find the length of the array.

`left = 0`

Start from the first index.

`right = n - 1`

Start from the last index.

`while left < right`

Continue until the pointers meet.

`s[left], s[right] = s[right], s[left]`

Swap the two elements.

`left += 1`

Move the left pointer inward.

`right -= 1`

Move the right pointer inward.

---

## 13. Why Is Time O(n)?

Although the array has `n` elements, we only process approximately half of them.

Each swap puts **two elements** into their correct positions.

Number of operations:

`n / 2`

Ignoring constants:

`Time = O(n)`

---

## 14. Space Complexity

Only a constant number of variables are used:

- `n`
- `left`
- `right`

No additional array is created.

Therefore:

`Space = O(1)`

---

## 15. Common Mistakes

### Mistake 1: Creating Another Array

This violates the in-place requirement.

---

### Mistake 2: Moving Only One Pointer

Both pointers must move:

`left += 1`

`right -= 1`

---

### Mistake 3: Using `left <= right`

Not necessary.

The middle element does not need to be swapped.

Use:

`while left < right`

---

### Mistake 4: Wrong Right Boundary

The last index is:

`n - 1`

not:

`n`

---

### Mistake 5: Returning `s`

The LeetCode function asks for:

`Do not return anything, modify s in-place instead.`

So:

`return s`

is unnecessary.

The array itself has already been modified.

## 16. Pattern Recognition

### Pattern

**Two Pointers + In-Place Swapping**

Whenever you see:

- Array or string
- Reverse
- In-place requirement
- Elements from opposite ends need to exchange positions

Think:

**Two Pointers**

General structure:

`left = 0`

`right = n - 1`

↓

`swap(left, right)`

↓

`left++`

`right--`

↓

Repeat until:

`left >= right`

```python
class Solution(object):
    def reverseString(self, s):
        """
        :type s: List[str]
        :rtype: None Do not return anything, modify s in-place instead.
        """
        n=len(s)
        left=0
        right=n-1
        while left<right:
            s[left],s[right]=s[right],s[left]
            left+=1
            right-=1
        return s
```
