# LC 167 - Two Sum II - Input Array Is Sorted

## 1. Problem

Given a **sorted** array `numbers` and a target value, find two numbers whose sum equals the target.

Return their **1-indexed positions**.

### Example

Input:

`numbers = [2,7,11,15]`

`target = 9`

Output:

`[1,2]`

Because:

`2 + 7 = 9`

---

## 2. Core Idea

Use the **Two Pointer** approach.

- `left` starts at the beginning.
- `right` starts at the end.
- Calculate the sum of the two values.
- Use the sorted property to decide which pointer to move.

### Mental Model

`left → [smaller values ........ larger values] ← right`

---

## 3. Pointer Initialization

Initialize:

`left = 0`

`right = n - 1`

For:

`numbers = [2,7,11,15]`

Think:

`Index:   0  1   2   3`

`Value:  [2, 7, 11, 15]`

`          ↑          ↑`

`        left       right`

---

## 4. Main Decision Logic

Calculate:

`sum = numbers[left] + numbers[right]`

There are three cases.

### Case 1: Sum is too small

`sum < target`

We need a larger sum.

Move:

`left += 1`

### Case 2: Sum is too large

`sum > target`

We need a smaller sum.

Move:

`right -= 1`

### Case 3: Sum equals target

`sum == target`

We found the answer.

Return:

`[left + 1, right + 1]`

### Memory Rule

**Too Small → Left++**

**Too Large → Right--**

**Equal → Found**

---

## 5. Why Move `left` When Sum Is Too Small?

Suppose:

`numbers[left] + numbers[right] < target`

The array is sorted.

If we move `right` backward, we choose an even smaller number.

That makes the sum smaller.

That cannot help.

So we move `left` forward.

Because the array is sorted, moving `left` forward gives us a larger value.

Therefore:

`sum < target → left++`

---

## 6. Why Move `right` When Sum Is Too Large?

Suppose:

`numbers[left] + numbers[right] > target`

If we move `left` forward, we choose a larger number.

That makes the sum even larger.

That cannot help.

So we move `right` backward.

Because the array is sorted, moving `right` backward gives us a smaller value.

Therefore:

`sum > target → right--`

---

## 7. Why Does Sorting Matter?

The two-pointer logic works because the array is sorted.

Moving:

`left →`

means the value can only increase.

Moving:

`← right`

means the value can only decrease.

Therefore, based on whether the current sum is too small or too large, we know exactly which pointer to move.

Without sorting, this logic would not work.

## 12. Why Return `left + 1` and `right + 1`?

Python uses **0-based indexing**.

`Index:     0  1   2   3`

`Position:  1  2   3   4`

`Value:    [2, 7, 11, 15]`

The problem asks for **1-indexed positions**.

Therefore:

`index 0 → position 1`

`index 1 → position 2`

So we return:

`[left + 1, right + 1]`

---

## 13. Why `while left < right`?

We need two different elements.

If:

`left == right`

both pointers are pointing to the same element.

We cannot use the same element twice.

Therefore:

`while left < right`

is correct.

---

## 14. Brute Force vs Two Pointers

### Brute Force

Check every possible pair.

Time complexity:

`O(n²)`

### Two Pointer

Start from both ends and eliminate impossible pairs using the sorted property.

Time complexity:

`O(n)`

The sorted property allows us to eliminate many impossible pairs without checking them individually.

## 15. Code

class Solution(object):
    def twoSum(self, numbers, target):
        """
        :type numbers: List[int]
        :type target: int
        :rtype: List[int]
        """
        n = len(numbers)
        left = 0
        right = n - 1

        while left < right:
            sum = numbers[left] + numbers[right]

            if sum < target:
                left += 1
            elif sum > target:
                right -= 1
            if sum == target:
                return [left + 1, right + 1]

---

### Step 2

### Step 3

### Step 4

### Step 7

`[left + 1, right + 1]`

## 17. Important Code Observation

Your code uses:

`if sum < target:`

`    left += 1`

`elif sum > target:`

`    right -= 1`

`if sum == target:`

`    return [left + 1, right + 1]`

This works because `sum` was calculated before moving the pointer.

A cleaner structure is:

`if sum < target:`

`    left += 1`

`elif sum > target:`

`    right -= 1`

`else:`

`    return [left + 1, right + 1]`

The algorithmic idea is exactly the same.

## 18. Common Mistakes

### Mistake 1: Using a Hash Map Automatically

Normal Two Sum often uses a hash map.

But LC 167 gives us a **sorted array**.

That sorted property should immediately trigger:

**Two Pointers**

### Mistake 2: Moving the Wrong Pointer

Remember:

`sum < target → left++`

`sum > target → right--`

### Mistake 3: Returning 0-Based Indexes

The problem requires 1-based positions.

Wrong:

`[left, right]`

Correct:

`[left + 1, right + 1]`

### Mistake 4: Using `left <= right`

This can allow the same element to be used twice.

Use:

`left < right`

### Mistake 5: Ignoring the Sorted Property

The entire pointer movement strategy depends on the array being sorted.

## 19. Complexity

### Time Complexity

`O(n)`

Both pointers move only toward each other.

`left` only moves forward.

`right` only moves backward.

Therefore the total number of pointer movements is linear.

### Space Complexity

`O(1)`

Only constant-size variables are used.

No hash map or additional array is required.
