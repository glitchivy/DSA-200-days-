# LC 125 - Valid Palindrome

## 1. Problem

Given a string `s`, determine whether it is a palindrome after:

- Converting uppercase letters to lowercase
- Removing all non-alphanumeric characters

A palindrome reads the same forward and backward.

### Example

Input:

`s = "A man, a plan, a canal: Panama"`

After removing spaces and punctuation and converting to lowercase:

`"amanaplanacanalpanama"`

Output:

`True`

---

## 2. Core Idea

Use the **Two Pointer** approach.

- `left` starts from the beginning.
- `right` starts from the end.
- Ignore characters that are not alphanumeric.
- Compare the valid characters from both ends.
- If they differ, return `False`.
- If they match, move both pointers inward.

### Mental Model

`left →  [valid characters]  ← right`

Compare both sides and move toward the center.

---

## 3. Pointer Initialization

For:

`s = "A man, a plan, a canal: Panama"`

Initialize:

`left = 0`

`right = len(s) - 1`

So:

```text
left  → A
...
right → a

````python
class Solution(object):
    def isPalindrome(self, s):
        """
        :type s: str
        :rtype: bool
        """
        left=0
        right=len(s)-1
        while left<right:
            if not s[left].isalnum():
                left+=1
            elif not s[right].isalnum():
                right-=1
            else:
                if s[left].lower()==s[right].lower():
                    left+=1
                    right-=1
                else:
                    return False
        return True
````
