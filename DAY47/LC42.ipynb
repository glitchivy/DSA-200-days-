# LC 42 - Trapping Rain Water

## 1. Problem

Given an array `height`, where each element represents the height of a vertical bar, calculate how much rainwater can be trapped.

### Example

Input:
`height = [0,1,0,2,1,0,1,3,2,1,2,1]`

Output:
`6`

### Main Formula

For any index:

`Water = min(maxLeft, maxRight) - height[i]`

The smaller boundary determines the maximum water level.

---

## 2. Core Idea

Water can be trapped only when there is:

- A wall on the left
- A wall on the right
- A lower region between them

Example:

`[3, 0, 2, 0, 4]`

The left boundary is `3`.

The right boundary is `4`.

The water level is:

`min(3,4) = 3`

So:

- At `0`: `3 - 0 = 3`
- At `2`: `3 - 2 = 1`
- At `0`: `3 - 0 = 3`

Total:

`3 + 1 + 3 = 7`

---

## 3. Brute Force Approach

For every index:

1. Find the maximum height on the left.
2. Find the maximum height on the right.
3. Take the smaller maximum.
4. Subtract the current height.
5. Add the result to the answer.

Formula:

`water = min(maxLeft, maxRight) - height[i]`

### Complexity

Time: `O(n²)`

Space: `O(1)`

The problem is that we repeatedly search for left and right maximums.

---

## 4. Better Approach

We can precompute:

`leftMax[i]`

and:

`rightMax[i]`

Then each index can calculate its water in `O(1)`.

### Complexity

Time: `O(n)`

Space: `O(n)`

This is better in time, but we can remove the extra arrays.

---

## 5. Optimal Approach

Use:

**Two Pointers + Running Maximums**

Initialize:

`left = 0`

`right = n - 1`

Maintain:

`maxLeft = 0`

`maxRight = 0`

`water = 0`

At every step compare:

`height[left]`

and:

`height[right]`

---

## 6. Most Important Decision

### If:

`height[left] <= height[right]`

Process the left side.

### Otherwise:

`height[left] > height[right]`

Process the right side.

### Mental Rule

**Always process the side with the smaller current boundary.**

This is the key idea of the two-pointer solution.

---

## 7. Why Do We Process the Smaller Side?

Suppose:

`height[left] <= height[right]`

We know the right side has a boundary at least as tall as the current left boundary.

Therefore, the left side is the limiting side.

So we can safely calculate the water on the left using:

`maxLeft - height[left]`

We don't need the exact maximum right boundary at this point.

Similarly, when:

`height[right] < height[left]`

the right side is the limiting side.

So we process the right.

---

## 8. What Is `maxLeft`?

`maxLeft` stores the tallest wall we have encountered from the left.

Example:

`[3,1,2]`

When we reach `1`:

`maxLeft = 3`

Water:

`3 - 1 = 2`

Update using:

`maxLeft = max(maxLeft, height[left])`

Therefore `maxLeft` always stores the highest left boundary seen so far.

---

## 9. What Is `maxRight`?

`maxRight` stores the tallest wall we have encountered from the right.

Example:

`[3,1,2]`

From the right:

`maxRight = 2`

At height `1`:

`water = 2 - 1`

`water = 1`

Update:

`maxRight = max(maxRight, height[right])`

---

## 10. Why Update Maximum Before Calculating Water?

Suppose:

`height[left] = 5`

and:

`maxLeft = 3`

If we calculate immediately:

`3 - 5 = -2`

That cannot represent water.

Instead:

`maxLeft = max(3,5)`

So:

`maxLeft = 5`

Then:

`water += 5 - 5`

`water += 0`

Therefore:

**Update the maximum first, then calculate water.**

---

## 11. Complete Dry Run

Input:

`height = [3,0,2,0,4]`

Expected answer:

`7`

Initial state:

```text
left = 0
right = 4
maxLeft = 0
maxRight = 0
water = 0

````python
class Solution(object):
    def trap(self, height):
        """
        :type height: List[int]
        :rtype: int
        """
        #intailize
        left=0
        right=len(height)-1
        maxright=0
        maxleft=0 
        water=0 
        while left<right:
            #decide
            if height[left] <= height[right]:
                maxleft=max(maxleft,height[left])
                water+=maxleft-height[left]
                left+=1
            else:
                maxright=max(maxright,height[right])
                water+=maxright-height[right]
                right-=1
        return water
````
