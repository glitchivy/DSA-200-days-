# LC 41 - First Missing Positive

## Problem

Given an unsorted integer array `nums`, find the smallest missing positive integer.

Examples:
- `[1,2,0]` → `3`
- `[3,4,-1,1]` → `2`
- `[7,8,9,11,12]` → `1`

## Pattern

**Cyclic Sort / Index Placement**

For an array of length `n`:

`1 → index 0`
`2 → index 1`
`3 → index 2`
`...`
`n → index n-1`

Therefore, for a number `x`, its correct index is:

`x - 1`

---

## Approach

Only numbers in the range:

`1 <= nums[i] <= n`

are useful.

For every index `i`:

1. Look at `nums[i]`.
2. If it is between `1` and `n`, find its correct index using `nums[i] - 1`.
3. If it is not already in its correct position, swap it into that position.
4. Keep doing this with a `while` loop because after a swap, a new number comes into `nums[i]`.
5. Ignore negative numbers, `0`, and numbers greater than `n`.

The duplicate check:

`nums[i] != nums[nums[i] - 1]`

prevents infinite swapping when the target position already contains the same value.

---

## Code

class Solution(object):
    def firstMissingPositive(self, nums):
        """
        :type nums: List[int]
        :rtype: int
        """
        n = len(nums)

        for i in range(n):
            while 1 <= nums[i] <= n and nums[i] != nums[nums[i] - 1]:
                target = nums[i] - 1
                nums[i], nums[target] = nums[target], nums[i]

        for i in range(n):
            if nums[i] != i + 1:
                return i + 1

        return n + 1

## Dry Run

Input:

`[3,1,2]`

### i = 0

Current number:

`3`

Correct index:

`3 - 1 = 2`

Swap:

`[2,1,3]`

We stay at `i = 0`.

Current number is now:

`2`

Correct index:

`2 - 1 = 1`

Swap:

`[1,2,3]`

Current number is now:

`1`

Correct index:

`0`

It is already home, so move to the next index.

Final:

`[1,2,3]`

---

## Finding the Answer

After placement, the ideal array is:

`[1,2,3,4,...]`

So at index `i`, we expect:

`i + 1`

The first mismatch gives the answer.

Example:

`[1,-1,3,4]`

Index `0`:

`1 == 1` → correct

Index `1`:

`-1 != 2` → mismatch

Therefore:

`answer = 1 + 1 = 2`

If every position is correct, then numbers `1` through `n` all exist.

Therefore:

`answer = n + 1`

---

## Why O(n) Time Despite the While Loop?

At first glance, this looks like:

`for loop + while loop = O(n²)`

But that is not automatically true.

Every successful iteration of the `while` loop performs a swap that places an element into its correct position.

Across the entire algorithm, there can only be `O(n)` useful swaps.

Therefore:

`Outer loop = O(n)`

`Total swaps = O(n)`

So:

`Total Time = O(n)`

This is an example of **amortized analysis**.

### Interview Explanation

> Although there is a nested while loop, every successful iteration places an element into its correct position. Across the entire algorithm, there can be at most O(n) such placements. Therefore, the total complexity remains O(n).

---

## Space Complexity

The array is modified in-place.

No extra array, set, or hash map is used.

Therefore:

`Space = O(1)`

## Why We Don't Sort

Normal sorting would take:

`O(n log n)`

But the problem requires an `O(n)` solution with `O(1)` extra space.

Instead of sorting, we exploit the relationship between:

`number → index`

This allows us to place values directly where they belong.

Mental model:

`x → index x - 1`

Then:

`Place numbers → Scan array → Find first broken position`

---

## Key Takeaway

The main idea is:

> Use the array's indices as a substitute for extra memory.

Every useful number tries to go to its "home":

`1 → 0`
`2 → 1`
`3 → 2`
`...`

After placement, the first index `i` where:

`nums[i] != i + 1`

means:

`i + 1`

is the first missing positive.
