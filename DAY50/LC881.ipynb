# LC 881 - Boats to Save People

## 1. Problem

You are given:

- `people` where `people[i]` is the weight of a person.
- `limit` which is the maximum weight a boat can carry.

Each boat can carry **at most two people**.

Find the minimum number of boats required to rescue everyone.

### Example

Input:

`people = [1,2]`

`limit = 3`

Output:

`1`

Because:

`1 + 2 = 3`

So both people can share one boat.

---

## 2. Core Idea

Use:

**Sorting + Two Pointers + Greedy**

Sort the people by weight.

Then use:

- `left` → lightest person
- `right` → heaviest person

Always consider the heaviest person first.

Try to pair the heaviest person with the lightest person.

### Memory Rule

**Heaviest must go → try pairing with lightest**

If they fit together:

`lightest + heaviest <= limit`

put both in one boat.

Otherwise:

the heaviest person must go alone.

---

## 3. Why Sort?

Sorting lets us make a greedy decision.

Example:

`people = [3,2,2,1]`

After sorting:

`[1,2,2,3]`

Suppose the heaviest person weighs:

`3`

Try the lightest:

`1`

`3 + 1 = 4`

If:

`limit = 3`

they cannot share a boat.

Since `3` is already the heaviest person, pairing them with anyone else would only be heavier.

Therefore:

**3 must take a boat alone.**

This is why sorting is powerful here.

---

## 4. Pointer Initialization

Initialize:

`left = 0`

`right = len(people) - 1`

`boat = 0`

Meaning:

- `left` points to the lightest remaining person.
- `right` points to the heaviest remaining person.
- `boat` counts the boats used.

---

## 5. Main Decision Logic

Calculate:

`value = people[left] + people[right]`

Then check whether the two people can share a boat.

### Case 1

If:

`value <= limit`

Both people fit.

Put them together.

Move:

`left += 1`

`right -= 1`

Increase:

`boat += 1`

### Case 2

If:

`value > limit`

They cannot fit together.

The heaviest person must go alone.

Move:

`right -= 1`

Increase:

`boat += 1`

### Memory Rule

**They fit → both move**

**They don't fit → only right moves**

---

## 6. Why Does the Heaviest Person Go Alone?

Suppose:

`people[right] = 80`

`people[left] = 50`

and:

`limit = 100`

Then:

`80 + 50 = 130`

Too heavy.

Could we pair the `80` with someone else?

No.

The person at `left` is already the lightest person.

Everyone else is at least as heavy as `50`.

So:

`80 + any other person >= 80 + 50`

Therefore the `80` cannot share a boat.

It must go alone.

This is the greedy proof.

## 15. Code

class Solution(object):
    def numRescueBoats(self, people, limit):
        """
        :type people: List[int]
        :type limit: int
        :rtype: int
        """
        people.sort()
        left = 0
        right = len(people) - 1
        boat = 0

        while left <= right:
            value = people[left] + people[right]

            if value <= limit:
                right -= 1
                boat += 1
                left += 1

            elif value > limit:
                right -= 1
                boat += 1

            else:
                boat += 1
                right -= 1

        return boat

## 17. Important Code Observation

Your final `else` is actually unnecessary.

You already check:

`if value <= limit`

and:

`elif value > limit`

These two conditions cover every possible value.

So the final:

`else`

can never execute.

Conceptually, there are only **two cases**:

1. They fit → both go.
2. They do not fit → heaviest goes alone.

The algorithm's logic is therefore cleaner as a two-case decision.

---

## 18. Why This Is Greedy

At every step, we make the locally optimal decision:

**Take the heaviest remaining person.**

Then:

- If the lightest person can fit with them, pair them.
- Otherwise, send the heaviest person alone.

Why is this safe?

Because if the heaviest person cannot fit with the lightest person, they cannot fit with anyone else.

Therefore we never need to reconsider that person.

This permanently reduces the problem.

```python
class Solution(object):
    def numRescueBoats(self, people, limit):
        """
        :type people: List[int]
        :type limit: int
        :rtype: int
        """
        people.sort()
        left=0
        right=len(people)-1
        boat=0
        while left<=right:
            value=people[left]+people[right]
            if value <= limit:
                right-=1
                boat+=1
                left+=1
            elif value>limit:
                right-=1
                boat+=1
            else:
                boat+=1   
                right-=1
        return boat
```
