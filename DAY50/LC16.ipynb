# LC 16 - 3Sum Closest

## 1. Problem

Given an integer array `nums` and an integer `target`, find three integers in `nums` such that their sum is closest to `target`.

Return the sum of those three integers.

### Important

You must return the **sum**, not the three numbers themselves.

### Example

Input:

`nums = [-1,2,1,-4]`

`target = 1`

Possible sums include:

`-1 + 2 + 1 = 2`

`-1 + 2 + -4 = -3`

`-1 + 1 + -4 = -4`

`2 + 1 + -4 = -1`

The closest sum to `1` is:

`2`

Output:

`2`

---

## 2. Core Idea

This problem is almost the same pattern as **3Sum**.

Use:

**Sorting + Fix One + Two Pointers**

The difference is:

### 3Sum

We need:

`sum == 0`

### 3Sum Closest

We need:

`sum` as close as possible to `target`

So instead of only checking whether the sum equals the target, we maintain the best sum found so far.

### Memory Rule

**Sort → Fix `i` → Two Pointers → Calculate difference → Update closest**

---

## 3. Why Do We Sort?

Sorting allows the two-pointer movement to work.

After sorting:

- If `current < target`, increase the sum by moving `left`.
- If `current > target`, decrease the sum by moving `right`.
- If `current == target`, we have the best possible answer and can return immediately.

Example:

`[-4,-1,1,2]`

Because the array is ordered, pointer movement directly controls the sum.

---

## 4. Initial Closest Value

Your code initializes:

`close = nums[0] + nums[1] + nums[2]`

This gives us a valid initial three-number sum.

Then every new combination is compared against this value.

We update `close` whenever the new sum is closer to the target.

### Memory Rule

**Always keep the best sum found so far.**

---

## 5. Difference From Target

For every current sum:

`current`

calculate:

`abs(current - target)`

This tells us how far the current sum is from the target.

Example:

`target = 10`

Current sum:

`7`

Difference:

`|7 - 10| = 3`

Another sum:

`12`

Difference:

`|12 - 10| = 2`

Therefore:

`12` is closer.

---

## 6. Pointer Initialization

For every fixed index:

`i`

initialize:

`left = i + 1`

`right = n - 1`

So the three elements are:

`nums[i]`

`nums[left]`

`nums[right]`

This is the same setup used in 3Sum.

---

## 7. Main Decision Logic

Calculate:

`current = nums[i] + nums[left] + nums[right]`

Then compare its distance from the target.

### Step 1

Calculate:

`abs(current - target)`

### Step 2

Compare with:

`abs(close - target)`

If the current sum is closer:

`close = current`

### Step 3

Decide pointer movement.

If:

`current < target`

the sum is too small.

Move:

`left += 1`

If:

`current > target`

the sum is too large.

Move:

`right -= 1`

If:

`current == target`

we have an exact answer.

Return:

`current`

---

## 8. Why Move `left` When `current < target`?

The array is sorted.

If the current sum is too small, we need a larger sum.

Moving:

`left += 1`

moves toward a larger value.

Therefore the total sum can increase.

### Memory Rule

**Too small → increase `left`**

---

## 9. Why Move `right` When `current > target`?

If the current sum is too large, we need a smaller sum.

Because the array is sorted, moving:

`right -= 1`

moves toward a smaller value.

Therefore the total sum can decrease.

### Memory Rule

**Too large → decrease `right`**

## 15. Code

class Solution(object):
    def threeSumClosest(self, nums, target):
        """
        :type nums: List[int]
        :type target: int
        :rtype: int
        """
        nums.sort()
        current = 0
        n = len(nums)
        close = nums[0] + nums[1] + nums[2]

        for i in range(n):
            left = i + 1
            right = n - 1

            while left < right:
                current = nums[i] + nums[left] + nums[right]
                diff = abs(current - target)

                if abs(current - target) < abs(close - target):
                    close = current

                if current < target:
                    left += 1
                elif current > target:
                    right -= 1
                else:
                    return current

        return close

## 16. Code Breakdown

### Step 1

`nums.sort()`

Sort the array so that pointer movement becomes meaningful.

### Step 2

`n = len(nums)`

Store the array length.

### Step 3

`close = nums[0] + nums[1] + nums[2]`

Start with a valid three-element sum as the initial closest value.

### Step 4

`for i in range(n):`

Fix one element.

This turns the problem into a two-pointer search for the other two elements.

### Step 5

`left = i + 1`

Start the second pointer immediately after `i`.

### Step 6

`right = n - 1`

Start the third pointer at the end.

### Step 7

`current = nums[i] + nums[left] + nums[right]`

Calculate the current three-number sum.

### Step 8

`diff = abs(current - target)`

Calculate how far the current sum is from the target.

### Step 9

`if abs(current - target) < abs(close - target):`

Check whether the current sum is closer than the best sum found so far.

If yes:

`close = current`

### Step 10

If:

`current < target`

move:

`left += 1`

### Step 11

If:

`current > target`

move:

`right -= 1`

### Step 12

If:

`current == target`

return immediately.

An exact target has:

`difference = 0`

which is the smallest possible difference.

### Step 13

After all combinations:

`return close`

Return the closest sum found.

## 17. Important Code Observation

The most important difference from **3Sum** is this:

### 3Sum

We only care about:

`current == 0`

### 3Sum Closest

We care about:

`abs(current - target)`

So the algorithm continuously maintains:

**Which sum is currently closest to the target?**

This is the key modification.

---

## 18. Duplicate Handling

Unlike the previous **3Sum** problem, your code does not explicitly skip duplicate values.

That is fine here.

Why?

The goal is not to return unique triplets.

We only need the closest **sum**.

Even if the same combination is considered more than once, it does not change the final closest value.

### Important Difference

**3Sum:**

Duplicate triplets matter.

Therefore duplicate skipping is necessary.

**3Sum Closest:**

We return only one integer.

Duplicate combinations do not create duplicate output.

Therefore explicit duplicate skipping is not required.

## 19. Common Mistakes

### Mistake 1: Using `abs(current)` Instead of `abs(current - target)`

The target is not always zero.

You need:

`abs(current - target)`

Example:

`target = 10`

`current = 9`

Distance:

`1`

not:

`9`

### Mistake 2: Moving the Wrong Pointer

If:

`current < target`

you need a larger sum.

Move:

`left`

If:

`current > target`

you need a smaller sum.

Move:

`right`

### Mistake 3: Forgetting to Update `close`

Finding a sum is not enough.

You must compare it with the best sum found so far.

### Mistake 4: Returning the First Sum

The first sum is not necessarily the closest.

You must continue searching unless:

`current == target`

### Mistake 5: Confusing This With 3Sum

3Sum asks:

**Does a triplet sum to zero?**

3Sum Closest asks:

**Which triplet produces the sum closest to the target?**

## 20. Complexity + Pattern Recognition

### Time Complexity

Sorting:

`O(n log n)`

For every fixed `i`, the two pointers scan the array:

`O(n)`

Across all `i` values:

`O(n²)`

Overall:

`O(n²)`

### Space Complexity

`O(1)` extra space, excluding the sorting implementation and output, since no result array is created.
