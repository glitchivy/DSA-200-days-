# LC 75 - Sort Colors

## 1. Problem

Given an array `nums` containing only:

`0`, `1`, and `2`

sort the array in-place so that all:

- `0`s come first
- `1`s come next
- `2`s come last

You must solve it without using a built-in sorting function.

### Example

Input:

`nums = [2,0,2,1,1,0]`

Output:

`[0,0,1,1,2,2]`

---

## 2. Core Idea

Use the **Dutch National Flag Algorithm**.

Maintain three regions:

- `0` → left region
- `1` → middle region
- `2` → right region

Use three pointers:

`low`

`mid`

`high`

### Memory Rule

**0 → send left**

**1 → leave in middle**

**2 → send right**

---

## 3. Three Pointers

### `low`

Marks where the next `0` should go.

Everything before `low` is already:

`0`

### `mid`

The current element being examined.

Everything between `low` and `mid - 1` is already:

`1`

### `high`

Marks where the next `2` should go.

Everything after `high` is already:

`2`

### Mental Picture

`[ 0s | 1s | UNKNOWN | 2s ]`

Pointers:

`      low   mid       high`

---

## 4. Initialization

Initialize:

`low = 0`

`mid = 0`

`high = n - 1`

Initially the entire array is unknown.

Example:

`[2,0,2,1,1,0]`

`low`
` ↓`

`mid`
` ↓`

`2  0  2  1  1  0`

`                    ↑`
`                   high`

---

## 5. Main Condition

Continue while:

`mid <= high`

Why?

Because the region between `mid` and `high` contains elements we have not processed yet.

Once:

`mid > high`

there are no unknown elements left.

The array is sorted.

---

## 6. Case 1 - `nums[mid] == 0`

If the current element is `0`, it belongs on the left.

Swap:

`nums[mid]` with `nums[low]`

Then:

`low += 1`

`mid += 1`

### Why increase both?

The `0` has been placed correctly.

The element moved into `mid` came from the already processed left region, so after the swap it is safe to move forward.

### Memory Rule

**See 0 → swap with low → move low and mid**

---

## 7. Case 2 - `nums[mid] == 1`

A `1` belongs in the middle.

It is already in the correct region.

No swap is required.

Simply:

`mid += 1`

### Memory Rule

**See 1 → do nothing → move mid**

---

## 8. Case 3 - `nums[mid] == 2`

A `2` belongs on the right.

Swap:

`nums[mid]` with `nums[high]`

Then:

`high -= 1`

### Important

Do **not** increment `mid`.

Why?

Because the element that came from `high` is unknown.

We have not examined it yet.

So `mid` must stay where it is and inspect the new value.

### Memory Rule

**See 2 → swap with high → move high only**

This is the most important detail of the algorithm.

---

## 16. Pointer Invariant

At every moment, maintain:

### Before `low`

Everything is:

`0`

### Between `low` and `mid`

Everything is:

`1`

### Between `mid` and `high`

Everything is:

`UNKNOWN`

### After `high`

Everything is:

`2`

Visual:

`[ 0s | 1s | UNKNOWN | 2s ]`

This invariant is the key to understanding the algorithm.

---

## 17. Why `mid` Does Not Move When We See `2`

This is the most common mistake.

Suppose:

`[0,1,2,0,2]`

and:

`mid` points to `2`.

We swap it with the element at `high`.

The new element arriving at `mid` could be:

`0`

or:

`1`

or:

`2`

We do not know.

Therefore we must examine it.

So:

`high -= 1`

but:

`mid` stays unchanged.

### Memory Rule

**2 gets sent right, but the replacement at mid is still unknown.**

---

## 18. Why `mid` Moves When We See `0`

When `nums[mid] == 0`, we swap it with `nums[low]`.

The element coming from `low` belongs to the already processed region.

Therefore the new element at `mid` is already known to be safe.

So:

`low += 1`

and:

`mid += 1`

---

## 19. Why `1` Only Moves `mid`

A `1` belongs exactly in the middle region.

It requires no swap.

Therefore:

`mid += 1`

That's it.

---

## 20. Code

class Solution(object):
    def sortColors(self, nums):
        """
        :type nums: List[int]
        :rtype: None Do not return anything, modify nums in-place instead.
        """

        low = 0
        n = len(nums)
        high = n - 1
        mid = 0

        while mid <= high:
            if nums[mid] == 0:
                nums[mid], nums[low] = nums[low], nums[mid]
                low += 1
                mid += 1

            elif nums[mid] == 2:
                nums[mid], nums[high] = nums[high], nums[mid]
                high -= 1

            else:
                mid += 1

        return nums

## 21. Final Summary + Status

### Problem

Sort an array containing only:

`0, 1, 2`

in-place.

### Approach

**Dutch National Flag Algorithm**

Use:

`low`

`mid`

`high`

Maintain:

`[ 0s | 1s | UNKNOWN | 2s ]`

### Decision Table

| Current Value | Action | Pointer Movement |
|---|---|---|
| `0` | Swap with `low` | `low++`, `mid++` |
| `1` | Already correct | `mid++` |
| `2` | Swap with `high` | `high--` |

### Most Important Rule

**When seeing `2`, do NOT increment `mid`.**

The incoming value is still unknown.

### Complexity

`Time = O(n)`

Each element is processed a constant number of times.

`Space = O(1)`

Only three pointers are used.

### Pattern

**Three Pointers + Partitioning**

### Master Mental Model

**0 → Left**

**1 → Middle**

**2 → Right**

### One-Line Rule

**0: low + mid | 1: mid | 2: high**

### Interview Explanation

"I use the Dutch National Flag approach with three pointers: low, mid, and high. The region before low contains only zeros, the region between low and mid contains only ones, and the region after high contains only twos. When mid points to zero, I swap it with low and move both. When it points to one, I simply move mid. When it points to two, I swap it with high and move only high because the new value at mid has not been processed yet. This gives O(n) time and O(1) extra space."
