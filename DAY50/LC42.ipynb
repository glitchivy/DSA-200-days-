# LC 42 - Trapping Rain Water

## 1. Problem

Given an array `height` where each value represents the height of a vertical bar, calculate how much rainwater can be trapped after raining.

### Example

Input:

`height = [0,1,0,2,1,0,1,3,2,1,2,1]`

Output:

`6`

The trapped water is determined by the tallest boundary on the left and right of each position.

---

## 2. Core Idea

Use:

**Two Pointers + Left Maximum + Right Maximum**

Maintain:

- `left` → left pointer
- `right` → right pointer
- `max_left` → maximum height seen from the left
- `max_right` → maximum height seen from the right
- `water` → total trapped water

The key decision is:

Compare:

`max_left`

and:

`max_right`

### Memory Rule

**Process the side with the smaller known maximum.**

If:

`max_left <= max_right`

process the left side.

Otherwise:

process the right side.

---

## 3. Why Does This Work?

Water above a position is determined by:

`min(left_max, right_max) - height`

The smaller boundary is the limiting factor.

If:

`max_left <= max_right`

then the left side is the limiting side.

Therefore we can safely calculate the water at `left`.

If:

`max_right < max_left`

then the right side is the limiting side.

Therefore we can safely calculate the water at `right`.

### Core Insight

**The smaller maximum determines the water level.**

---

## 4. Pointer Initialization

Initialize:

`left = 0`

`right = len(height) - 1`

Both pointers start at the ends of the array.

Also initialize:

`max_left = 0`

`max_right = 0`

`water = 0`

---

## 5. Main Decision

While:

`left <= right`

compare:

`max_left`

and:

`max_right`

### Case 1

If:

`max_left > max_right`

process the right side.

Update:

`max_right`

using:

`height[right]`

Then calculate:

`max_right - height[right]`

Move:

`right -= 1`

### Case 2

Otherwise:

`max_left <= max_right`

process the left side.

Update:

`max_left`

using:

`height[left]`

Then calculate:

`max_left - height[left]`

Move:

`left += 1`

---

## 6. Why Process the Smaller Maximum?

Suppose:

`max_left = 3`

`max_right = 5`

The water level is limited by:

`3`

because:

`min(3,5) = 3`

Therefore the left side can be safely processed.

We do not need the exact right boundary for that position because we already know the right boundary is at least `5`, which is higher than the left maximum.

### Memory Rule

**Smaller maximum = guaranteed limiting boundary**

---

## 7. Dry Run

Use:

`height = [0,1,0,2,1,0,1,3,2,1,2,1]`

Initial:

`left = 0`

`right = 11`

`max_left = 0`

`max_right = 0`

`water = 0`

---

## 8. Dry Run - Step 1

Compare:

`max_left = 0`

`max_right = 0`

Since:

`max_left <= max_right`

process left.

Current:

`height[left] = 0`

Update:

`max_left = max(0,0) = 0`

Water:

`0 - 0 = 0`

Move:

`left = 1`

---

## 9. Dry Run - Step 2

Current:

`height[left] = 1`

`max_left = 0`

`max_right = 0`

Process left because:

`max_left <= max_right`

Update:

`max_left = 1`

Water:

`1 - 1 = 0`

Move:

`left = 2`

---

## 10. Dry Run - Step 3

Current:

`height[left] = 0`

Now:

`max_left = 1`

`max_right = 0`

Since:

`max_left > max_right`

process right.

Current:

`height[right] = 1`

Update:

`max_right = 1`

Water:

`1 - 1 = 0`

Move:

`right = 10`

---

## 11. Continue the Same Pattern

At every step:

1. Compare `max_left` and `max_right`.
2. Process the side with the smaller maximum.
3. Update that maximum.
4. Calculate trapped water.
5. Move that pointer inward.

The water accumulated over the full traversal becomes:

`6`

Therefore:

`return 6`

---

## 12. Important Water Formula

For a processed position:

`water += max_height - current_height`

For the left side:

`water += max_left - height[left]`

For the right side:

`water += max_right - height[right]`

The current maximum is always updated first.

This prevents negative water.

---

## 13. Why Update Maximum First?

Suppose:

`max_left = 2`

and:

`height[left] = 5`

The new bar is taller than the previous maximum.

Update:

`max_left = 5`

Then:

`max_left - height[left]`

becomes:

`5 - 5 = 0`

Correct.

If we calculated water before updating the maximum:

`2 - 5 = -3`

which is impossible.

### Memory Rule

**Update max first → calculate water second.**

---

## 14. Code

class Solution(object):
    def trap(self, height):
        """
        :type height: List[int]
        :rtype: int
        """
        # initialize
        left = 0
        right = len(height) - 1
        max_right = 0
        max_left = 0
        water = 0

        while left <= right:
            if max_left > max_right:
                max_right = max(max_right, height[right])
                water += max_right - height[right]
                right -= 1
            else:
                max_left = max(max_left, height[left])
                water += max_left - height[left]
                left += 1

        return water

---

## 15. Code Breakdown

### `left = 0`

Start from the leftmost bar.

### `right = len(height) - 1`

Start from the rightmost bar.

### `max_left = 0`

Stores the tallest bar encountered from the left.

### `max_right = 0`

Stores the tallest bar encountered from the right.

### `water = 0`

Stores the total trapped water.

### `while left <= right`

Continue processing until the two pointers cross.

### `if max_left > max_right`

The right side currently has the smaller maximum.

Process the right side.

### `max_right = max(max_right, height[right])`

Update the right maximum.

### `water += max_right - height[right]`

Calculate water trapped at the current right position.

### `right -= 1`

Move the right pointer inward.

### `else`

This means:

`max_left <= max_right`

So process the left side.

### `max_left = max(max_left, height[left])`

Update the left maximum.

### `water += max_left - height[left]`

Calculate water trapped at the current left position.

### `left += 1`

Move the left pointer inward.

### `return water`

Return the total trapped water.

---

## 16. Why This Is O(n)

There are two pointers:

`left`

and:

`right`

Every iteration moves exactly one pointer.

`left` only moves right.

`right` only moves left.

Neither pointer ever moves backward.

Therefore, across the entire algorithm, there are at most `n` pointer movements.

So:

`Time = O(n)`

There is no nested loop.

---

## 17. Complexity

### Time Complexity

`O(n)`

Every element is processed at most once.

### Space Complexity

`O(1)`

Only a constant number of variables are used.

No extra arrays are required.

`max_left`

## 19. Common Mistakes

### Mistake 1: Comparing `height[left]` and `height[right]`

The important comparison is:

`max_left`

vs.

`max_right`

Not simply the current heights.

### Mistake 2: Calculating Water Before Updating Maximum

Always update:

`max_left`

or:

`max_right`

first.

Then calculate the trapped water.

### Mistake 3: Moving Both Pointers

Only process one side at a time.

The side with the smaller maximum gets processed.

### Mistake 4: Moving the Wrong Pointer

If:

`max_left > max_right`

move:

`right`

Otherwise move:

`left`

### Mistake 5: Using Extra Arrays

A prefix-max and suffix-max solution works, but uses:

`O(n)`

extra space.

The two-pointer solution reduces this to:

`O(1)`

---

## 20. Comparison With Other Approaches

### Brute Force

For every position, find the maximum on the left and right.

Time:

`O(n²)`

Space:

`O(1)`

Too slow for large inputs.

### Prefix/Suffix Maximum Arrays

Precompute:

`left_max`

and:

`right_max`

Time:

`O(n)`

Space:

`O(n)`

Works, but uses extra memory.

### Two Pointers

Maintain only:

`max_left`

and:

`max_right`

Time:

`O(n)`

Space:

`O(1)`

This is the optimal approach.
