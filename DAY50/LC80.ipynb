# LC 80 - Remove Duplicates from Sorted Array II

## 1. Problem

Given an integer array `nums` sorted in non-decreasing order, remove duplicates **in-place** so that each unique element appears **at most twice**.

Return the number of elements remaining after removing the extra duplicates.

The relative order of the remaining elements must be maintained.

### Example

Input:

`nums = [1,1,1,2,2,3]`

Output:

`5`

Modified array:

`[1,1,2,2,3,...]`

Only the first `5` positions matter.

---

## 2. Core Idea

Use the **Two Pointer / Slow-Fast Pointer** approach.

Because the array is already sorted, duplicates are next to each other.

We want:

- First occurrence → keep
- Second occurrence → keep
- Third and later occurrences → remove

The key observation is:

To determine whether the current number can be kept, compare it with the element **2 positions behind the slow pointer**.

### Memory Rule

**Allow at most 2 copies → compare with `slow - 2`**

---

## 3. Pointer Roles

### `slow`

Represents the position where the next valid element should be written.

Everything before `slow` is already valid.

### `i`

Scans through the original array looking for elements that can be kept.

### Initial State

`slow = 2`

Why?

Because the first two elements can always stay.

Example:

`[1,1,1,2,2,3]`

The first two:

`[1,1]`

are automatically valid.

---

## 4. Main Logic

Start:

`slow = 2`

Then scan:

`i = 2`

through:

`n - 1`

For each element:

`nums[i]`

compare it with:

`nums[slow - 2]`

### If Different

If:

`nums[i] != nums[slow - 2]`

the current value can safely be included.

Write:

`nums[slow] = nums[i]`

Then:

`slow += 1`

### If Same

If:

`nums[i] == nums[slow - 2]`

the value would become the third occurrence.

So skip it.

### Memory Rule

**Different from two positions back → keep**

**Same as two positions back → skip**

---

## 5. Why Compare With `slow - 2`?

This is the most important insight.

We are allowed to keep **two copies** of every value.

Suppose the valid portion currently ends with:

`[1,1]`

Now another:

`1`

comes in.

We compare it with the element two positions behind:

`nums[slow - 2]`

which is:

`1`

They are equal.

That means adding the current `1` would create:

`[1,1,1]`

which is not allowed.

So we skip it.

---

## 6. Why Does This Work?

Because the array is sorted.

All duplicates appear consecutively.

Suppose we have:

`[1,1,1,2,2,3]`

For the third `1`, comparing against the element two valid positions behind tells us whether we already have two `1`s.

For `2`:

The first `2` can be kept.

The second `2` can be kept.

A third `2` would match the element two positions behind and therefore be rejected.

This automatically limits every value to at most two occurrences.

The important point is that we never need to count how many times each number appears.

The sorted property plus:

`nums[i] != nums[slow - 2]`

## 14. Code

class Solution(object):
    def removeDuplicates(self, nums):
        """
        :type nums: List[int]
        :rtype: int
        """
        n = len(nums)
        slow = 2

        for i in range(2, n):
            if nums[i] != nums[slow - 2]:
                nums[slow] = nums[i]
                slow += 1

        return slow

---

## 15. Code Breakdown

### Step 1

`n = len(nums)`

Store the array length.

### Step 2

`slow = 2`

The first two elements are always allowed.

### Step 3

`for i in range(2, n):`

Start scanning from the third element.

### Step 4

`nums[i] != nums[slow - 2]`

Check whether keeping the current value would create a third consecutive copy.

### Step 5

If different:

`nums[slow] = nums[i]`

Write the valid value into the next available position.

### Step 6

`slow += 1`

Increase the length of the valid portion.

### Step 7

If the values are equal:

Do nothing.

The current value is skipped.

### Step 8

`return slow`

`slow` represents the number of valid elements.

Only:

`nums[0:slow]`

matters after the function finishes.

---

## 16. Important Code Observation

The algorithm does **not actually delete elements**.

It overwrites the beginning of the array with the valid values.

Example:

Original:

`[1,1,1,2,2,3]`

After processing:

`[1,1,2,2,3,3]`

The final `3` is irrelevant.

Because the returned length is:

`5`

we only care about:

`nums[:5]`

which is:

`[1,1,2,2,3]`

### Memory Rule

**In-place removal = overwrite valid values + return valid length**

## 17. Why This Is O(n)

There is:

`for i in range(2,n)`

The loop runs approximately `n` times.

Inside the loop:

- One comparison
- Possibly one assignment
- One increment

All are:

`O(1)`

There is no nested loop.

Therefore:

`O(n)`

time.

## 19. Pattern Recognition

### Pattern

**Two Pointers + In-Place Overwrite**

### Trigger

When you see:

- Sorted array
- Remove duplicates
- Keep at most `k` occurrences
- Modify array in-place
- Return the valid length

Think:

**Slow/Fast Pointer + Compare With `slow - k`**

For this problem:

`k = 2`

Therefore:

`nums[i] != nums[slow - 2]`

### General Pattern

For "keep at most `k` duplicates":

`if nums[i] != nums[slow-k]:`

keep the element.
