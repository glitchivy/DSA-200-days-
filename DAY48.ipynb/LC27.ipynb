# LC 27 - Remove Element

## 1. Problem

Given an integer array `nums` and an integer `val`, remove all occurrences of `val` **in-place**.

Return the number of elements that are **not equal to `val`**.

The order of the remaining elements can be preserved.

### Example

Input:

`nums = [3,2,2,3]`

`val = 3`

Output:

`2`

After modification:

`nums = [2,2,_,_]`

Only the first `2` positions matter because the returned length is `2`.

---

## 2. Core Idea

Use the **Write Pointer / Slow-Fast Pointer** approach.

- `i` scans every element.
- `slow` tells us where the next valid element should be placed.
- If `nums[i] != val`, keep it.
- If `nums[i] == val`, skip it.

At the end:

`slow`

is the number of elements that remain.

### Mental Model

`i → scans everything`

`slow → next position for a valid element`

---

## 3. Pointer Initialization

Initialize:

`slow = 0`

The `slow` pointer represents:

**"How many valid elements have I found so far?"**

The `i` pointer scans from:

`0 → n-1`

Example:

`nums = [3,2,2,3]`

`val = 3`

Initially:

`slow = 0`

---

## 4. Main Decision Logic

For every element:

`if nums[i] != val`

The element should remain.

Place it at:

`nums[slow] = nums[i]`

Then:

`slow += 1`

If:

`nums[i] == val`

Skip it.

Do not increment `slow`.

### Memory Rule

**Valid element → Write at slow → slow++**

**Target value → Skip**

**End → slow is the answer**

---

## 5. Why Does `slow` Work?

`slow` always represents the next position where a valid element should be placed.

Suppose:

`nums = [3,2,2,3]`

`val = 3`

The first element is `3`, so skip it.

The next element is `2`.

Write it at index `0`.

Now:

`slow = 1`

The next `2` goes to index `1`.

Now:

`slow = 2`

So the first `slow` positions contain exactly the elements we want to keep.

---

## 6. Why Don't We Delete Elements?

The problem asks for an **in-place** modification.

We don't actually need to physically delete elements.

Instead, we overwrite the beginning of the array with all valid elements.

Only the first `slow` positions matter after the operation.

Everything after `slow` can be ignored.

## 13. Why Return `slow`?

Every time we find a valid element, we do:

`slow += 1`

Therefore `slow` counts exactly how many elements are different from `val`.

Example:

`nums = [3,2,2,3]`

`val = 3`

Valid elements:

`2`

`2`

Number of valid elements:

`2`

Therefore:

`slow = 2`

So:

`return slow`

---

## 14. Brute Force vs Write Pointer

### Brute Force

Repeatedly remove matching elements from the array.

This can require shifting many elements.

Potential time complexity:

`O(n²)`

### Write Pointer

Scan once and overwrite invalid positions.

Time complexity:

`O(n)`

Space complexity:

`O(1)`

The write-pointer approach is the correct in-place pattern here.

---

## 15. Code

class Solution(object):
    def removeElement(self, nums, val):
        """
        :type nums: List[int]
        :type val: int
        :rtype: int
        """
        slow = 0

        for i in range(len(nums)):
            if nums[i] != val:
                nums[slow] = nums[i]
                slow += 1

        return slow

---

## 16. Code Breakdown

### Step 1

`slow = 0`

Tracks the next position for a valid element.

### Step 2

`for i in range(len(nums))`

Scans every element.

### Step 3

`if nums[i] != val`

Checks whether the current element should remain.

### Step 4

`nums[slow] = nums[i]`

Places the valid element at the next available position.

### Step 5

`slow += 1`

Increases the count of valid elements.

### Step 6

`return slow`

Returns the number of elements that are not equal to `val`.

## 18. Common Mistakes

### Mistake 1: Actually Deleting Elements

Using operations such as removing elements while iterating can cause shifting and make the solution inefficient.

### Mistake 2: Incrementing `slow` for `val`

If:

`nums[i] == val`

do not increment `slow`.

That element is not part of the final valid portion.

### Mistake 3: Returning `len(nums)`

The array length does not change.

The answer is the number of valid elements:

`slow`

### Mistake 4: Caring About Elements After `slow`

Only the first `slow` elements matter.

Anything after that is irrelevant.

### Mistake 5: Creating a New Array

That would violate the intended in-place approach.
