# LC 26 - Remove Duplicates from Sorted Array

## 1. Problem

Given a sorted integer array `nums`, remove the duplicates **in-place** so that each unique element appears only once.

Return the number of unique elements.

The relative order of the unique elements must be maintained.

### Example

Input:

`nums = [1,1,2]`

Output:

`2`

Modified array:

`[1,2,_]`

Only the first `2` positions matter because the returned length is `2`.

---

## 2. Core Idea

Use the **Two Pointer / Write Pointer** approach.

Because the array is sorted, all duplicates are next to each other.

We can use:

- `i` to scan the array.
- `slow` to track the next position where a new unique element should be written.

### Key Insight

Compare the current element with the last unique element:

`nums[i] != nums[slow - 1]`

If they are different, we found a new unique value.

---

## 3. Pointer Initialization

Initialize:

`slow = 1`

Why `1`?

The first element is automatically unique.

For example:

`nums = [1,1,2,2,3]`

The first unique element is already at:

`index 0`

So the next unique element should be written at:

`index 1`

Therefore:

`slow = 1`

Also:

`n = len(nums)`

---

## 4. Main Decision Logic

For every element:

`if nums[i] != nums[slow - 1]`

Then the current element is different from the last unique element.

Therefore:

`nums[slow] = nums[i]`

Then:

`slow += 1`

If:

`nums[i] == nums[slow - 1]`

It is a duplicate.

Skip it.

### Memory Rule

**Different from last unique → Keep**

**Same as last unique → Skip**

**slow → Number of unique elements**

---

## 5. Why Does Sorting Matter?

The array is sorted.

Therefore duplicates are adjacent.

Example:

`[1,1,1,2,2,3,3,3]`

All `1`s are together.

All `2`s are together.

All `3`s are together.

So we only need to compare the current element with the most recently stored unique element.

Without sorting, duplicates could appear anywhere, and this simple comparison would not work.

---

## 6. Why Compare With `nums[slow - 1]`?

`slow - 1` points to the **last unique element we have stored**.

Example:

`nums = [1,1,2,2,3]`

After processing the first two values:

`slow = 1`

The last unique value is:

`nums[slow - 1]`

`nums[0] = 1`

Now when we see `2`:

`2 != 1`

So `2` is a new unique value.

`slow` counts how many unique elements have been stored.

For:

`[1,1,2,2,3]`

`1`

`2`

`3`

Therefore:

`slow = 3`

So:

`return slow`

The first `slow` positions contain the required answer.

## 15. Code

class Solution(object):
    def removeDuplicates(self, nums):
        """
        :type nums: List[int]
        :rtype: int
        """
        slow = 1
        n = len(nums)

        if n == 0:
            return 0

        for i in range(len(nums)):
            if nums[i] != nums[slow - 1]:
                nums[slow] = nums[i]
                slow += 1

        return slow

## 18. Common Mistakes

### Mistake 1: Using a Set

A set can identify unique values, but it does not directly solve the required in-place modification pattern.

The sorted property gives us a simpler O(1)-space solution.

### Mistake 2: Comparing With `nums[i - 1]`

This can work conceptually because the array is sorted, but after overwriting elements, using the last stored unique value through:

`nums[slow - 1]`

is the safer pattern for the write-pointer approach.

### Mistake 3: Starting `slow = 0`

The first element is already unique.

Therefore the next write position should be:

`1`

### Mistake 4: Incrementing `slow` for Duplicates

Only increment `slow` when a new unique element is found.

### Mistake 5: Returning `len(nums)`

The array length does not change.

Return:

`slow`

because `slow` represents the number of unique elements.
