# LC 392 - Is Subsequence

## 1. Problem

Given two strings `s` and `t`, determine whether `s` is a subsequence of `t`.

A string is a subsequence of another string if all characters of the first string appear in the same relative order in the second string, but they do not need to be adjacent.

### Example 1

Input:

`s = "abc"`

`t = "ahbgdc"`

Output:

`True`

Because we can find:

`a → h → b → g → d → c`

and the characters `a`, `b`, `c` appear in the correct order.

### Example 2

Input:

`s = "axc"`

`t = "ahbgdc"`

Output:

`False`

Because `x` does not appear in `t`.

---

## 2. Core Idea

Use the **Two Pointer** approach.

We use:

- `i` → points to the current character we need from `s`
- `j` → scans through `t`

The goal is to match every character of `s` in order inside `t`.

### Memory Rule

**`j` scans `t` → if characters match, move `i` → always move `j`**

---

## 3. Pointer Initialization

Initialize:

`i = 0`

`j = 0`

Here:

`i` points to the first character of `s`.

`j` points to the first character of `t`.

Example:

`s = "abc"`

`t = "ahbgdc"`

Initially:

`i → a`

`j → a`

---

## 4. Main Decision Logic

The loop runs while both strings still have characters to process:

`while i < len(s) and j < len(t)`

Compare:

`s[i]`

with:

`t[j]`

### Case 1: Characters Match

If:

`s[i] == t[j]`

we successfully found the current required character of `s`.

So:

`i += 1`

Then we continue searching for the next character of `s`.

### Case 2: Characters Do Not Match

Do nothing to `i`.

We simply continue searching through `t`.

### Important

`j` moves forward **every time**, whether there is a match or not.

So:

`j += 1`

### Memory Rule

**Match → move both**

**No match → move only `j`**

---

## 5. Why Does `i` Only Move When There Is a Match?

Because `i` represents the next character of `s` that we need to find.

Example:

`s = "abc"`

`t = "ahbgdc"`

We first need:

`a`

When we find `a`, we move to:

`b`

We cannot move to `b` before finding `a`.

Therefore:

`i` only moves after a successful match.

---

## 6. Why Does `j` Always Move?

`j` is scanning `t`.

Every character of `t` can be considered once.

If:

`t[j]`

does not match:

`s[i]`

we simply skip it and move forward.

Example:

`s = "abc"`

`t = "ahbgdc"`

After matching `a`, we need `b`.

The next character is:

`h`

`h != b`

So we skip `h`.

Then:

`j` moves to `b`.

## 15. Code

class Solution(object):
    def isSubsequence(self, s, t):
        """
        :type s: str
        :type t: str
        :rtype: bool
        """
        i = 0
        j = 0

        while i < len(s) and j < len(t):
            if s[i] == t[j]:
                i += 1
            j += 1

        return i == len(s)

---

## 16. Code Breakdown

### Step 1

`i = 0`

Pointer for `s`.

It tells us which character of `s` we are currently searching for.

### Step 2

`j = 0`

Pointer for `t`.

It scans through `t`.

### Step 3

`while i < len(s) and j < len(t):`

Continue while:

- We still have characters left in `s`
- We still have characters left in `t`

### Step 4

`if s[i] == t[j]:`

Check whether the current characters match.

### Step 5

`i += 1`

Only move `i` when a match is found.

This means we successfully matched one more character of `s`.

### Step 6

`j += 1`

Always move `j`.

We are scanning through `t`.

### Step 7

`return i == len(s)`

If `i` reached the end of `s`, every character of `s` was successfully matched.

Therefore:

`True`

Otherwise:

`False`

---

## 17. Important Code Observation

The most important line is:

`return i == len(s)`

We do not need another counter.

`i` itself tells us how many characters of `s` have been successfully matched.

If:

`i == len(s)`

then the entire string `s` was found.

### Example

`s = "abc"`

After matching:

`a → i = 1`

`b → i = 2`

`c → i = 3`

Since:

`len(s) = 3`

we know the entire string was matched.

---

## 18. Common Mistakes

### Mistake 1: Moving `i` When There Is No Match

Wrong idea:

"No match → move both pointers."

This can skip a character that we still need to find.

Correct:

**No match → move only `j`.**

### Mistake 2: Requiring Adjacent Characters

A subsequence does not require adjacency.

For:

`s = "abc"`

`t = "ahbgdc"`

The characters are separated by other characters.

That is completely valid.

### Mistake 3: Changing the Order

The order must remain the same.

For:

`s = "abc"`

`t = "acb"`

The answer is:

`False`

because `b` comes after `c` in `t`.

### Mistake 4: Returning `True` When One Character Matches

Finding one character is not enough.

Every character in `s` must be matched.

### Mistake 5: Comparing From Both Ends

Unlike some other two-pointer problems, this problem does not need left and right pointers.

Both pointers move forward.

## 20. Complexity + Pattern Recognition

### Time Complexity

`O(n)`

More precisely:

`O(len(t))`

because `j` moves through `t` at most once.

### Space Complexity

`O(1)`

Only two pointers are used.

---

### Pattern

**Two Pointers + Forward Scanning**

### Trigger

When you see:

- Determine whether one sequence appears inside another
- Preserve relative order
- Characters/elements can be skipped
- No need for contiguous matching

Think:

**Subsequence → Two Forward Pointers**

### General Pattern

`i → current required element`

`j → scanner`

If:

`match`

then:

`i++`

Always:

`j++`
