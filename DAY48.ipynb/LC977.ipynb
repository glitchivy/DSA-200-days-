# LC 977 - Squares of a Sorted Array

## 1. Problem

Given an integer array `nums` sorted in non-decreasing order, return an array containing the squares of every number, also sorted in non-decreasing order.

### Example

Input:

`nums = [-4,-1,0,3,10]`

Output:

`[0,1,9,16,100]`

---

## 2. Core Idea

Use the **Two Pointer** approach.

The array is sorted, but after squaring it, the order can change because negative numbers become positive.

The largest square must come from either:

- The leftmost negative number
- The rightmost positive number

So compare their absolute values.

---

## 3. Pointer Initialization

Initialize:

`left = 0`

`right = n - 1`

Also create:

`pos = n - 1`

`res = [0] * n`

Why?

We are filling the result array from **right to left** because we find the largest square first.

---

## 4. Main Decision Logic

Compare:

`abs(nums[left])`

with:

`abs(nums[right])`

### Case 1

If:

`abs(nums[left]) > abs(nums[right])`

The left value has the larger square.

Put:

`nums[left]²`

at:

`res[pos]`

Then:

`left += 1`

### Case 2

Otherwise:

`abs(nums[right]) >= abs(nums[left])`

The right value has the larger square.

Put:

`nums[right]²`

at:

`res[pos]`

Then:

`right -= 1`

After either case:

`pos -= 1`

---

## 5. Why Compare Absolute Values?

Example:

`nums = [-7,2]`

The array is sorted:

`-7 < 2`

But after squaring:

`(-7)² = 49`

`2² = 4`

So the numerical value itself does not tell us which square is larger.

We compare:

`abs(-7) = 7`

`abs(2) = 2`

Therefore:

`7 > 2`

So `49` is the larger square.

---

## 6. Why Fill From Right to Left?

We are looking for the **largest square first**.

The largest square will always come from one of the two ends.

So:

`pos = n - 1`

stores the largest square.

Then:

`pos -= 1`

moves toward the beginning.

Therefore the result gets constructed in sorted order.

---

## 7. Why Only Compare the Two Ends?

The array is already sorted.

For example:

`[-10,-5,-2,1,3,8]`

The largest absolute values can only be found at the ends.

- Left side contains the most negative values.
- Right side contains the largest positive values.

So we only need:

`left`

and:

`right`

This eliminates the need to sort the squared values afterward.

## 15. Code

class Solution(object):
    def sortedSquares(self, nums):
        """
        :type nums: List[int]
        :rtype: List[int]
        """
        n = len(nums)
        right = n - 1
        left = 0
        pos = n - 1
        res = [0] * n

        while left <= right:
            if abs(nums[left]) > abs(nums[right]):
                res[pos] = nums[left] * nums[left]
                left += 1
            else:
                res[pos] = nums[right] * nums[right]
                right -= 1
            pos -= 1

        return res

## 16. Code Breakdown

### Initialization

`n = len(nums)`

Gets the size of the array.

`left = 0`

Starts from the smallest element.

`right = n - 1`

Starts from the largest element.

`pos = n - 1`

Points to the last position of the result.

`res = [0] * n`

Creates the result array.

### Main Loop

`while left <= right:`

Continue until all elements have been processed.

### Compare

`abs(nums[left]) > abs(nums[right])`

Checks which end produces the larger square.

### Left Wins

`res[pos] = nums[left] * nums[left]`

Store the left square.

Then:

`left += 1`

### Right Wins

`res[pos] = nums[right] * nums[right]`

Store the right square.

Then:

`right -= 1`

### Move Result Position

`pos -= 1`

Move to the next position from right to left.

---

## 17. Important Code Observation

The important trick is that you **do not square everything first and then sort**.

That would take:

`O(n log n)`

because of sorting.

Instead, you exploit the sorted input and use two pointers.

This gives:

`O(n)`

time.

Also notice that the result array is filled from the back because the largest squares are discovered first.

---

## 18. Common Mistakes

### Mistake 1: Squaring First and Sorting

Example:

`[-4,-1,0,3,10]`

Square:

`[16,1,0,9,100]`

Then sort.

This works, but costs:

`O(n log n)`

The two-pointer solution is better.

### Mistake 2: Comparing Actual Values

Comparing:

`nums[left] > nums[right]`

is not enough.

You need:

`abs(nums[left]) > abs(nums[right])`

because negative values can produce larger squares.

### Mistake 3: Filling From the Beginning

The largest square is found first.

Therefore fill:

`right → left`

not:

`left → right`

### Mistake 4: Forgetting `pos -= 1`

Every selected square needs to go into the next position from the right.

### Mistake 5: Using `left < right`

You must process the final remaining element.

Therefore:

`while left <= right`

---

## 19. Complexity

### Time Complexity

`O(n)`

Each element is processed exactly once.

The `left` pointer only moves forward.

The `right` pointer only moves backward.

### Space Complexity

`O(n)`

The result array contains `n` elements.

If the output array is excluded from auxiliary-space analysis, the extra working space is:

`O(1)`
