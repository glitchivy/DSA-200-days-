# LC 283 - Move Zeroes

## 1. Problem

Given an integer array `nums`, move all `0`s to the end of the array while maintaining the relative order of the non-zero elements.

The operation must be performed **in-place**.

### Example

Input:

`nums = [0,1,0,3,12]`

Output:

`[1,3,12,0,0]`

The relative order of the non-zero elements remains:

`1 → 3 → 12`

---

## 2. Core Idea

Use the **Two Pointer / Slow-Fast Pointer** approach.

- `i` scans every element.
- `slow` tells us where the next non-zero element should go.

Whenever we find a non-zero value:

`nums[slow] = nums[i]`

Then:

`slow += 1`

After all non-zero elements have been placed, every position from `slow` onward should contain `0`.

---

## 3. Pointer Initialization

Initialize:

`slow = 0`

The `slow` pointer represents:

**"Where should the next non-zero element be placed?"**

The `i` pointer scans the entire array.

Example:

`nums = [0,1,0,3,12]`

Initially:

`slow = 0`

`i = 0`

---

## 4. Main Decision Logic

For every element:

`if nums[i] != 0`

Then the current value is useful and should be placed at the `slow` position.

Perform:

`nums[slow] = nums[i]`

Then:

`slow += 1`

If:

`nums[i] == 0`

Do nothing.

The scanning pointer continues moving.

### Memory Rule

**Non-zero → Place at slow**

**Zero → Skip**

**After scan → Fill remaining positions with zero**

---

## 5. Why Does `slow` Work?

`slow` always points to the next position where a non-zero number belongs.

Suppose:

`nums = [0,1,0,3,12]`

At the beginning:

`slow = 0`

The first non-zero number is `1`.

So place:

`1` at index `0`.

Now:

`slow = 1`

The next non-zero number is `3`.

Place it at index `1`.

Now:

`slow = 2`

So `slow` effectively keeps track of the boundary between:

**processed non-zero elements**

and

**positions that still need to be filled**

---

## 6. Why Don't We Need to Move the Zeros Immediately?

We do not need to physically move every zero when we encounter it.

Instead, we only care about collecting all the non-zero values toward the front.

After that, we already know that every remaining position must be zero.

This makes the algorithm simple:

**First compact non-zero values → Then fill remaining positions with zero**

Therefore positions:

`3` through `n-1`

`nums[i] = 0`

## 15. Code

class Solution(object):
    def moveZeroes(self, nums):
        """
        :type nums: List[int]
        :rtype: None Do not return anything, modify nums in-place instead.
        """
        slow = 0

        for i in range(len(nums)):
            if nums[i] != 0:
                nums[slow] = nums[i]
                slow += 1

        for i in range(slow, len(nums)):
            nums[i] = 0

## 16. Code Breakdown

### Step 1

`slow = 0`

Tracks where the next non-zero element should be placed.

### Step 2

`for i in range(len(nums))`

The `i` pointer scans every element.

### Step 3

`if nums[i] != 0`

Only non-zero values need to be moved toward the front.

### Step 4

`nums[slow] = nums[i]`

Place the current non-zero value at the next available position.

### Step 5

`slow += 1`

Move the insertion position forward.

### Step 6

Second loop:

`for i in range(slow, len(nums))`

Every position after the last non-zero element must contain zero.

### Step 7

`nums[i] = 0`

Fill all remaining positions with zero.

---

## 17. Important Code Observation

Your solution uses a very useful pattern:

**Compaction / Write Pointer**

Instead of swapping every zero immediately, you first move all useful values toward the front.

The important invariant is:

**Before `slow`, all positions contain the non-zero elements in their original order.**

At the end:

**From `slow` onward, every position becomes zero.**

This is why you can solve the problem without an additional array.

---

## 18. Common Mistakes

### Mistake 1: Creating Another Array

For example:

`result = []`

This uses extra space.

The problem asks for an in-place solution.

### Mistake 2: Losing Relative Order

The non-zero elements must remain in the same order.

For:

`[0,1,0,3,12]`

the answer must begin:

`[1,3,12,...]`

not some rearrangement such as:

`[12,3,1,...]`

### Mistake 3: Moving `slow` When You See Zero

Do not increment `slow` for zero.

`slow` represents the next position for a non-zero value.

### Mistake 4: Forgetting the Second Loop

After moving the non-zero values, the leftover positions are not automatically zero.

You must explicitly fill them.

### Mistake 5: Returning the Array

The LeetCode problem asks for an in-place modification.

The function should modify `nums` directly and does not need to return it.
