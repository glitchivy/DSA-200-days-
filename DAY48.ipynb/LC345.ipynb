# LC 345 - Reverse Vowels of a String

## 1. Problem

Given a string `s`, reverse only the vowels in the string.

The consonants and their positions must remain unchanged.

Vowels are:

`A, E, I, O, U, a, e, i, o, u`

### Example

Input:

`s = "hello"`

Output:

`"holle"`

Only the vowels are reversed:

`e → o`

`o → e`

---

## 2. Core Idea

Use the **Two Pointer** approach.

We place:

- `left` at the beginning
- `right` at the end

Then:

- Move `left` until it finds a vowel.
- Move `right` until it finds a vowel.
- Swap those two vowels.
- Move both pointers inward.

### Memory Rule

**Left finds vowel → Right finds vowel → Swap → Move both**

---

## 3. Why Convert the String to a List?

Python strings are immutable.

That means we cannot directly do:

`s[left] = ...`

So we convert:

`s`

into:

`list(s)`

Now the characters can be modified in-place.

At the end:

`"".join(s)`

converts the list back into a string.

---

## 4. Pointer Initialization

Initialize:

`left = 0`

`right = len(s) - 1`

Example:

`s = "hello"`

After conversion:

`['h','e','l','l','o']`

Pointers:

`left → h`

`right → o`

---

## 5. Vowel Detection

Create a collection containing all vowels:

`['A','E','I','O','U','a','e','i','o','u']`

For the left pointer:

If:

`s[left] not in value`

then it is a consonant.

Move:

`left += 1`

For the right pointer:

If:

`s[right] not in value`

then it is a consonant.

Move:

`right -= 1`

---

## 6. Main Decision Logic

There are three situations.

### Case 1: Left is not a vowel

Move `left` forward.

### Case 2: Right is not a vowel

Move `right` backward.

### Case 3: Both are vowels

Swap them.

Then:

`left += 1`

`right -= 1`

### Memory Rule

**Not vowel → Move pointer**

**Both vowels → Swap**

## 14. Code

class Solution(object):
    def reverseVowels(self, s):
        """
        :type s: str
        :rtype: str
        """
        left = 0
        right = len(s) - 1
        s = list(s)
        value = ['A','E','I','O',"U",'a','e','i','o','u']

        while left < right:
            if s[left] not in value:
                left += 1
            elif s[right] not in value:
                right -= 1
            else:
                s[right], s[left] = s[left], s[right]
                right -= 1
                left += 1

        return "".join(s)

## 16. Important Code Observation

The key part of the algorithm is that we do **not** swap arbitrary characters.

We only swap when:

`left → vowel`

and:

`right → vowel`

This guarantees that only vowels change positions.

The consonants are simply skipped.

---

## 17. Common Mistakes

### Mistake 1: Swapping Every Character

That would reverse the entire string.

The problem only wants vowels reversed.

### Mistake 2: Moving Both Pointers Every Time

You should move only the pointer that is currently pointing to a consonant.

Only move both after a successful vowel swap.

### Mistake 3: Forgetting Uppercase Vowels

You need:

`A E I O U`

as well as:

`a e i o u`

### Mistake 4: Trying to Modify a String Directly

Python strings are immutable.

Convert the string to a list first.

### Mistake 5: Forgetting to Convert Back

The required return type is a string.

Therefore:

`"".join(s)`

is needed.

### Mistake 6: Using `while left <= right`

`left < right` is enough because once the pointers meet or cross, there is nothing left to swap.

---

## 18. Complexity

### Time Complexity

`O(n)`

Each pointer moves only forward or backward through the string.

Together, they scan the string at most a constant number of times.

### Space Complexity

`O(n)`

Because the string is converted into a list of characters.

The algorithm itself uses only `O(1)` additional pointer space, but the mutable character list requires `O(n)` space.

## 19. Pattern Recognition + Interview Explanation

### Pattern

**Two Pointers + Selective Swapping**

### Trigger

When you see:

- Reverse only certain elements
- Keep other elements unchanged
- Need to process from both ends
- The selected elements can be identified by a condition

Think:

**Two Pointers**

### General Pattern

`left → search from beginning`

`right → search from end`

When both satisfy the condition:

`swap`

Then:

`left++`

`right--`

### Interview Explanation

"I use two pointers starting from both ends of the string. The left pointer searches for the next vowel from the beginning, while the right pointer searches for the next vowel from the end. When both pointers point to vowels, I swap them and move both pointers inward. Consonants are skipped. Since Python strings are immutable, I convert the string to a list first and join it back into a string at the end."
