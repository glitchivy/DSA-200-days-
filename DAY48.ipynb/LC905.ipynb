# LC 905 - Sort Array By Parity

## 1. Problem

Given an integer array `nums`, move all even integers to the beginning of the array and all odd integers to the end.

The relative order of the elements does not need to be maintained.

### Example

Input:

`nums = [3,1,2,4]`

Output:

`[2,4,3,1]`

Any arrangement where all even numbers come before all odd numbers is valid.

---

## 2. Core Idea

Use the **Two Pointer / Write Pointer** approach.

We use:

- `i` to scan the array.
- `slow` to track the position where the next even number should go.

Whenever we find an even number, we move it toward the front.

The basic idea is:

**Find even → put it at `slow` → move `slow` forward**

---

## 3. Pointer Initialization

Initialize:

`slow = 0`

The `slow` pointer represents:

**"Where should the next even number be placed?"**

The `i` pointer scans every element from left to right.

Example:

`nums = [3,1,2,4]`

Initially:

`slow = 0`

---

## 4. Main Decision Logic

For every element:

Check:

`nums[i] % 2 == 0`

If the current number is even and the current `slow` position contains an odd number:

`nums[i], nums[slow] = nums[slow], nums[i]`

Then:

`slow += 1`

After that, the code checks:

`if nums[slow] % 2 == 0`

and increments `slow`.

### Memory Rule

**Even → move toward front**

**Odd → leave it**

**slow → next position for an even number**

---

## 5. Why Does `slow` Work?

`slow` represents the boundary between the positions where even numbers should be placed and the remaining unsorted area.

Example:

`[3,1,2,4]`

Initially:

`slow = 0`

The first even number encountered is `2`.

It should move toward the beginning.

So:

`2` swaps with `3`.

Array becomes:

`[2,1,3,4]`

Now:

`slow = 1`

The next even number can be placed at index `1`.

---

## 6. Why Use Swapping?

We need all even values toward the beginning and odd values toward the end.

Instead of creating a second array, we can swap:

`even number`

with the current odd position.

Example:

`[3,1,2,4]`

When we find `2`:

`3 ↔ 2`

Result:

`[2,1,3,4]`

This keeps the operation in-place.

## 13. Important Observation About Your Code

Your solution uses:

`if nums[i] % 2 == 0 and nums[slow] % 2 != 0:`

This checks two things:

1. The current element is even.
2. The position where we want an even number currently contains an odd number.

If both are true, swap them.

Then your second condition:

`if nums[slow] % 2 == 0:`

moves `slow` forward when the current position already contains an even number.

This is an important pointer-boundary idea.

## 14. Code

class Solution(object):
    def sortArrayByParity(self, nums):
        """
        :type nums: List[int]
        :rtype: List[int]
        """
        slow = 0

        for i in range(len(nums)):
            if nums[i] % 2 == 0 and nums[slow] % 2 != 0:
                nums[i], nums[slow] = nums[slow], nums[i]
                slow += 1

            if nums[slow] % 2 == 0:
                slow += 1

        return nums

## 16. Important Code Observation

This problem is another variation of the **Two Pointer / Partition** pattern.

The array is being divided into two regions:

`[ EVEN | UNKNOWN | ODD ]`

The `slow` pointer represents the boundary of the even region.

The scanning pointer `i` searches for even values that need to be moved into that region.

This is closely related to the partition idea used in algorithms such as Quick Sort.

---

## 17. Common Mistakes

### Mistake 1: Sorting the Whole Array

Using a normal sorting algorithm is unnecessary.

We only need to partition the array into:

`EVEN | ODD`

### Mistake 2: Creating a Second Array

That works conceptually but uses `O(n)` extra space.

The problem can be solved in-place.

### Mistake 3: Forgetting the Boundary

You need a pointer that tells you where the next even number belongs.

That is `slow`.

### Mistake 4: Assuming the Output Has One Exact Arrangement

The problem does not require the even or odd elements to maintain their original relative order.

For example, both can be valid:

`[2,4,3,1]`

and:

`[4,2,1,3]`

The only requirement is:

**all evens before all odds.**

### Mistake 5: Confusing Parity

Even:

`num % 2 == 0`

Odd:

`num % 2 != 0`

## 19. Pattern Recognition + Interview Explanation

### Pattern

**Two Pointers + In-Place Partition**

### Trigger

When you see:

- Separate elements into two groups
- One condition defines the first group
- Modify in-place
- Relative order is not important
- Need a linear-time solution

Think:

**Partition / Two Pointers**

### General Pattern

`slow → boundary of desired group`

`i → scans for elements belonging to that group`

When a valid element is found:

`swap(i, slow)`

Then move the boundary.

### Interview Explanation

"I use a two-pointer partition approach. The `slow` pointer represents the next position where an even number should be placed, while `i` scans the array. Whenever `i` finds an even number and the slow position contains an odd number, I swap them and advance the boundary. This partitions the array into an even section followed by an odd section. The solution modifies the array in-place and runs in O(n) time with O(1) extra space."
