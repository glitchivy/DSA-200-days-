# LC 88 - Merge Sorted Array

## 1. Problem

You are given two sorted arrays:

- `nums1` with `m` actual elements and extra space at the end
- `nums2` with `n` elements

Merge `nums2` into `nums1` so that `nums1` becomes one sorted array.

The merge must be done **in-place**.

### Example

Input:

`nums1 = [1,2,3,0,0,0]`

`m = 3`

`nums2 = [2,5,6]`

`n = 3`

Output:

`[1,2,2,3,5,6]`

---

## 2. Core Idea

Use **Three Pointers + Merge From the End**.

We have:

- `i` → last actual element in `nums1`
- `j` → last element in `nums2`
- `k` → last available position in `nums1`

Why merge from the end?

Because `nums1` already contains its elements at the beginning.

If we started from the front, writing into `nums1` could overwrite elements that we still need.

Starting from the end avoids this problem.

### Memory Rule

**Compare the largest remaining elements → put the larger one at `k` → move backward**

---

## 3. Pointer Initialization

Initialize:

`i = m - 1`

`j = n - 1`

`k = m + n - 1`

For:

`nums1 = [1,2,3,0,0,0]`

`nums2 = [2,5,6]`

We get:

`i = 2`

`j = 2`

`k = 5`

Visual:

`nums1: [1, 2, 3, 0, 0, 0]`
`          i        k`

`nums2: [2, 5, 6]`
`             j`

---

## 4. Main Decision Logic

While both arrays still have elements:

`while i >= 0 and j >= 0`

Compare:

`nums1[i]`

and:

`nums2[j]`

### Case 1: nums1 is larger

If:

`nums1[i] > nums2[j]`

Place:

`nums1[i]`

at:

`nums1[k]`

Then:

`i -= 1`

`k -= 1`

### Case 2: nums2 is larger or equal

Otherwise:

`nums2[j]`

goes into:

`nums1[k]`

Then:

`j -= 1`

`k -= 1`

### Memory Rule

**Larger → put at the back**

---

## 5. Why Merge From the End?

Suppose:

`nums1 = [1,3,5,0,0,0]`

`nums2 = [2,4,6]`

The largest element is:

`6`

It belongs at the final position.

So we can safely place it at:

`nums1[5]`

Then the next largest element goes at:

`nums1[4]`

And so on.

This means we never overwrite an unprocessed element in the useful portion of `nums1`.

---

## 6. Why Does `k = m + n - 1`?

The final merged array has:

`m + n`

elements.

Array indices start from `0`.

Therefore the final index is:

`m + n - 1`

Example:

`m = 3`

`n = 3`

Total:

`6`

Last index:

`5`

Therefore:

`k = 5`

## 13. Why Is There a Second While Loop?

The second loop is:

`while j >= 0`

This handles the case where some elements of `nums2` are still remaining after all relevant elements of `nums1` have been processed.

Example:

`nums1 = [4,5,6,0,0,0]`

`nums2 = [1,2,3]`

Eventually:

`i = -1`

but:

`j >= 0`

So we must copy the remaining `nums2` elements into `nums1`.

### Important

We do **not** need a second loop for `nums1`.

Why?

Because if `nums2` is exhausted first, the remaining elements of `nums1` are already in their correct positions.

## 15. Code

class Solution(object):
    def merge(self, nums1, m, nums2, n):
        i = m - 1
        j = n - 1
        k = m + n - 1

        while i >= 0 and j >= 0:
            if nums1[i] > nums2[j]:
                nums1[k] = nums1[i]
                k -= 1
                i -= 1
            else:
                nums1[k] = nums2[j]
                j -= 1
                k -= 1

        while j >= 0:
            nums1[k] = nums2[j]
            j -= 1
            k -= 1

        return nums1

## 16. Code Breakdown

### Step 1

`i = m - 1`

Points to the last actual element in `nums1`.

Ignore the extra zero placeholders.

### Step 2

`j = n - 1`

Points to the last element in `nums2`.

### Step 3

`k = m + n - 1`

Points to the final position of the merged array.

### Step 4

`while i >= 0 and j >= 0`

Continue comparing while both arrays have unprocessed elements.

### Step 5

`if nums1[i] > nums2[j]`

If the current `nums1` value is larger, put it at `k`.

### Step 6

`else`

Otherwise, put the current `nums2` value at `k`.

This also handles equality.

### Step 7

Move the appropriate pointer and always move:

`k`

because one position has been filled.

### Step 8

`while j >= 0`

Copy any remaining `nums2` elements.

### Step 9

`return nums1`

Return the merged array.

---

## 17. Important Code Observation

The key trick is:

**Do not merge from the beginning.**

Merge from the **end**.

You have three positions:

`i → nums1`

`j → nums2`

`k → final position`

The invariant is:

**Everything after `k` is already correctly placed.**

Each iteration places the largest remaining value into `k`.

This makes the in-place merge possible without an extra array.

---

## 18. Common Mistakes

### Mistake 1: Starting From the Front

If you start at index `0`, you can overwrite elements in `nums1` that you still need.

### Mistake 2: Using `len(nums1) - 1` for `i`

`nums1` contains extra zero placeholders.

`i` must start at:

`m - 1`

not:

`len(nums1) - 1`

### Mistake 3: Forgetting `k`

You need a separate pointer for where the next largest value should be placed.

### Mistake 4: Forgetting the Remaining `nums2` Elements

After the main loop, `nums2` may still have values.

That is why:

`while j >= 0`

is required.

### Mistake 5: Adding a Loop for Remaining `nums1`

Not necessary.

If `nums2` is exhausted, the remaining `nums1` values are already correctly positioned.

### Mistake 6: Comparing the Wrong Values

Always compare:

`nums1[i]`

with:

`nums2[j]`

These are the largest unprocessed elements in each array.

## 19. Complexity

### Time Complexity

`O(m + n)`

Every element from both arrays is processed at most once.

### Space Complexity

`O(1)`

No extra array is created.

The merge happens directly inside `nums1`.

## 20. Pattern Recognition + Interview Explanation

### Pattern

**Two Pointers + Reverse Merge + In-Place Array Modification**

### Trigger

When you see:

- Two sorted arrays
- One array has enough extra space
- Need to merge in-place
- Need sorted output

Think:

**Merge From the End**

### General Pattern

`i → end of nums1's actual data`

`j → end of nums2`

`k → end of nums1`

Compare:

`nums1[i]` vs `nums2[j]`

Put the larger value at:

`nums1[k]`

Then move backward.

### Interview Explanation

"Both arrays are sorted, and `nums1` has enough empty space at the end. I use three pointers: `i` for the last valid element in `nums1`, `j` for the last element in `nums2`, and `k` for the last position in the final array. I compare the two largest remaining elements and place the larger one at `k`, moving the pointers backward. This prevents overwriting unprocessed values in `nums1`. Once one array is exhausted, only remaining elements from `nums2` need to be copied. The solution takes O(m+n) time and O(1) extra space."
