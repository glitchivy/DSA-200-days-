# LC 643 - Maximum Average Subarray I

## 1. Problem

Given an integer array `nums` and an integer `k`, find the contiguous subarray of length exactly `k` that has the maximum average value.

Return the maximum average.

### Example

Input:

`nums = [1,12,-5,-6,50,3]`

`k = 4`

Possible windows:

`[1,12,-5,-6]`

Sum:

`2`

Average:

`0.5`

---

`[12,-5,-6,50]`

Sum:

`51`

Average:

`12.75`

---

`[-5,-6,50,3]`

Sum:

`42`

Average:

`10.5`

Therefore:

`12.75`

is the maximum average.

---

## 2. Core Idea

Use the **Fixed-Size Sliding Window** pattern.

The subarray length is always exactly:

`k`

So instead of recalculating the sum of every window from scratch, maintain a running sum.

### Memory Rule

**First window → slide by removing left → add new right**

---

## 3. Why Sliding Window?

Every possible subarray has the same size:

`k`

That is the key trigger.

If the current window is:

`[1,2,3,4]`

and `k = 4`

the next window is:

`[2,3,4,5]`

Only two things changed:

- Remove `1`
- Add `5`

Therefore:

`new_sum = old_sum - outgoing + incoming`

This avoids recalculating the entire sum.

---

## 4. Initial Window

Your code starts with:

`max_sum = sum(nums[:k])`

and:

`curr_sum = sum(nums[:k])`

This calculates the first window of size `k`.

Example:

`nums = [1,12,-5,-6,50,3]`

`k = 4`

First window:

`[1,12,-5,-6]`

Sum:

`1 + 12 - 5 - 6 = 2`

Therefore:

`curr_sum = 2`

and:

`max_sum = 2`

---

## 5. Sliding the Window

Start scanning from:

`i = k`

because the first `k` elements have already been processed.

For every new element:

`nums[i]`

we add it to the window.

But we must remove the element that is no longer inside the window:

`nums[i-k]`

Therefore:

`curr_sum += nums[i] - nums[i-k]`

This is the central sliding-window operation.

---

## 6. Why `nums[i-k]`?

Suppose:

`k = 3`

Current window:

`[1,2,3]`

Next element:

`4`

The next window should be:

`[2,3,4]`

The element leaving the window is:

`1`

Its index is:

`i-k`

If:

`i = 3`

then:

`i-k = 3-3 = 0`

So:

`nums[i-k] = nums[0] = 1`

Therefore:

`new_sum = old_sum - nums[i-k] + nums[i]`

---

`nums = [1,12,-5,-6,50,3]`

`k = 4`

Initial window:

`[1,12,-5,-6]`

Sum:

`curr_sum = 2`

`max_sum = 2`

`nums[4] = 50`

`nums[0] = 1`

`curr_sum = 2 + 50 - 1`

`i = 5`

Incoming:

`nums[5] = 3`

`nums[1] = 12`

Current window:

`[-5,-6,50,3]`

## 10. Final Calculation

We found:

`max_sum = 51`

The average is:

`max_sum / k`

Therefore:

`51 / 4 = 12.75`

Return:

`12.75`

## 12. Code

class Solution:
    def findMaxAverage(self, nums: list[int], k: int) -> float:
        max_sum = sum(nums[:k])
        curr_sum = sum(nums[:k])

        for i in range(k, len(nums)):
            curr_sum += nums[i] - nums[i-k]
            max_sum = max(curr_sum, max_sum)

        return max_sum / k

---

## 13. Code Breakdown

### Step 1

`max_sum = sum(nums[:k])`

Calculate the sum of the first window.

### Step 2

`curr_sum = sum(nums[:k])`

Keep track of the current window's sum.

### Step 3

`for i in range(k, len(nums)):`

Start from the first element outside the initial window.

### Step 4

`curr_sum += nums[i] - nums[i-k]`

Slide the window:

- Add the incoming element.
- Remove the outgoing element.

### Step 5

`max_sum = max(curr_sum, max_sum)`

Keep the largest window sum.

### Step 6

`return max_sum / k`

Every window has exactly `k` elements.

Therefore the window with the largest sum also has the largest average.

---

## 14. Important Observation

We do **not** need to calculate the average for every window.

All windows have exactly the same size:

`k`

Therefore:

`average = sum / k`

Since `k` is constant:

**Maximum sum = Maximum average**

So we only track the maximum sum.

This is an important optimization and reasoning step.

---

## 15. Why Not Recalculate Every Window?

A brute-force solution could calculate:

`sum(nums[i:i+k])`

for every possible starting position.

Each window takes:

`O(k)`

to calculate.

There are approximately:

`O(n)`

windows.

Therefore:

`O(nk)`

time.

The sliding window avoids this repeated work.

Instead of recalculating:

`k` elements

we only:

- remove one element
- add one element

per shift.

Therefore:

`O(n)`

time.

## 16. Pattern Recognition

### Pattern

**Fixed-Size Sliding Window**

### Trigger

When you see:

- Contiguous subarray
- Exactly `k` elements
- Need maximum/minimum/sum/average
- Every window has the same size

Think:

**Fixed-Size Sliding Window**

### General Pattern

1. Build the first window.
2. Store its state.
3. Slide one position.
4. Remove the outgoing element.
5. Add the incoming element.
6. Update the answer.

### Memory Rule

**Fixed `k` → Window slides one step at a time**

---

## 17. Common Mistakes

### Mistake 1: Recalculating Every Window

This gives:

`O(nk)`

instead of:

`O(n)`

### Mistake 2: Removing the Wrong Element

When the current index is `i`, the outgoing element is:

`nums[i-k]`

### Mistake 3: Forgetting the Initial Window

You need to initialize the first `k` elements before starting the sliding process.

### Mistake 4: Tracking Average Instead of Sum

Not necessary.

Because every window has the same size:

`k`

the maximum sum automatically gives the maximum average.

### Mistake 5: Using a Variable Window

This is not variable-size sliding window.

The size is always exactly:

`k`

Therefore this is **Fixed-Size Sliding Window**.

---

## 18. Edge Cases

### `k = 1`

Every individual element is a window.

The answer is simply the maximum element.

### `k = len(nums)`

There is only one possible window.

Return the average of the entire array.

### Negative Numbers

The algorithm still works.

Example:

`nums = [-5,-2,-8,-1]`

`k = 2`

The maximum average comes from the pair with the largest sum.

### Duplicate Values

No special handling is needed.

### Decimal Result

The final division:

`max_sum / k`

produces the average.

---

## 19. Complexity

### Time Complexity

`O(n)`

The first window takes `O(k)` and the remaining elements are processed once.

Since:

`k <= n`

overall:

`O(n)`

### Space Complexity

`O(1)`

Only:

- `max_sum`
- `curr_sum`
- `i`

and a few constant variables are used.

No extra array is created.

### Fixed-Size

Window size is explicitly fixed:

`k`

### Variable-Size

Window expands/shrinks based on a condition.

Typical trigger:

"Longest/shortest subarray satisfying condition"

This distinction is extremely important for pattern recognition.
