# LC 2090 - K Radius Subarray Averages

## 1. Problem

Given an integer array `nums` and an integer `k`, calculate the average of every subarray centered at each index with radius `k`.

For an index `i`, the required window is:

`nums[i-k ... i ... i+k]`

Therefore, every valid window contains:

`2*k + 1`

elements.

If there are not enough elements on both sides of an index to form this window, the answer for that index is:

`-1`

### Example

`nums = [7,4,3,9,1,8,5,2,6]`

`k = 3`

Window size:

`2*k + 1 = 7`

For index `3`:

`[7,4,3,9,1,8,5]`

Average:

`37 // 7 = 5`

So:

`result[3] = 5`

---

## 2. Core Idea

This is a **Fixed-Size Sliding Window + Center Mapping** problem.

The size of every required window is:

`2*k + 1`

Instead of calculating the sum of every window from scratch, maintain a running sum.

When the window moves one position:

- Add the new element entering the window.
- Remove the old element leaving the window.
- Calculate the average.
- Store it at the correct center index.

The key idea is:

**The window moves by one position, but the answer belongs to the center of that window.**

---

## 3. Pointer / Variable Roles

### `size`

Stores the size of each averaging window:

`size = 2*k + 1`

### `result`

Stores the final averages.

Initially:

`[-1, -1, -1, ...]`

The `-1` represents indices where a complete radius-`k` window cannot exist.

### `n`

Stores the length of `nums`.

### `current_sum`

Stores the sum of the current window.

### `i`

Represents the right endpoint of the current window.

### `center`

Represents the center index corresponding to the current window.

Formula:

`center = i-k`

---

## 4. Initialization

First calculate the window size:

`size = 2*k + 1`

Then initialize:

`result = [-1] * len(nums)`

This automatically marks every position as invalid.

If:

`size > n`

then no complete window exists anywhere.

Therefore:

`return result`

Otherwise, calculate the first window:

`current_sum = sum(nums[:size])`

The first window covers:

`indices 0 ... size-1`

Its center is:

`k`

Therefore:

`result[k] = current_sum // size`

---

## 5. Main Decision Logic

After processing the first window, slide it across the array.

For every:

`i` from `size` to `n-1`

the new element entering the window is:

`nums[i]`

The element leaving is:

`nums[i-size]`

So:

`current_sum += nums[i] - nums[i-size]`

Now the current window's center is:

`center = i-k`

Calculate:

`result[center] = current_sum // size`

Continue until all possible windows have been processed.

---

## 6. Why It Works

Consider:

`k = 2`

Then:

`size = 2*2 + 1 = 5`

For center index `2`, the window is:

`[0,1,2,3,4]`

For center index `3`, the window becomes:

`[1,2,3,4,5]`

The two windows overlap heavily.

Instead of recalculating the second window's entire sum:

`nums[1] + nums[2] + nums[3] + nums[4] + nums[5]`

we reuse the previous sum.

Remove:

`nums[0]`

Add:

`nums[5]`

Therefore:

`new_sum = old_sum - nums[0] + nums[5]`

This gives constant-time window updates.

`0 1 2 3 4 5 6`

Values:

`7 4 3 9 1 8 5`

Sum:

`37`

Average:

`37 // 7 = 5`

Center:

`3`

Therefore:

`result[3] = 5`

## 8. Dry Run Continuation

### Next window

Add:

`nums[7] = 2`

Remove:

`nums[0] = 7`

New sum:

`37 + 2 - 7 = 32`

Current window:

`[4,3,9,1,8,5,2]`

Center:

`7 - 3 = 4`

Average:

`32 // 7 = 4`

Therefore:

`result[4] = 4`

---

### Next window

Add:

`nums[8] = 6`

Remove:

`nums[1] = 4`

New sum:

`32 + 6 - 4 = 34`

Current window:

`[3,9,1,8,5,2,6]`

Center:

`8 - 3 = 5`

Average:

`34 // 7 = 4`

Therefore:

`result[5] = 4`

Final result:

`[-1,-1,-1,5,4,4,-1,-1,-1]`

````python
class Solution(object):
    def getAverages(self, nums, k):
        """
        :type nums: List[int]
        :type k: int
        :rtype: List[int]
        """
        size=2*k+1 
        result=[-1]*len(nums)
        n=len(nums)
        if size > n:
            return result
        
        current_sum=sum(nums[:size])
        result[k]=current_sum//size

        for i in range(size,n):
            current_sum+=nums[i]-nums[i-size]

            center=i-k
            result[center]=current_sum//size 
        return result


        
````
