# LC 1876 - Substrings of Size Three with Distinct Characters

## 1. Problem

Given a string `s`, count the number of substrings of length exactly `3` where all three characters are distinct.

### Example

`s = "xyzzaz"`

Substrings of length `3`:

- `"xyz"` → `x, y, z` are distinct → ✅
- `"yzz"` → `z` repeats → ❌
- `"zza"` → `z` repeats → ❌
- `"zaz"` → `z` repeats → ❌

Answer:

`1`

---

## 2. Core Idea

This is a **Fixed-Size Sliding Window** problem.

The window size is always:

`3`

For every group of 3 consecutive characters, we need to check whether all three characters are different.

For three characters:

`a, b, c`

the condition is:

`a != b`

`b != c`

`a != c`

If all three conditions are true, the substring is good.

---

## 3. Pointer / Variable Roles

### `count`

Stores the number of good substrings found.

### `loop`

Stores the first substring of length `3`.

### `set(loop)`

Removes duplicates from the first window.

If:

`len(set(loop)) == 3`

then all three characters are distinct.

### `i`

Represents the index of the new character entering the current size-3 window.

---

## 4. Initialization

The first window is:

`loop = s[:3]`

Then convert it to a set:

`loop = set(loop)`

If:

`len(loop) == 3`

then all three characters are unique.

Therefore:

`count += 1`

This handles the first window separately.

---

## 5. Main Decision Logic

After processing the first window, every new window can be represented by:

`s[i-2], s[i-1], s[i]`

Instead of creating a new set for every window, directly compare the three characters.

The substring is good if:

`s[i] != s[i-1]`

AND

`s[i] != s[i-2]`

AND

`s[i-1] != s[i-2]`

If all three are different:

`count += 1`

---

## 6. Why It Works

A substring has exactly three characters.

For three characters to be distinct, every pair must be different.

For:

`a, b, c`

we need:

`a != b`

`a != c`

`b != c`

Your condition checks exactly these three pairwise relationships.

Therefore, the condition is both necessary and sufficient.

---

## 7. Dry Run

### Input

`s = "xyzzaz"`

First window:

`"xyz"`

Set:

`{'x','y','z'}`

Length:

`3`

Therefore:

`count = 1`

---

## 8. Dry Run Continuation

### Window 2

`"yzz"`

Characters:

`y, z, z`

Check:

`z != z`

False.

Therefore:

`count = 1`

---

### Window 3

`"zza"`

Characters:

`z, z, a`

Check:

`z != z`

False.

Therefore:

`count = 1`

---

### Window 4

`"zaz"`

Characters:

`z, a, z`

Check:

`z != z`

False.

Therefore:

`count = 1`

Final answer:

`1`

---

## 9. Visual Dry Run

```text
s = x y z z a z
    └─────┘
      xyz       → distinct → ✓

      └─────┘
        yzz     → z repeats → ✗

        └─────┘
          zza   → z repeats → ✗

          └─────┘
            zaz → z repeats → ✗

Answer = 1

````python
class Solution(object):
    def countGoodSubstrings(self, s):
        """
        :type s: str
        :rtype: int
        """
        count=0 
        # len is 3 
        loop=s[:3]
        loop=set(loop)

        if len(loop)==3:
            count+=1
        for i in range(3,len(s)):
            if s[i]!=s[i-1] and s[i]!=s[i-2] and s[i-1]!=s[i-2]:
                count+=1
        return count 
````
