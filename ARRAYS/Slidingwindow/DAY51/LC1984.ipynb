# LC 167 - Two Sum II - Input Array Is Sorted

## 1. Problem

Given a **sorted** array `numbers` and a target value, find two numbers whose sum equals the target.

Return their **1-indexed positions**.

### Example

Input:

`numbers = [2,7,11,15]`

`target = 9`

Output:

`[1,2]`

Because:

`2 + 7 = 9`

---

## 2. Core Idea

Use the **Two Pointer** approach.

- `left` starts at the beginning.
- `right` starts at the end.
- Calculate the sum of the two values.
- Use the sorted property to decide which pointer to move.

Mental model:

`left → [smaller values ........ larger values] ← right`

---

## 3. Pointer Initialization

Initialize:

`left = 0`

`right = n - 1`

For:

`numbers = [2,7,11,15]`

we have:

```text
Index:   0  1   2   3
Value:  [2, 7, 11, 15]
          ↑          ↑
        left       right

````python
class Solution(object):
    def minimumDifference(self, nums, k):
        """
        :type nums: List[int]
        :type k: int
        :rtype: int
        """
        nums.sort()
        max_value=nums[k-1]-nums[0]
        for i in range(k,len(nums)):
            differnce=nums[i]-nums[i-k+1]
            max_value=min(max_value,differnce)
        return max_value
````
