# LC 1343 - Number of Sub-arrays of Size K and Average Greater than or Equal to Threshold

## 1. Problem

Given an integer array `arr`, an integer `k`, and a `threshold`, count how many contiguous subarrays of size exactly `k` have an average greater than or equal to `threshold`.

### Example

`arr = [2,2,2,2,5,5,5,8]`  
`k = 3`  
`threshold = 4`

We need to check every window of size `3`:

- `[2,2,2]` → average = 2 → ❌
- `[2,2,2]` → average = 2 → ❌
- `[2,2,5]` → average = 3 → ❌
- `[2,5,5]` → average = 4 → ✅
- `[5,5,5]` → average = 5 → ✅
- `[5,5,8]` → average = 6 → ✅

Answer = `3`

---

## 2. Core Idea

This is a **fixed-size sliding window** problem.

Every subarray must contain exactly `k` elements.

Instead of calculating the sum of every window from scratch, maintain the sum of the current window.

When the window moves one position:

- Remove the element leaving the window.
- Add the new element entering the window.

So:

`new_sum = old_sum - outgoing_element + incoming_element`

---

## 3. Pointer / Variable Roles

### `count`

Stores the number of windows whose average is greater than or equal to `threshold`.

### `current_sum`

Stores the sum of the current window of size `k`.

### `i`

Represents the index of the new element entering the sliding window.

### `n`

Stores the length of the array.

---

## 4. Initialization

First create the first window:

`current_sum = sum(arr[:k])`

This gives the sum of the first `k` elements.

Then check whether:

`current_sum / k >= threshold`

If yes:

`count += 1`

This handles the first window before the sliding process begins.

---

## 5. Main Decision Logic

After processing the first window, slide the window from left to right.

For every new element `arr[i]`:

1. Add `arr[i]`
2. Remove `arr[i-k]`
3. Calculate the new average
4. If the average is at least `threshold`, increase `count`

The update is:

`current_sum += arr[i] - arr[i-k]`

This is the key sliding-window operation.

---

## 6. Why It Works

Suppose the current window is:

`[2, 5, 5]`

and the next window should be:

`[5, 5, 8]`

We do not need to calculate:

`5 + 5 + 8`

from scratch.

Instead:

`old_sum = 2 + 5 + 5`

Remove `2` and add `8`:

`new_sum = old_sum - 2 + 8`

Therefore:

`new_sum = old_sum + arr[i] - arr[i-k]`

This avoids repeatedly summing `k` elements.

````python
class Solution(object):
    def numOfSubarrays(self, arr, k, threshold):
        """
        :type arr: List[int]
        :type k: int
        :type threshold: int
        :rtype: int
        """
        # avg > threshoold 
        # needs to count 
        count=0 
        #return count 
        n=len(arr)
        current_sum=sum(arr[:k])
        if current_sum/k>=threshold:
            count+=1
        for i in range(k,n):
            current_sum+=arr[i]-arr[i-k]
            value =current_sum/k 
            if value>=threshold:
                count+=1
            # current_sum+=arr[i]-arr[i-k]
        return count
````
