# LC 1456 - Maximum Number of Vowels in a Substring of Given Length

## 1. Problem

Given a string `s` and an integer `k`, find the maximum number of vowels in any substring of length exactly `k`.

Vowels are:

`a, e, i, o, u`

### Example

`s = "abciiidef"`  
`k = 3`

Possible windows:

- `"abc"` → 1 vowel
- `"bci"` → 1 vowel
- `"cii"` → 2 vowels
- `"iii"` → 3 vowels
- `"iid"` → 2 vowels
- `"ide"` → 2 vowels
- `"def"` → 1 vowel

Answer = `3`

---

## 2. Core Idea

This is a **Fixed-Size Sliding Window** problem.

Every substring must have exactly `k` characters.

Instead of counting vowels from scratch for every substring, maintain the number of vowels in the current window.

When the window moves one position:

- Remove the character leaving the window.
- Add the character entering the window.
- Update the maximum vowel count.

The core operation is:

`new_count = old_count - outgoing + incoming`

---

## 3. Pointer / Variable Roles

### `value`

Stores the vowels:

`{'a','e','i','o','u'}`

Used to check whether a character is a vowel.

### `vowel_count`

Stores the number of vowels in the current window.

### `max_vowel`

Stores the maximum number of vowels found so far.

### `i`

Represents the incoming character's index.

### `k`

Represents the fixed window size.

### `window`

Represents the first window of size `k`.

---

## 4. Initialization

First create the first window:

`window = s[:k]`

Count all vowels inside this window.

Then:

`max_vowel = vowel_count`

The first window becomes the initial answer.

This is important because the sliding loop starts only after the first window has already been processed.

---

## 5. Main Decision Logic

For every character entering the window:

### Step 1: Remove outgoing character

The outgoing character is:

`s[i-k]`

If it is a vowel:

`vowel_count -= 1`

### Step 2: Add incoming character

The incoming character is:

`s[i]`

If it is a vowel:

`vowel_count += 1`

### Step 3: Update maximum

`max_vowel = max(max_vowel, vowel_count)`

### Step 4: Early termination

If:

`max_vowel == k`

then the current window contains `k` vowels.

Since the window has only `k` characters, it is impossible to do better.

Therefore:

`return k`

---

## 6. Why It Works

Consider:

`"abcii"`

Suppose we move the window one position:

`"abcii"` → `"bciii"`

We do not need to recount all characters.

The outgoing character is:

`a`

Since `a` is a vowel:

`vowel_count -= 1`

The incoming character is:

`i`

Since `i` is a vowel:

`vowel_count += 1`

So the new vowel count is obtained in `O(1)` time.

The overlapping part of the two windows is reused instead of recalculated.

---

## 7. Dry Run

### Input

`s = "abciiidef"`

`k = 3`

Vowels:

`{'a','e','i','o','u'}`

### First window

`"abc"`

Vowels:

`a`

Therefore:

`vowel_count = 1`

`max_vowel = 1`

---

## 8. Dry Run Continuation

### Window 2

`"bci"`

Outgoing:

`a`

`a` is a vowel.

`vowel_count = 1 - 1 = 0`

Incoming:

`i`

`i` is a vowel.

`vowel_count = 0 + 1 = 1`

`max_vowel = 1`

---

### Window 3

`"cii"`

Outgoing:

`b`

`b` is not a vowel.

Incoming:

`i`

`i` is a vowel.

`vowel_count = 2`

`max_vowel = 2`

---

### Window 4

`"iii"`

Outgoing:

`c`

`c` is not a vowel.

Incoming:

`i`

`i` is a vowel.

`vowel_count = 3`

`max_vowel = 3`

Since:

`max_vowel == k`

we can immediately return:

`3`

Final answer:

`3`

---

## 9. Visual Dry Run

```text
s = a b c i i i d e f
    └─────┘
      abc       → 1 vowel

      └─────┘
        bci     → 1 vowel

        └─────┘
          cii   → 2 vowels

          └─────┘
            iii → 3 vowels ✓

Answer = 3

````python
class Solution(object):
    def maxVowels(self, s, k):
        """
        :type s: str
        :type k: int
        :rtype: int
        """
        value={'a','e','i','o','u'}
        vowel_count=0
        window=s[:k]
        for i in window:
            if i in value:
                vowel_count+=1
        max_vowel=vowel_count
        for i in range(k,len(s)):
            if s[i-k] in value:
                vowel_count-=1
            if s[i] in value:
                vowel_count+=1
            max_vowel=max(max_vowel,vowel_count)
            if max_vowel==k:
                return k 
        return max_vowel
````
