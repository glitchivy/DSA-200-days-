# LC 1052 - Grumpy Bookstore Owner

## 1. Problem

Given `customers`, `grumpy`, and an integer `minutes`, maximize the number of satisfied customers by using a technique that suppresses the owner's grumpiness for `minutes` consecutive minutes.

- `customers[i]` represents customers arriving during minute `i`.
- `grumpy[i] == 0`: Customers are already satisfied.
- `grumpy[i] == 1`: Customers are initially unsatisfied.

The goal is to return the maximum number of satisfied customers.

### Example

Input:

`customers = [1,0,1,2,1,1,7,5]`

`grumpy = [0,1,0,1,0,1,0,1]`

`minutes = 3`

Already satisfied customers:

`base = 1 + 1 + 1 + 7 = 10`

The best window is indices `5` to `7`, recovering `1 + 5 = 6` additional customers.

Answer:

`10 + 6 = 16`

---

## 2. Core Idea

This problem combines **Fixed-Size Sliding Window + Baseline + Maximum Gain**.

Separate the answer into two parts:

1. `base`: Customers already satisfied when the owner is not grumpy.
2. `max_gain`: Maximum additional customers recovered by applying the technique to a window of size `minutes`.

Formula:

`answer = base + max_gain`

The baseline is calculated once. The sliding window tracks only the additional customers who can be recovered.

---

## 3. Pointer / Variable Roles

- `base`: Total customers already satisfied, where `grumpy[i] == 0`.
- `count`: Recoverable customers in the current window, counting only positions where `grumpy[i] == 1`.
- `max_gain`: Maximum recoverable customer count across all windows.
- `i`: Index of the incoming minute.
- `minutes`: Fixed window size.
- `grumpy[i-minutes]`: Determines whether the outgoing minute contributed to the current gain.

---

## 4. Initialization

### Step 1: Calculate the baseline

Traverse the entire array.

Whenever `grumpy[i] == 0`, add `customers[i]` to `base`.

### Step 2: Calculate the first window's gain

Traverse the first `minutes` positions.

Whenever `grumpy[i] == 1`, add `customers[i]` to `count`.

### Step 3: Initialize the maximum

`max_gain = count`

The first window becomes the initial best window.

---

## 5. Main Decision Logic

For every incoming index `i`, starting at `minutes`:

### Step 1: Add the incoming contribution

If `grumpy[i] == 1`:

`count += customers[i]`

### Step 2: Remove the outgoing contribution

If `grumpy[i-minutes] == 1`:

`count -= customers[i-minutes]`

### Step 3: Update the maximum gain

`max_gain = max(max_gain, count)`

Finally:

`return base + max_gain`

Both window updates must happen before updating the maximum.

---

## 6. Why It Works

The total satisfied customers consist of:

**Already satisfied customers + Additional customers recovered by the technique.**

The baseline is independent of the chosen window.

Only the additional gain changes depending on which `minutes` consecutive positions are selected.

For each window, we count customers who were initially unsatisfied. Every such customer becomes satisfied when the technique is applied.

Since the algorithm examines every possible window and maintains the maximum gain, it finds the optimal result.

## 10. Code

    class Solution(object):
        def maxSatisfied(self, customers, grumpy, minutes):
            base = 0
            count = 0

            for i in range(len(customers)):
                if grumpy[i] == 0:
                    base += customers[i]

            for i in range(minutes):
                if grumpy[i] == 1:
                    count += customers[i]

            max_gain = count

            for i in range(minutes, len(customers)):
                if grumpy[i] == 1:
                    count += customers[i]

                if grumpy[i-minutes] == 1:
                    count -= customers[i-minutes]

                max_gain = max(max_gain, count)

            return max_gain + base

---

## 11. Code Breakdown

### Step 1: Initialize

`base = 0` and `count = 0`.

### Step 2: Calculate the baseline

Add customers from every minute where `grumpy[i] == 0`.

### Step 3: Calculate the first window

Count customers only at grumpy positions within the first `minutes` elements.

### Step 4: Initialize the maximum gain

`max_gain = count`

### Step 5: Slide the window

Start from index `minutes`.

### Step 6: Add the incoming contribution

If the incoming minute is grumpy, add its customers to `count`.

### Step 7: Remove the outgoing contribution

If the outgoing minute is grumpy, subtract its customers from `count`.

### Step 8: Update the maximum

Keep the larger of the current gain and the previous maximum.

### Step 9: Return the total

Add `base` and `max_gain`.

---

## 12. Important Observation

The baseline and sliding-window gain must remain separate.

- `base` contains customers already satisfied.
- `count` contains only customers recoverable inside the current window.

Do not count customers from non-grumpy minutes in the window gain because they are already included in `base`.

This decomposition simplifies the problem significantly.

---

## 13. Edge Cases

### Case 1: No grumpy minutes

Everyone is already satisfied.

Answer: `sum(customers)`

### Case 2: All minutes are grumpy

The baseline is `0`. The answer is the maximum sum of any window of size `minutes`.

### Case 3: `minutes = 1`

Choose the single minute that gives the greatest additional gain.

### Case 4: `minutes = len(customers)`

The technique covers the entire array.

Answer: `sum(customers)`

### Case 5: Incoming minute is not grumpy

No additional gain is added.

### Case 6: Outgoing minute is not grumpy

No additional gain is removed.

---

## 14. Common Mistakes

### Mistake 1: Counting all customers inside the window

Only customers at grumpy positions contribute to the recoverable gain.

### Mistake 2: Adding the baseline repeatedly

Calculate the baseline once, outside the sliding-window loop.

### Mistake 3: Forgetting the outgoing contribution

The previous window's contribution must be removed.

### Mistake 4: Confusing `base` with `count`

`base` is already satisfied. `count` is recoverable.

### Mistake 5: Returning only `max_gain`

The final answer must include both components:

`base + max_gain`

## 15. Complexity

### Time Complexity

`O(n)`

- Baseline calculation: `O(n)`
- Initial window: `O(minutes)`
- Sliding window: `O(n)`

Overall: `O(n)`.

### Space Complexity

`O(1)` auxiliary space because only a constant number of variables are used.

**Pattern: Fixed-Size Sliding Window + Baseline + Maximum Gain**

Recognize this pattern when:

- Part of the answer is already guaranteed.
- An operation improves the result within a fixed-length window.
- The window contribution can be updated incrementally.
- The objective is to maximize the additional benefit.

Calculate the guaranteed part once, then optimize the remaining part.

## 17. Pattern Trigger / Generalization

When an operation can improve a result for exactly `k` consecutive positions, ask:

1. What is already correct without the operation?
2. What additional benefit can the operation provide?
3. Can that benefit be tracked with a fixed-size window?
4. How should the baseline and best gain be combined?

General formula:

`answer = baseline + maximum_gain`

For this problem:

- Baseline: customers during non-grumpy minutes.
- Gain: customers during grumpy minutes inside the chosen window.

```python
class Solution(object):
    def maxSatisfied(self, customers, grumpy, minutes):
        """
        :type customers: List[int]
        :type grumpy: List[int]
        :type minutes: int
        :rtype: int
        """
        base=0 
        count=0
        for i in range(len(customers)):
            if grumpy[i]==0:
                base+=customers[i]
        
        for i in range(minutes):
            if grumpy[i]==1:
                count+=customers[i]
        max_gain=count
        for i in range(minutes,len(customers)):
            if grumpy[i]==1:
                count+=customers[i]
            if grumpy[i-minutes]==1:
                count-=customers[i-minutes]
            max_gain=max(max_gain,count)
        return max_gain+base
```
