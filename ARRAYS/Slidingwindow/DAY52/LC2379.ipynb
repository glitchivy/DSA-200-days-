# LC 2379 - Minimum Recolors to Get K Consecutive Black Blocks

## 1. Problem

Given a string `blocks` containing `B` (black) and `W` (white), find the minimum number of white blocks that must be recolored black to obtain `k` consecutive black blocks.

### Example

`blocks = "WBBWWBBWBW"`

`k = 7`

We need to find a contiguous window of exactly `k` blocks containing the fewest white blocks.

**Answer = minimum number of white blocks in any window of size `k`.**

---

## 2. Core Idea

This is a **Fixed-Size Sliding Window + Running Count** problem.

Every candidate window contains exactly `k` blocks.

To make all blocks black, every white block in the selected window must be recolored.

Instead of counting white blocks from scratch for every window, maintain the number of white blocks in the current window.

When the window moves:
- Add the incoming character's contribution.
- Remove the outgoing character's contribution.
- Update the minimum white count.

The core operation is:

`new_count = old_count + incoming_white - outgoing_white`

---

## 3. Pointer / Variable Roles

### `seq`
Stores the first substring of length `k`.

### `count`
Stores the number of white blocks in the current window.

### `min_count`
Stores the minimum number of white blocks found in any window so far.

### `i`
Represents the index of the incoming character.

### `blocks[i-k]`
Represents the outgoing character leaving the current window.

### `k`
The fixed window size.

---

## 4. Initialization

First, extract the initial window:

`seq = blocks[:k]`

Count the white blocks in this window:

- If a character is `"W"`, increment `count`.
- If a character is `"B"`, do nothing.

Then initialize:

`min_count = count`

The first window becomes the initial best answer.

---

## 5. Main Decision Logic

For each incoming character, starting at index `k`:

### Step 1: Add the incoming character

If:

`blocks[i] == "W"`

then:

`count += 1`

### Step 2: Remove the outgoing character

If:

`blocks[i-k] == "W"`

then:

`count -= 1`

### Step 3: Update the minimum

`min_count = min(count, min_count)`

Both updates happen before evaluating the minimum for the new window.

---

## 6. Why It Works

Suppose the current window contains:

`"BWWB"`

It has two white blocks, so two recolors are required.

When the window moves one position, one character leaves and another enters.

- An incoming `"W"` increases the count.
- An outgoing `"W"` decreases the count.
- Black blocks do not affect the count.

This maintains the exact number of white blocks in the current window.

Since every possible window of size `k` is evaluated, the smallest white count found is the minimum number of recolors required.

---

## 7. Dry Run

### Input

`blocks = "WBBWWBBWBW"`

`k = 7`

The first window is:

`"WBBWWBB"`

White blocks:

`W, W, W`

Therefore:

`count = 3`

Initialize:

`min_count = 3`

---

## 8. Dry Run Continuation

### Window 1

`"WBBWWBB"`

White count = `3`

`min_count = 3`

### Window 2

`"BBWWBBW"`

Outgoing character: `"W"`

Incoming character: `"W"`

Count:

`3 - 1 + 1 = 3`

`min_count = min(3, 3) = 3`

### Window 3

`"BWWBBWB"`

Outgoing character: `"B"`

Incoming character: `"B"`

Count remains:

`3`

`min_count = 3`

### Window 4

`"WWBBWBW"`

Outgoing character: `"B"`

Incoming character: `"W"`

Count:

`3 + 1 = 4`

`min_count = min(3, 4) = 3`

Final answer:

`3`

        4 white blocks

Minimum white count = 3

````python
class Solution(object):
    def minimumRecolors(self, blocks, k):
        """
        :type blocks: str
        :type k: int
        :rtype: int
        """
        seq=blocks[:k]
        count=0
        for o in seq:
            if o=="W":
                count+=1
        min_count=count
        for i in range(k,len(blocks)):
            if blocks[i]=="W":
                count+=1
            if blocks[i-k]=="W":
                count-=1
            min_count=min(count,min_count)
        return min_count
        
````
