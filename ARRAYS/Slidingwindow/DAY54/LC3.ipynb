# LC 3: Longest Substring Without Repeating Characters

## 1. Problem

Given a string `s`, find the length of the longest substring without repeating characters.

A substring is a contiguous sequence of characters within a string.

Return the maximum length of a substring in which every character appears at most once.

### Example

Input:

`s = "abcabcbb"`

Output:

`3`

Explanation:

The longest substrings without repeating characters include `"abc"`, `"bca"`, and `"cab"`.

Each has length `3`.

---

## 2. Core Idea

This problem uses **Variable-Size Sliding Window + Frequency Hash Map**.

Maintain a window containing no repeated characters.

- Expand the window using the right pointer.
- Track the frequency of each character.
- If the newly added character appears more than once, shrink the window from the left.
- Continue shrinking until the duplicate is removed.
- Calculate the window length and update the maximum.

The key condition is:

`freq[s[i]] <= 1`

Every character inside the valid window must appear exactly once.

---

## 3. Pointer / Variable Roles

- `s`: Input string.
- `left`: Starting index of the current window.
- `i`: Right endpoint of the current window.
- `freq`: Dictionary storing character frequencies in the current window.
- `window_size`: Length of the current valid window.
- `max_count`: Maximum valid substring length found so far.

Important distinction:

- `freq` tracks the number of occurrences of each character.
- `max_count` tracks the longest valid substring length.
- `left` moves forward whenever the current window contains a duplicate.

---

## 4. Initialization

Initialize:

`left = 0`

`max_count = 0`

`freq = {}`

Initially, the window is empty.

The loop moves the right endpoint through every character of the string.

Each character is added to the frequency map before the algorithm checks whether the window remains valid.

---

## 5. Main Decision Logic

### Step 1: Expand the window

For each character `s[i]`, add it to the frequency map.

If it is not present:

`freq[s[i]] = 1`

Otherwise:

`freq[s[i]] += 1`

### Step 2: Check for duplicates

While:

`freq[s[i]] > 1`

the current character appears more than once inside the window.

### Step 3: Shrink the window

Decrease the frequency of the character at the left endpoint:

`freq[s[left]] -= 1`

Then move the left pointer:

`left += 1`

Continue until the duplicate is removed.

### Step 4: Calculate the window length

`window_size = i - left + 1`

### Step 5: Update the maximum

`max_count = max(max_count, window_size)`

Finally, return `max_count`.

---

## 6. Why It Works

The algorithm maintains a window containing no repeated characters.

When a duplicate appears, the window becomes invalid. Moving the left pointer forward removes characters from the beginning until the duplicate character occurs only once again.

After the `while` loop, every character in the current window is unique.

The algorithm then calculates the length of this valid window and compares it with the longest valid window found earlier.

Because the right pointer visits every character and the left pointer only moves forward, the algorithm efficiently finds the longest substring without repeating characters.

Notice that the character `"a"` does not need to be deleted from the dictionary when its frequency becomes zero. A zero-frequency key does not cause a problem in this implementation because the algorithm checks the frequency of the incoming character, not `len(freq)`.

## 10. Code

    class Solution(object):
        def lengthOfLongestSubstring(self, s):
            left = 0
            max_count = 0
            freq = {}

            for i in range(len(s)):
                if s[i] not in freq:
                    freq[s[i]] = 1
                else:
                    freq[s[i]] += 1

                while freq[s[i]] > 1:
                    freq[s[left]] -= 1
                    left += 1

                window_size = i - left + 1
                max_count = max(max_count, window_size)

            return max_count

---

## 11. Code Breakdown

### Step 1: Initialize the left pointer

`left = 0`

This marks the beginning of the current substring.

### Step 2: Initialize the frequency map

`freq = {}`

This records how many times each character appears in the current window.

### Step 3: Add the incoming character

The character at index `i` is added to the dictionary or its frequency is increased.

### Step 4: Restore validity

If the incoming character appears more than once, shrink the window from the left.

Each removed character has its frequency decreased.

### Step 5: Calculate the current length

`i - left + 1`

The current window is valid after the `while` loop.

### Step 6: Update the maximum

Compare the current valid length with `max_count`.

### Step 7: Return the answer

After the loop, `max_count` contains the length of the longest valid substring.

---

## 12. Important Observation

The `while` loop is essential.

A single left-pointer movement may not remove the duplicate.

For example:

`s = "abba"`

When the second `"b"` enters, the window contains `"abb"`.

The algorithm must remove `"a"` and then the first `"b"` before the window becomes valid.

The condition:

`while freq[s[i]] > 1`

keeps shrinking until the newly added character appears only once.

Also, zero-frequency dictionary entries do not affect correctness here because the code checks the frequency of the current character directly.

## 13. Edge Cases

### Case 1: Empty string

`s = ""`

No iterations occur.

Answer: `0`

### Case 2: One character

`s = "a"`

Answer: `1`

### Case 3: All characters are identical

`s = "bbbbb"`

The window shrinks whenever a repeated character enters.

Answer: `1`

### Case 4: All characters are distinct

`s = "abcdef"`

The window keeps expanding without shrinking.

Answer: `6`

### Case 5: Duplicate characters are separated

`s = "pwwkew"`

The longest valid substring is `"wke"`.

Answer: `3`

### Case 6: Duplicate character at the beginning

`s = "abba"`

The algorithm must shrink enough to remove the earlier duplicate.

Answer: `2`

## 14. Common Mistakes

### Mistake 1: Using `if` instead of `while`

A duplicate may require removing multiple characters from the left.

### Mistake 2: Updating the maximum before shrinking

An invalid window must not contribute to the answer.

### Mistake 3: Forgetting the `+1`

The correct length is:

`i - left + 1`

### Mistake 4: Confusing substring with subsequence

A substring must be contiguous. You cannot skip characters.

### Mistake 5: Deleting a dictionary key unnecessarily

This implementation remains correct without deleting zero-frequency keys because it checks only the incoming character's frequency.

### Mistake 6: Assuming every repeated character requires moving `left` once

The left pointer must continue moving until the current window is valid.

## 15. Complexity

### Time Complexity

`O(n)`

The right pointer visits each character once.

The left pointer also moves forward at most `n` times.

Although the `while` loop is nested inside the `for` loop, the total number of left-pointer movements is bounded by `n`.

### Space Complexity

`O(min(n, Σ))` auxiliary space, where `Σ` is the size of the character set.

The frequency dictionary stores characters encountered in the window. Since this is a string problem, the space is bounded by the number of distinct characters that can appear.

## 16. Pattern Recognition

**Pattern: Variable-Size Sliding Window + Frequency Map**

Recognize this pattern when:

- The problem asks for the longest substring.
- The substring must satisfy a condition involving duplicates.
- You can maintain the condition using character frequencies.
- The window can expand and shrink while both pointers move only forward.

Typical structure:

1. Expand the right pointer.
2. Update the frequency map.
3. Shrink while the window is invalid.
4. Calculate the valid window length.
5. Track the maximum.

## 17. Pattern Trigger / Generalization

Ask these questions:

1. Is the problem asking for a contiguous substring?
2. Does validity depend on repeated characters?
3. Can a frequency map track the current window?
4. Can shrinking from the left restore validity?
5. Can both pointers move only forward?

If yes, variable-size sliding window is a strong candidate.

This pattern can be adapted to:

- Longest substring without repeating characters.
- Longest substring with at most `k` distinct characters.
- Longest window containing at most `k` replacements or violations.
- Minimum window substring, with a different validity condition and objective.

```python
class Solution(object):
    def lengthOfLongestSubstring(self, s):
        """
        :type s: str
        :rtype: int
        """
        left=0
        max_count=0 
        
        freq={}
        for i in range(len(s)):
            if s[i] not in freq:
            
                freq[s[i]]=1
            else:
                freq[s[i]]+=1
            
            while freq[s[i]]>1:
                freq[s[left]]-=1
                left+=1
                
            window_size=i-left+1
            max_count=max(max_count,window_size)
        return max_count

```
