# LC 424: Longest Repeating Character Replacement

## 1. Problem

Given a string `s` and an integer `k`, return the length of the longest substring that can be made to contain the same character by replacing at most `k` characters.

You can replace any character with any uppercase English letter.

### Example

Input:

`s = "AABABBA"`

`k = 1`

Output:

`4`

Explanation:

The substring `"AABA"` can be converted into `"AAAA"` by replacing one `B` with `A`.

Therefore, the maximum length is `4`.

---

## 2. Core Idea

This problem uses **Variable-Size Sliding Window + Frequency Map + Maximum Frequency**.

Maintain a window in which we can make every character identical using at most `k` replacements.

The key observation is that we should replace every character except the most frequent character in the window.

For a window of length `window_size`:

`replacements_needed = window_size - max_freq`

Where:

- `window_size` is the number of characters in the current window.
- `max_freq` is the frequency of the most common character in the window.

The window is valid when:

`window_size - max_freq <= k`

If this condition is violated, shrink the window from the left.

---

## 3. Pointer / Variable Roles

- `s`: Input string.
- `k`: Maximum number of character replacements allowed.
- `freq`: Dictionary storing character frequencies in the current window.
- `left`: Starting index of the current window.
- `i`: Right endpoint of the current window.
- `max_freq`: Frequency of the most common character in the window.
- `window_size`: Length of the current window.
- `max_window`: Maximum valid window length found so far.

Important distinction:

`max_freq` is the count of the most frequent character, not the character itself.

---

## 4. Initialization

Initialize:

`freq = {}`

`left = 0`

`max_window = 0`

The window begins empty.

The right pointer moves through the string one character at a time.

Each new character is added to the frequency map before the algorithm checks whether the window remains valid.

---

## 5. Main Decision Logic

### Step 1: Expand the window

For each character `s[i]`, increase its frequency in `freq`.

If the character is new, initialize its frequency to `1`.

### Step 2: Find the most frequent character count

`max_freq = max(freq.values())`

This gives the frequency of the most common character in the current window.

### Step 3: Calculate the window length

`window_size = i - left + 1`

### Step 4: Check validity

Calculate:

`window_size - max_freq`

If:

`window_size - max_freq > k`

the window needs too many replacements.

### Step 5: Shrink the window

Decrease the frequency of `s[left]` and increment `left`.

Recalculate the window length and continue shrinking while the window is invalid.

### Step 6: Update the answer

`max_window = max(window_size, max_window)`

After processing the entire string, return `max_window`.

---

## 6. Why It Works

Suppose a window contains several different characters.

To make the entire window identical, keep the most frequent character and replace all other characters.

If the window length is `L` and the most frequent character occurs `F` times, the number of replacements required is:

`L - F`

The window can be converted using at most `k` replacements exactly when:

`L - F <= k`

The algorithm expands the window to explore longer substrings and shrinks it whenever the replacement constraint is violated.

It tracks the longest valid window encountered, producing the answer.

## 10. Code

    class Solution(object):
        def characterReplacement(self, s, k):
            freq = {}
            left = 0
            max_window = 0

            for i in range(len(s)):
                if s[i] not in freq:
                    freq[s[i]] = 1
                else:
                    freq[s[i]] += 1

                max_freq = max(freq.values())
                window_size = i - left + 1

                while window_size - max_freq > k:
                    freq[s[left]] -= 1
                    left += 1
                    window_size = i - left + 1

                max_window = max(window_size, max_window)

            return max_window

---

## 11. Code Breakdown

### Step 1: Initialize the state

`freq = {}`

`left = 0`

`max_window = 0`

### Step 2: Add the incoming character

Increase the frequency of `s[i]`.

### Step 3: Find the dominant character frequency

`max_freq = max(freq.values())`

This tells us how many characters can remain unchanged if we convert the rest of the window to the same character.

### Step 4: Calculate required replacements

`window_size - max_freq`

This represents how many characters must be replaced to make the current window uniform, based on the current frequency counts.

### Step 5: Shrink while invalid

If the number of replacements exceeds `k`, remove characters from the left until the condition is satisfied.

### Step 6: Update the maximum

After restoring validity, compare the current window length with `max_window`.

---

## 12. Important Observation

The most frequent character determines how many replacements are needed.

For example:

`Window = "AABAB"`

Frequencies:

`A = 3`

`B = 2`

Window length:

`5`

Required replacements:

`5 - 3 = 2`

Keeping `A` requires replacing the two `B` characters.

This is more efficient than trying every possible target character separately.

**Implementation note:** The code recalculates `max(freq.values())` on each iteration. This is correct, but it can take up to `O(Σ)` time per iteration, where `Σ` is the number of distinct characters in the map. With uppercase English letters, this is bounded by 26.

---

## 13. Edge Cases

### Case 1: Empty string

`s = ""`

Answer: `0`

### Case 2: All characters identical

`s = "AAAA"`

Any window is already uniform.

Answer: `4`

### Case 3: `k = 0`

No replacements are allowed.

The answer is the length of the longest substring containing only one repeated character.

### Case 4: `k` is greater than or equal to the string length

Every character can be converted to the same character.

Answer: `len(s)`

### Case 5: Alternating characters

`s = "ABABAB"`

With `k = 2`, a window of length `5` needs at least two replacements because either character appears at most three times.

The answer is `5`.

### Case 6: Multiple characters tie for highest frequency

Any of the most frequent characters can be chosen as the target.

The formula `window_size - max_freq` still works.

---

## 14. Common Mistakes

### Mistake 1: Using `max_freq` as the replacement count

`max_freq` is the number of characters kept unchanged.

The replacement count is:

`window_size - max_freq`

### Mistake 2: Shrinking when replacements equal `k`

A window is valid when replacements are exactly `k`.

Shrink only when:

`window_size - max_freq > k`

### Mistake 3: Forgetting to update the frequency map

Every incoming and outgoing character must update its frequency.

### Mistake 4: Using the wrong window length

Use:

`i - left + 1`

### Mistake 5: Assuming the target character must be selected explicitly

The most frequent character is implicitly the best target because it minimizes the number of replacements.

### Mistake 6: Confusing this with LC 3

LC 3 requires every character to be distinct.

LC 424 allows repeated characters and limits how many characters must be replaced to make the window uniform.

## 15. Complexity

### Time Complexity

With the given implementation:

`O(n * Σ)`

where `Σ` is the number of distinct characters tracked in `freq`, because `max(freq.values())` scans the frequency values on every iteration.

For uppercase English letters, `Σ <= 26`, so this is effectively:

`O(n)`

### Space Complexity

`O(Σ)`

The frequency map stores counts for the distinct characters encountered.

For uppercase English letters, the map has at most 26 keys, so auxiliary space is `O(1)` under the fixed alphabet constraint.

## 16. Pattern Recognition

**Pattern: Variable-Size Sliding Window + Frequency Map + Replacement Budget**

Recognize this pattern when:

- The problem asks for the longest substring.
- Characters can be replaced or modified.
- There is a maximum number of allowed replacements.
- A frequency map can determine the minimum number of changes needed.

The key formula is:

`replacements_needed = window_size - max_freq`

## 17. Pattern Trigger / Generalization

Ask:

1. Is the problem asking for the longest contiguous substring?
2. Can some characters be changed?
3. Is the number of allowed changes limited?
4. Can the minimum changes be calculated from the most frequent character?

If yes, consider a variable-size sliding window with a replacement budget.

The same broad idea applies to problems where a window can contain at most a certain number of deviations from a desired property.

```python
class Solution(object):
    def characterReplacement(self, s, k):
        """
        :type s: str
        :type k: int
        :rtype: int
        """
        freq={}
        left=0
        max_window=0
        for i in range(len(s)):
            if s[i] not in freq:
                freq[s[i]]=1
            else:
                freq[s[i]]+=1
            max_freq = max(freq.values())
            window_size=i-left+1
            
            while window_size-max_freq>k:
                    
                freq[s[left]]-=1
                left+=1
                window_size=i-left+1
            max_window=max(window_size,max_window)
        return max_window 
                    

```
