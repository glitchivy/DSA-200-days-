# LC 2461: Maximum Sum of Distinct Subarrays With Length K

## 1. Problem

Given an integer array `nums` and an integer `k`, find the maximum sum of a subarray of length exactly `k` such that all elements in that subarray are distinct.

Return `0` if no valid subarray exists.

### Example

Input:

`nums = [1,5,4,2,9,9,9]`

`k = 3`

Output:

`15`

Explanation:

The subarray `[5,4,2]` contains three distinct elements and has sum `11`.

The subarray `[4,2,9]` contains three distinct elements and has sum `15`.

The maximum valid sum is `15`.

---

## 2. Core Idea

This problem combines **Fixed-Size Sliding Window + Frequency Hash Map + Distinctness Check**.

We need to maintain two properties for every window of size `k`:

1. The sum of all elements in the current window.
2. The frequency of every element in the current window.

A window is valid when:

`len(freq) == k`

Why?

A window of length `k` contains exactly `k` elements. If the frequency map contains `k` distinct keys, every element must be distinct.

For each valid window, update the maximum sum.

---

## 3. Pointer / Variable Roles

- `nums`: Input array.
- `k`: Required window size.
- `start`: Slice containing the first `k` elements.
- `sum_num`: Sum of the current window.
- `freq`: Dictionary mapping each element to its frequency in the current window.
- `max_sum`: Maximum sum among valid windows.
- `i`: Index of the incoming element.
- `nums[i-k]`: Element leaving the window.

Important distinction:

- `sum_num` tracks the total of the current window.
- `freq` tracks the number of occurrences of each value.
- `len(freq)` tracks the number of distinct values.

---

## 4. Initialization

### Step 1: Select the first window

`start = nums[:k]`

This creates a slice containing the first `k` elements.

### Step 2: Initialize the state

`sum_num = 0`

`freq = {}`

`max_sum = 0`

### Step 3: Build the first window

For every element in the first `k` positions:

- Add it to `sum_num`.
- Add it to `freq`, or increase its frequency if already present.

### Step 4: Check validity

If `len(freq) == k`, all elements are distinct.

Set:

`max_sum = sum_num`

Otherwise, keep `max_sum = 0`.

---

## 5. Main Decision Logic

Slide the window from left to right.

For every incoming index `i`, starting at `k`:

### Step 1: Update the sum

`sum_num += nums[i] - nums[i-k]`

This adds the incoming element and removes the outgoing element.

### Step 2: Add the incoming element to the frequency map

If `nums[i]` is already in `freq`, increase its frequency.

Otherwise, initialize its frequency to `1`.

### Step 3: Remove the outgoing element

Decrease:

`freq[nums[i-k]] -= 1`

If its frequency becomes zero, delete that key.

### Step 4: Check distinctness

If:

`len(freq) == k`

the current window contains `k` distinct elements.

### Step 5: Update the maximum

`max_sum = max(max_sum, sum_num)`

After processing every window, return `max_sum`.

---

## 6. Why It Works

Every possible subarray of length `k` appears exactly once as the sliding window moves through the array.

The algorithm maintains the correct sum by adding the incoming element and subtracting the outgoing element.

It maintains the frequency map by adding the incoming element and removing the outgoing element's contribution.

A window of length `k` is valid precisely when it has `k` distinct elements.

Therefore, the algorithm checks every valid window and keeps the largest sum.

If no valid window exists, `max_sum` remains `0`, which is the required result.

## 10. Code

    class Solution(object):
        def maximumSubarraySum(self, nums, k):
            start = nums[:k]
            sum_num = 0
            freq = {}

            for i in range(len(start)):
                sum_num += nums[i]

                if nums[i] not in freq:
                    freq[nums[i]] = 1
                else:
                    freq[nums[i]] += 1

            max_sum = 0

            if len(freq) == k:
                max_sum = sum_num

            for i in range(k, len(nums)):
                sum_num += nums[i] - nums[i-k]

                if nums[i] not in freq:
                    freq[nums[i]] = 1
                else:
                    freq[nums[i]] += 1

                freq[nums[i-k]] -= 1

                if freq[nums[i-k]] == 0:
                    del freq[nums[i-k]]

                if len(freq) == k:
                    max_sum = max(max_sum, sum_num)

            return max_sum

## 11. Code Breakdown

### Step 1: Build the first window

`start = nums[:k]`

The slice represents the first window. The loop uses its length to process the first `k` elements.

### Step 2: Track the sum

`sum_num += nums[i]`

Every element in the initial window contributes to the running sum.

### Step 3: Build the frequency map

If the element already exists, increase its frequency. Otherwise, insert it with frequency `1`.

### Step 4: Validate the first window

`len(freq) == k`

If true, every element is distinct, so the initial sum is a valid candidate.

### Step 5: Slide the window

`sum_num += nums[i] - nums[i-k]`

The sum changes in constant time instead of being recalculated from scratch.

### Step 6: Update the incoming frequency

Increase the count of `nums[i]`.

### Step 7: Update the outgoing frequency

Decrease the count of `nums[i-k]`.

Delete the key when its frequency reaches zero. This is essential because `len(freq)` must count only elements currently present in the window.

### Step 8: Update the answer

Only valid windows can update `max_sum`.

---

## 12. Important Observation

**The number of distinct elements equals the window size if and only if all elements in the window are distinct.**

For a window of length `k`:

- `len(freq) == k`: Every element is distinct.
- `len(freq) < k`: At least one element is repeated.

The frequency map must remove keys whose frequency becomes zero. Otherwise, it could count elements that are no longer in the current window.

This is why the deletion step matters:

`if freq[nums[i-k]] == 0: del freq[nums[i-k]]`

---

## 13. Edge Cases

### Case 1: All elements are distinct

Every window is valid. Return the maximum window sum.

### Case 2: All elements are identical

No window of length greater than `1` is valid.

The result is `0`.

### Case 3: No valid window exists

`max_sum` remains `0`, so the algorithm correctly returns `0`.

### Case 4: `k = 1`

Every individual element forms a valid window because a single element is always distinct.

The answer is the maximum element, or `0` if all elements are negative under the problem's nonnegative-input constraints.

### Case 5: The outgoing value appears more than once

Decrease its frequency but do not delete it unless the frequency becomes zero.

### Case 6: Incoming value equals outgoing value

The frequency map first increases the incoming value's count and then decreases the outgoing value's count.

The final frequency remains correct.

### Case 7: `k > len(nums)`

The problem constraints generally prevent this case. Without such constraints, the implementation would need an explicit check before creating the initial window.

---

## 14. Common Mistakes

### Mistake 1: Recalculating the sum for every window

This can lead to `O(nk)` time.

Instead, add the incoming value and subtract the outgoing value.

### Mistake 2: Checking only whether the sum is large

A high-sum window may contain duplicates.

Always check `len(freq) == k`.

### Mistake 3: Keeping zero-frequency keys

A key with frequency zero is no longer present in the window and must be deleted.

### Mistake 4: Decreasing the outgoing frequency before updating the sum

The sum and frequency map are separate state variables. Update both consistently for every window.

### Mistake 5: Updating the maximum for an invalid window

Only update `max_sum` when the window has exactly `k` distinct elements.

### Mistake 6: Confusing frequency with distinct count

`freq[value]` is the number of occurrences of one value.

`len(freq)` is the number of distinct values in the current window.

## 15. Complexity

### Time Complexity

`O(n)` average time.

- Initial window construction: `O(k)`
- Sliding-window traversal: `O(n-k)`
- Dictionary lookups, insertions, updates, and deletions: average `O(1)` each

Overall:

`O(n)`

### Space Complexity

`O(k)` auxiliary space in the worst case for the frequency map.

The `start = nums[:k]` slice also uses `O(k)` space. It is unnecessary because the initial loop can iterate directly over the first `k` elements.

## 16. Pattern Recognition

**Pattern: Fixed-Size Sliding Window + Frequency Hash Map**

Recognize this pattern when a problem asks for:

- A subarray or substring of exactly length `k`.
- A condition involving distinct elements or element frequencies.
- A maximum or minimum sum among valid windows.

Typical structure:

1. Build the initial window.
2. Track the necessary state.
3. Slide by removing one element and adding another.
4. Validate the current window.
5. Update the answer only when the condition holds.

---

## 17. Pattern Trigger / Generalization

Ask these questions:

1. Is the required subarray length fixed?
2. Can the sum be updated by removing one element and adding another?
3. Does validity depend on distinctness or frequency?
4. Can a dictionary track the necessary information?

If yes, use a fixed-size sliding window with a frequency map.

The same structure can be adapted to count valid windows, maximize sums, minimize costs, or enforce frequency constraints.
