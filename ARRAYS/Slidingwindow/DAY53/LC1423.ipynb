# LC 1423: Maximum Points You Can Obtain from Cards

## 1. Problem

Given an integer array `cardPoints` and an integer `k`, choose exactly `k` cards from either the beginning or the end of the array.

Each selected card contributes its points. Return the maximum possible score.

### Example

Input:

`cardPoints = [1,2,3,4,5,6,1]`

`k = 3`

Output:

`12`

Explanation:

Choose the last three cards: `5 + 6 + 1 = 12`.

---

## 2. Core Idea

This problem can be solved using **Fixed-Size Sliding Window + Complementary Subarray**.

Instead of choosing `k` cards from the two ends, consider the cards that remain unselected.

If the array has `n` cards and we choose `k` cards, then the number of unselected cards is:

`window_size = n - k`

The unselected cards must form one contiguous subarray in the middle.

Therefore:

`Maximum Score = Total Sum - Minimum Sum of a Window of Size (n - k)`

This transforms a problem about selecting cards from both ends into a standard fixed-size sliding-window problem.

---

## 3. Pointer / Variable Roles

- `total`: Sum of all card points.
- `n`: Total number of cards.
- `k`: Number of cards to select.
- `n-k`: Size of the subarray left unselected.
- `sum_first`: Sum of the current window of unselected cards.
- `minimum`: Minimum sum found among all windows of size `n-k`.
- `i`: Index used to move the window.

---

## 4. Initialization

### Step 1: Calculate the total

`total = sum(cardPoints)`

This represents the points available from all cards.

### Step 2: Calculate the array length

`n = len(cardPoints)`

### Step 3: Determine the window size

`window_size = n - k`

These are the cards that will not be selected.

### Step 4: Calculate the first window

Sum the first `n-k` elements:

`sum_first = sum(cardPoints[:n-k])`

### Step 5: Initialize the minimum

`minimum = sum_first`

The first window is the initial candidate for the minimum unselected sum.

---

## 5. Main Decision Logic

Slide the window through the array.

For every incoming index `i`:

### Step 1: Add the incoming element

`sum_first += cardPoints[i]`

### Step 2: Remove the outgoing element

`sum_first -= cardPoints[i-(n-k)]`

Combined update:

`sum_first += cardPoints[i] - cardPoints[i-(n-k)]`

### Step 3: Update the minimum

`minimum = min(minimum, sum_first)`

After checking every window, subtract the minimum unselected sum from the total:

`return total - minimum`

---

## 6. Why It Works

Suppose we choose exactly `k` cards from the two ends.

The remaining `n-k` cards form a contiguous middle subarray.

Every card belongs to one of two groups:

1. Selected cards.
2. Unselected cards.

Therefore:

`Selected Sum + Unselected Sum = Total Sum`

Rearranging:

`Selected Sum = Total Sum - Unselected Sum`

To maximize the selected sum, we must minimize the unselected sum.

Since the unselected cards form a window of fixed size `n-k`, we find the minimum-sum window and subtract it from the total.

This produces the maximum possible score.

## 10. Code

    class Solution(object):
        def maxScore(self, cardPoints, k):
            total = sum(cardPoints)
            n = len(cardPoints)
            sum_first = 0

            for i in range(n-k):
                sum_first += cardPoints[i]

            minimum = sum_first

            for i in range(n-k, n):
                sum_first += cardPoints[i] - cardPoints[i-(n-k)]
                minimum = min(minimum, sum_first)

            return total - minimum

---

## 11. Code Breakdown

### Step 1: Calculate the total sum

`total = sum(cardPoints)`

This gives the score of every card combined.

### Step 2: Calculate the first unselected window

The loop from `0` to `n-k-1` calculates the sum of the first `n-k` cards.

### Step 3: Initialize the minimum

`minimum = sum_first`

The first window is the smallest window sum found so far.

### Step 4: Slide the window

The second loop starts at index `n-k`.

For every step:

- Add the incoming card.
- Remove the outgoing card.
- Update the minimum window sum.

### Step 5: Calculate the maximum score

`return total - minimum`

The smallest unselected sum leaves the largest possible score among the selected cards.

---

## 12. Important Observation

The key insight is to **reverse the perspective**.

The direct approach considers how many cards to take from the left and how many to take from the right.

There are `k+1` possible splits:

- Take all `k` from the left.
- Take `k-1` from the left and `1` from the right.
- Continue until all `k` are taken from the right.

Instead, find the minimum-sum contiguous subarray of length `n-k`.

This reduces the problem to fixed-size sliding window and avoids checking every possible split separately.

---

## 13. Edge Cases

### Case 1: `k = 0`

No cards are selected.

The window contains all `n` cards.

`total - minimum = 0`

### Case 2: `k = n`

All cards are selected.

The window size is zero, so the initial implementation's empty-window handling needs special consideration.

For the given implementation, return `total` directly when `k == n`, or use a formulation that handles a zero-sized window.

### Case 3: `k = 1`

Choose the larger of the first and last cards.

The complementary window has size `n-1`, so minimizing the unselected sum is equivalent to selecting the better end card.

### Case 4: All card values are equal

Every valid selection gives the same score:

`k * cardPoints[0]`

### Case 5: Negative values

The complement relationship still holds mathematically. The algorithm can still work with negative values because it minimizes the unselected window sum.

---

## 14. Common Mistakes

### Mistake 1: Sliding a window of size `k`

The window represents the unselected cards, so its size is `n-k`, not `k`.

### Mistake 2: Maximizing the window sum

The window contains cards that are not selected.

We need to minimize its sum.

### Mistake 3: Forgetting the complement relationship

The answer is:

`total - minimum`

Not the minimum window sum itself.

### Mistake 4: Incorrect outgoing index

For a window of size `n-k`, the outgoing index is:

`i - (n-k)`

### Mistake 5: Ignoring `k == n`

A zero-sized window requires special handling in this implementation.

## 15. Complexity

### Time Complexity

`O(n)`

- Calculating the total takes `O(n)`.
- Building the initial window takes `O(n-k)`.
- Sliding through the remaining windows takes `O(k)`.

Overall time complexity is `O(n)`.

### Space Complexity

`O(1)` auxiliary space.

Only a constant number of variables are used, and the input array is not modified.

## 16. Pattern Recognition

**Pattern: Fixed-Size Sliding Window + Complementary Subarray**

Recognize this pattern when:

- You must select elements from the beginning and/or end.
- The elements not selected form one contiguous middle subarray.
- The total sum is easy to calculate.
- Maximizing the selected sum can be transformed into minimizing the remaining sum.

Core transformation:

`Maximum selected sum = Total sum - Minimum remaining sum`

---

## 17. Pattern Trigger / Generalization

When a problem asks you to select elements from both ends, ask:

1. What elements remain unselected?
2. Do the unselected elements form a contiguous subarray?
3. Is their length fixed?
4. Can the answer be expressed as total minus the remaining sum?

If the answers are yes, consider a complementary sliding-window solution.

This transformation is especially useful when selecting from both ends appears more complicated than scanning the middle.

## 20. Final Summary

The main insight is to solve the complementary problem.

Instead of maximizing the score of the `k` selected cards, minimize the sum of the `n-k` cards left behind.

Formula:

`Maximum Score = Total Sum - Minimum Window Sum`

Window size:

`n-k`

Complexity:

- Time: `O(n)`
- Auxiliary space: `O(1)`

Mental model:

**Selecting from both ends can become a minimum-sum window problem when the unselected elements form a fixed-size contiguous subarray.**

```python
class Solution(object):
    def maxScore(self, cardPoints, k):
        """
        :type cardPoints: List[int]
        :type k: int
        :rtype: int
        """
        total=sum(cardPoints)
        n=len(cardPoints)
        sum_first=0
        
        for i in range(n-k):
            sum_first+=cardPoints[i] 
        minimum=sum_first
        for i in range(n-k,n):
            sum_first+= cardPoints[i]-cardPoints[i-(n-k)]
            minimum=min(minimum,sum_first)
        return total-minimum
```
