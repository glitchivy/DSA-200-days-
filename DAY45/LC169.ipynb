# LC 169 - Majority Element

## Problem

Given an array `nums` of size `n`, return the element that appears more than `n/2` times.

The problem guarantees that a majority element always exists.

### Example

`nums = [2,2,1,1,1,2,2]`

Array length:

`n = 7`

A majority element must appear more than:

`7/2 = 3.5`

times.

`2` appears 4 times.

Therefore:

`2` is the majority element.

---

## Brute Force Approach

The simplest approach is to count the frequency of every number.

For each number, count how many times it appears.

If:

`frequency > n/2`

then that number is the majority element.

### Complexity

Time: `O(n²)` if counting each element separately.

Space: `O(1)` if no frequency map is used.

---

## Better Approach

Use a hash map to store frequencies.

For every number:

`frequency[num] += 1`

As soon as a number's frequency becomes greater than `n/2`, it is the majority element.

### Complexity

Time: `O(n)`

Space: `O(n)`

This is better than brute force, but we can do even better in space.

---

## Optimal Approach

Use the **Boyer-Moore Voting Algorithm**.

Maintain two variables:

- `value` → current candidate for majority element
- `counter` → current vote balance for that candidate

Rules:

- If `counter <= 0`, choose the current element as the new candidate.
- If the current element equals the candidate, increase `counter`.
- Otherwise, decrease `counter`.

The key idea is:

> A majority element appears more than all other elements combined, so it cannot be completely cancelled out by different elements.

---

## Key Observation

Think of every element as casting a vote.

If the current element is the candidate:

`+1 vote`

If it is different:

`-1 vote`

Example:

`[2,2,1,1,1,2,2]`

Start:

`candidate = 2`

`counter = 1`

Then:

`2 → +1`

`1 → -1`

Different elements cancel each other.

The majority element has enough occurrences to survive the cancellation.

---

## Why Cancellation Works

Suppose the majority element appears more than `n/2` times.

That means:

`majority_count > non_majority_count`

For example:

`[A,A,A,B,B]`

A possible cancellation is:

`A + B → cancel`

Another:

`A + B → cancel`

One `A` remains.

The majority element cannot be completely eliminated because there are more `A`s than all non-`A`s combined.

This is the fundamental reason Boyer-Moore works.

---

## Algorithm

1. Initialize `counter = 0`.
2. Traverse every element.
3. If `counter <= 0`:
   - make the current element the candidate.
   - set `counter = 1`.
4. Else if the current element equals the candidate:
   - increase `counter`.
5. Otherwise:
   - decrease `counter`.
6. Return the candidate.

---

## Dry Run

Consider:

`nums = [2,2,1,1,1,2,2]`

| Element | Candidate | Counter | Action |
|---|---:|---:|---|
| 2 | 2 | 1 | New candidate |
| 2 | 2 | 2 | Same → +1 |
| 1 | 2 | 1 | Different → -1 |
| 1 | 2 | 0 | Different → -1 |
| 1 | 1 | 1 | Counter reached 0 → new candidate |
| 2 | 1 | 0 | Different → -1 |
| 2 | 2 | 1 | Counter reached 0 → new candidate |

Final candidate:

`2`

Because the problem guarantees a majority element exists, this candidate is the answer.

---

## Primary Pattern

### Boyer-Moore Voting / Cancellation Pattern

The core pattern is:

**Candidate + Balance + Cancellation**

Think:

```text
Same as candidate
        ↓
      +1

Different
        ↓
      -1

Balance reaches 0
        ↓
Choose a new candidate

````python
class Solution(object):
    def majorityElement(self, nums):
        """
        :type nums: List[int]
        :rtype: int
        """
        counter=0 
        for i in nums:
            if counter<=0:
                value=i 
                counter=1
            elif value==i:
                counter+=1
            else:
                counter-=1
        return value
````
