# LC 121 - Best Time to Buy and Sell Stock

## Problem

You are given an array `prices` where:

`prices[i]` = stock price on day `i`.

You can:

- Buy the stock once.
- Sell the stock once.
- You must buy before selling.

Return the maximum profit possible.

If no profit is possible, return `0`.

### Example

`prices = [7,1,5,3,6,4]`

Best transaction:

- Buy at `1`
- Sell at `6`

Profit:

`6 - 1 = 5`

Answer:

`5`

---

## Brute Force Approach

The direct approach is to try every possible pair of days:

1. Choose a buying day.
2. Choose a later selling day.
3. Calculate:

`profit = selling_price - buying_price`

4. Keep the maximum profit.

This checks every possible transaction.

### Complexity

Time: `O(n²)`

Space: `O(1)`

This works, but there is unnecessary repeated work.

---

## Optimal Approach

We only need to know the **cheapest price seen so far**.

While traversing the array:

- `min_buy` = cheapest buying price seen so far.
- For the current price `i`, calculate:

`profit = i - min_buy`

- Update the maximum profit.
- If the current price is cheaper than `min_buy`, update `min_buy`.

The key idea is:

> At every selling day, the best possible buying price is the minimum price that appeared before it.

---

## Key Observation

For every current price, ask:

> "If I sell today, what is the cheapest price I could have bought at before today?"

That is exactly what `min_buy` stores.

Example:

`prices = [7,1,5,3,6,4]`

As we move through the array:

| Price | Cheapest Buy So Far | Profit If Sold Today | Maximum Profit |
|---:|---:|---:|---:|
| 7 | 7 | 0 | 0 |
| 1 | 1 | 0 | 0 |
| 5 | 1 | 4 | 4 |
| 3 | 1 | 2 | 4 |
| 6 | 1 | 5 | 5 |
| 4 | 1 | 3 | 5 |

Answer:

`5`

---

## Important Detail

The buying price must occur **before** the selling price.

Our traversal naturally guarantees this.

When processing the current price:

`min_buy`

only contains prices we have already seen.

So we never accidentally buy in the future.

---

## Algorithm

1. Set `min_buy` to the first price.
2. Set `overall = 0`.
3. Traverse every price.
4. Calculate:

`profit = current_price - min_buy`

5. Update the maximum:

`overall = max(overall, profit)`

6. If the current price is lower than `min_buy`, update:

`min_buy = current_price`

7. Return `overall`.

---

## Dry Run

Consider:

`prices = [7,1,5,3,6,4]`

### Price = 7

`min_buy = 7`

`profit = 7 - 7 = 0`

`overall = 0`

---

### Price = 1

`profit = 1 - 7 = -6`

No profit.

But `1` is cheaper than `7`.

Update:

`min_buy = 1`

`overall = 0`

---

### Price = 5

`profit = 5 - 1 = 4`

Update:

`overall = 4`

---

### Price = 3

`profit = 3 - 1 = 2`

`overall` stays `4`.

---

### Price = 6

`profit = 6 - 1 = 5`

Update:

`overall = 5`

---

### Price = 4

`profit = 4 - 1 = 3`

`overall` stays `5`.

Final answer:

`5`

---

## Why Negative Profit Is Not a Problem

Suppose:

`prices = [7,6,4,3,1]`

Every possible transaction loses money.

For example:

`6 - 7 = -1`

`4 - 6 = -2`

`3 - 4 = -1`

`1 - 3 = -2`

But the problem allows us to **not make a transaction**.

Therefore the answer should be:

`0`

That is why:

`overall = 0`

is initialized instead of using a negative value.

## Primary Pattern

### Running Minimum + Current Difference + Global Maximum

This is the main pattern.

The structure is:

`Maintain minimum seen so far`

↓

`Calculate current value - minimum`

↓

`Update maximum result`

In general:

> **When you need the maximum difference between a later value and an earlier value, maintain the minimum earlier value while traversing.**

---

## Pattern Recognition

When you see problems involving:

- buy and sell
- maximum profit
- maximum difference
- later value minus earlier value
- choose an earlier minimum and a later maximum

ask:

1. What value should I remember from the past?
2. Do I need the minimum or maximum previous value?
3. Can I calculate the answer using the current value and that stored value?

For LC 121:

`current price - minimum previous price`

---

## Generalization

This pattern goes beyond stocks.

Suppose you have:

`arr = [8,3,6,2,9]`

and want the maximum:

`later value - earlier value`

Maintain:

`minimum value seen so far`

Then for every current value:

`current - minimum`

and keep the largest result.

This gives the same pattern:

**Running Minimum → Difference → Global Maximum**

---

## Common Mistakes

### 1. Finding the global minimum first

A tempting approach is:

`minimum = min(prices)`

and then finding a maximum after it.

But this can fail because the minimum might occur **after** the selling price.

Example:

`[10,8,6,4,2]`

The global minimum is `2`, but you cannot buy at `2` and sell at an earlier price.

The solution must respect chronological order.

---

### 2. Finding the global maximum first

Same problem.

The maximum might occur before the minimum.

The transaction must follow:

`buy → sell`

not:

`sell → buy`

---

### 3. Returning a negative profit

If prices continuously decrease, you should return `0`.

You are allowed to make no transaction.

---

### 4. Updating the minimum too late

The current price can become the new minimum for **future days**.

That is why we continuously update `min_buy`.

---

## Complexity

Let `n` be the number of prices.

### Time Complexity

We traverse the array once:

`O(n)`

### Space Complexity

Only two variables are required:

- `min_buy`
- `overall`

Therefore:

`O(1)`

This is optimal.

---

## Interview Follow-Ups

### Q1. Why don't we need two loops?

Because we don't need to test every possible buy day.

For every selling day, only the **cheapest previous price** can produce the maximum profit.

So we compress all previous buying possibilities into one variable:

`min_buy`

---

### Q2. Why does the algorithm guarantee buy before sell?

Because `min_buy` represents only prices encountered earlier during the traversal.

The current price is always a later day.

---

### Q3. What if the prices only decrease?

Example:

`[7,6,5,4,3]`

No profitable transaction exists.

The algorithm keeps:

`overall = 0`

Answer:

`0`

---

### Q4. What if there is only one price?

There is no possible buy-and-sell transaction.

The answer is `0`.

```python
class Solution(object):
    def maxProfit(self, prices):
        """
        :type prices: List[int]
        :rtype: int
        """
        min_buy=prices[0]
        overall=0
        for i in prices:
            profit=i-min_buy
            if min_buy>i:
                min_buy=i
            overall=max(profit,overall)
        return overall
```
