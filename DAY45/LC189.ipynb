# LC 189 - Rotate Array

## Problem

Given an integer array `nums`, rotate the array to the right by `k` steps.

The rotation must be done **in-place**.

### Example

Input:

`nums = [1,2,3,4,5,6,7]`

`k = 3`

Output:

`[5,6,7,1,2,3,4]`

Each rotation moves the last element to the front.

After 3 rotations:

`[5,6,7,1,2,3,4]`

---

## Brute Force Approach

One simple approach is to rotate the array one position at a time.

For each rotation:

1. Save the last element.
2. Shift every other element one position to the right.
3. Put the saved element at index `0`.

Doing this `k` times gives:

**O(n × k)** time.

This can become expensive when `k` is large.

---

## Better Approach

Another approach is to create a new array and place each element directly at its rotated position.

The new position of an element at index `i` is:

`(i + k) % n`

This gives:

**O(n)** time

but requires:

**O(n)** extra space.

The problem specifically asks for an in-place solution, so we can do better in space.

---

## Optimal Approach

Use the **Reversal Algorithm**.

The idea is:

1. Reverse the entire array.
2. Reverse the first `k` elements.
3. Reverse the remaining `n-k` elements.

This produces the required rotation without using another array.

---

## Key Observation

Suppose:

`nums = [1,2,3,4,5,6,7]`

and:

`k = 3`

The elements that need to move to the front are:

`[5,6,7]`

The remaining elements are:

`[1,2,3,4]`

So the final structure is:

`[5,6,7] + [1,2,3,4]`

The reversal technique rearranges these two groups using only swaps.

---

## Why Three Reversals Work

Start:

`[1,2,3,4,5,6,7]`

### Step 1: Reverse the entire array

`[7,6,5,4,3,2,1]`

### Step 2: Reverse the first `k = 3` elements

`[5,6,7,4,3,2,1]`

### Step 3: Reverse the remaining elements

`[5,6,7,1,2,3,4]`

Final answer:

`[5,6,7,1,2,3,4]`

---

## Algorithm

1. Let `n = len(nums)`.
2. Normalize `k` using:

`k = k % n`

This is important because rotating by `n` positions produces the original array.

3. Reverse the entire array.
4. Reverse the first `k` elements.
5. Reverse the remaining `n-k` elements.

---

## Dry Run

Consider:

`nums = [1,2,3,4,5,6,7]`

`k = 3`

### Initial

`[1,2,3,4,5,6,7]`

### Reverse entire array

`[7,6,5,4,3,2,1]`

### Reverse first 3

`[5,6,7,4,3,2,1]`

### Reverse remaining 4

`[5,6,7,1,2,3,4]`

Final:

`[5,6,7,1,2,3,4]`

## Primary Pattern

### Array Reversal / In-Place Transformation

The main pattern is:

**Reverse → Reverse Part → Reverse Part**

This is a very important array manipulation technique.

Whenever an array needs to be rearranged **in-place**, ask:

> Can the desired arrangement be achieved using reversals?

---

## Pattern Recognition

When you see:

- rotate array
- shift elements
- move a suffix to the front
- move a prefix to the back
- rearrange an array in-place
- do not use extra space

think about:

**Reversal**

For right rotation by `k`:

`[A | B]`

where:

`A = first n-k elements`

`B = last k elements`

The desired result is:

`[B | A]`

The reversal algorithm transforms:

`[A | B]`

into:

`reverse(A | B)`

then reverses the appropriate sections to restore the internal order.

---

## Generalization

The same technique can be used for **left rotation**.

For left rotation by `k`:

`[A | B]`

where:

`A = first k elements`

`B = remaining elements`

Desired:

`[B | A]`

The reversal operations can be adapted accordingly.

The larger pattern is:

> **Use reversals to rearrange contiguous sections without allocating another array.**

---

## Common Mistakes

### 1. Forgetting `k % n`

If:

`n = 7`

and:

`k = 10`

rotating 10 times is equivalent to rotating:

`10 % 7 = 3`

times.

So the clean approach is:

`k = k % n`

Your solution handles `k > n`, which works, but directly doing modulo is simpler and more robust.

---

### 2. Reversing the wrong sections

For right rotation:

1. Reverse everything.
2. Reverse first `k`.
3. Reverse from `k` to `n-1`.

The boundaries matter.

---

### 3. Using `right = k`

The first section ends at:

`k - 1`

because array indices are zero-based.

So:

`left = 0`

`right = k - 1`

---

### 4. Forgetting the problem requires in-place modification

Creating another array is easier, but the problem specifically asks:

**Modify `nums` in-place.**

The reversal solution uses:

**O(1) auxiliary space.**

---

## Complexity

Let `n` be the length of the array.

### Time Complexity

We perform three reversals.

Each reversal takes at most `O(n)`.

Therefore:

**O(n) time**

Three passes still simplify to:

`O(n)`

---

### Space Complexity

Only a few pointer variables are used.

**O(1) auxiliary space**

This is optimal for the in-place requirement.

---

## Interview Follow-Ups

### Q1. Why use `k % n`?

Because rotating by `n` positions returns the array to its original state.

Therefore:

`k` rotations

is equivalent to:

`k % n` rotations.

---

### Q2. What happens when `k = n`?

The array remains unchanged.

For example:

`[1,2,3]`

rotated by `3`:

`[1,2,3]`

This is why:

`k % n = 0`

is useful.

---

### Q3. Why not repeatedly shift the array?

Repeated shifting takes:

`O(n × k)`

while the reversal technique takes:

`O(n)`.

---

### Q4. Why is the reversal approach considered optimal?

The problem requires modifying the array in-place.

We achieve:

**O(n) time + O(1) auxiliary space**

which is the desired complexity.
