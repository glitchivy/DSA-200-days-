# LC 53 - Maximum Subarray

## Problem

Given an integer array `nums`, find the **contiguous subarray** with the largest sum and return that sum.

A subarray must contain consecutive elements.

### Example

`nums = [-2,1,-3,4,-1,2,1,-5,4]`

The maximum-sum subarray is:

`[4,-1,2,1]`

Its sum is:

`4 + (-1) + 2 + 1 = 6`

Answer:

`6`

---

## Brute Force Approach

The direct approach is:

1. Choose a starting index.
2. Choose an ending index.
3. Calculate the sum of that subarray.
4. Keep the maximum sum.

If we calculate every subarray sum from scratch, the complexity becomes:

**O(n³)**

We can improve this using a running sum:

1. Fix a starting index.
2. Expand the ending index.
3. Maintain the current sum.

This gives:

**O(n²)** time and **O(1)** space.

But we can do even better.

---

## Optimal Approach

Use **Kadane's Algorithm**.

Maintain two variables:

- `current_sum` = maximum subarray sum ending at the current position
- `max_sum` = maximum subarray sum found anywhere so far

For every new element, we have two choices:

1. Extend the previous subarray.
2. Start a new subarray from the current element.

So:

`current_sum = max(current_sum + nums[i], nums[i])`

Then:

`max_sum = max(max_sum, current_sum)`

---

## Key Observation

The most important question is:

> **Should I continue the previous subarray or start a new one?**

Suppose:

`current_sum = -5`

and:

`nums[i] = 4`

Continuing gives:

`-5 + 4 = -1`

Starting fresh gives:

`4`

Clearly, `4` is better.

So:

**If the previous sum hurts the current element, discard it.**

This gives the central Kadane's Algorithm decision:

`extend previous subarray`

OR

`start new subarray`

Take whichever is larger.

---

## Algorithm

1. Initialize `current_sum` with the first element.
2. Initialize `max_sum` with the first element.
3. Traverse from the second element onward.
4. For each element:
   - Decide whether to extend the current subarray or start a new one.
   - Store the better choice in `current_sum`.
   - Update `max_sum`.
5. Return `max_sum`.

---

## Primary Pattern

### Kadane's Algorithm

The core pattern is:

**Running Best Ending Here + Global Best**

Think:

`current_sum`

= best subarray sum that **must end at the current index**

Then:

`max_sum`

= best subarray sum found anywhere so far.

The key decision is:

`Continue`

OR

`Restart`

---

## Pattern Recognition

When you see:

- maximum subarray sum
- contiguous subarray
- maximum sum of consecutive elements
- largest sum over a continuous segment

think:

> **Can I maintain the best subarray ending at the current position?**

If yes, Kadane's Algorithm is a major candidate.

The trigger question is:

> **Is the previous running sum helping or hurting the current element?**

If it is hurting, start fresh.

---

## Generalization

Kadane's core idea can be expressed as:

`best ending at i = max(best ending at i-1 + nums[i], nums[i])`

This is a form of **dynamic programming**.

At every index, we only need the previous state:

`previous best ending here`

So we don't need to store a DP array.

This reduces space from:

`O(n)`

to:

`O(1)`.

---

## Why We Need Both Variables

### `current_sum`

Answers:

> What is the best subarray sum that ends exactly here?

### `max_sum`

Answers:

> What is the best subarray sum I have seen anywhere?

They are different.

Example:

`[5,-10,4]`

After processing `4`:

`current_sum = 4`

But the overall best is still:

`max_sum = 5`

So we need both.

---

## Common Mistakes

### 1. Resetting whenever the current element is negative

A negative number does not automatically mean the subarray should restart.

Example:

`[4,-1,2]`

The `-1` is negative, but keeping it gives:

`4 + (-1) + 2 = 5`

which is better than starting at `2`.

The correct question is:

> Does the entire previous sum hurt the current element?

Not:

> Is the current element negative?

---

### 2. Starting `max_sum` at 0

This fails when all numbers are negative.

Example:

`[-5,-2,-8]`

The correct answer is:

`-2`

If we initialize `max_sum = 0`, we might incorrectly return `0`.

That is why your initialization:

`current_sum = nums[0]`

`max_sum = nums[0]`

is important.

---

### 3. Forgetting that the subarray must be contiguous

For:

`[1,-5,3]`

You cannot choose `[1,3]`.

That would be a subsequence, not a subarray.

---

### 4. Using nested loops unnecessarily

A brute-force solution can use two loops, but Kadane's Algorithm reduces the problem to one traversal.

---

## Complexity

Let `n` be the length of `nums`.

### Time Complexity

Every element is processed once.

**O(n)**

### Space Complexity

Only two variables are maintained.

**O(1)**

This is optimal.

## Interview Follow-Ups

### Q1. Why can we discard the previous subarray?

Suppose:

`current_sum < 0`

and the next element is `x`.

Continuing gives:

`current_sum + x`

Since `current_sum` is negative:

`current_sum + x < x`

Therefore, starting fresh at `x` is always better.

This is the mathematical reason behind Kadane's restart decision.

---

### Q2. Why does Kadane work with all-negative arrays?

Because we initialize using the first element instead of `0`.

For:

`[-3,-1,-2]`

the algorithm produces:

`-1`

which is the correct maximum subarray sum.

---

### Q3. Can we return the actual subarray instead of only the sum?

Yes.

We would additionally track:

- current starting index
- best starting index
- best ending index

The Kadane logic remains the same.

---

### Q4. Is Kadane dynamic programming?

Yes.

The DP state can be defined as:

`dp[i] = maximum subarray sum ending at index i`

Transition:

`dp[i] = max(dp[i-1] + nums[i], nums[i])`

Since we only need `dp[i-1]`, we can compress the DP array into one variable:

`current_sum`

**Primary Pattern:** Kadane's Algorithm - Running Best + Continue/Restart + Global Maximum

```python
class Solution(object):
    def maxSubArray(self, nums):
        """
        :type nums: List[int]
        :rtype: int
        """
        current_sum=nums[0]
        max_sum=nums[0]
        for i in range(1,len(nums)):
            current_sum=max(current_sum+nums[i],nums[i])
            max_sum=max(max_sum,current_sum)

        return max_sum
```
