# LeetCode 1482 - Minimum Number of Days to Make m Bouquets

## Problem

You are given:

- `bloomDay[i]` = the day on which the `i-th` flower blooms.
- `m` = number of bouquets required.
- `k` = number of **adjacent** flowers needed for one bouquet.

Return the **minimum number of days** needed to make `m` bouquets.

If it is impossible, return

```
-1
```

---

# Brute Force Approach

## Idea

Try every possible day from

```
min(bloomDay)
```

to

```
max(bloomDay)
```

For every day,

check whether it is possible to make

```
m
```

bouquets.

Return the first valid day.

---

## Time Complexity

```
O((maxDay-minDay) × n)
```

---

## Space Complexity

```
O(1)
```

---

# Optimal Approach (Binary Search on Answer)

## Key Observation

We are **not searching an array**.

We are searching the answer.

The answer is

```
Minimum Day
```

Possible answers lie between

```python
min(bloomDay)
```

and

```python
max(bloomDay)
```

---

## Edge Case

Before Binary Search,

check whether enough flowers exist.

Flowers required

```python
m * k
```

Flowers available

```python
len(bloomDay)
```

If

```python
m * k > len(bloomDay)
```

return

```python
-1
```

because making the required bouquets is impossible.

---

## Search Space

```python
low = min(bloomDay)
high = max(bloomDay)
```

---

# Algorithm

## Step 1

Initialize

```python
low = min(bloomDay)
high = max(bloomDay)
```

---

## Step 2

Repeat while

```python
low < high
```

Find

```python
mid = (low + high) // 2
```

Here,

```
mid
```

represents the candidate day.

Question being asked:

> Can we make `m` bouquets by day `mid`?

---

## Step 3

Initialize

```python
count = 0
bouquets = 0
```

where

- `count` = consecutive bloomed flowers
- `bouquets` = total bouquets formed

---

## Step 4

Traverse every flower.

If the flower has bloomed,

```python
day <= mid
```

increase

```python
count += 1
```

Otherwise,

the consecutive sequence breaks.

```python
count = 0
```

---

## Step 5

Whenever

```python
count == k
```

one bouquet is completed.

```python
bouquets += 1
count = 0
```

Resetting `count` ensures flowers are **not reused** in another bouquet.

---

## Step 6

After scanning the array,

check

```python
bouquets >= m
```

### If True

Current day works.

Try finding an even smaller day.

```python
high = mid
```

---

### Otherwise

Not enough bouquets.

Need more days.

```python
low = mid + 1
```

---

## Step 7

When

```python
low == high
```

that is the minimum valid day.

Return

```python
low
```

---

# Key Observations

- We search over **days**, not indices.
- More days mean more flowers have bloomed.
- If a day works, every larger day also works.

```
❌ ❌ ❌ ✅ ✅ ✅
```

We need the **first** valid day.

- Flowers in a bouquet must be **adjacent**.
- Flowers cannot be reused.

---

# Common Mistakes

- Forgetting the edge case

```python
m * k > len(bloomDay)
```

- Using

```python
low = 1
```

instead of

```python
low = min(bloomDay)
```

- Forgetting to reset `count` when an unbloomed flower appears.

- Forgetting to reset `count` after forming one bouquet.

- Using

```python
high = mid - 1
```

instead of

```python
high = mid
```

when using the `while low < high` template.

- Returning `mid` instead of `low`.

# Interview Takeaway

This problem is another classic **Binary Search on Answer** problem. We search over the possible number of days rather than array indices. For each candidate day, we simulate the blooming process, count consecutive flowers, and determine whether at least `m` bouquets can be formed. Since any day that works guarantees all later days also work, the answers form a monotonic pattern, allowing an efficient **O(n × log(maxDay-minDay))** solution.

````python
class Solution(object):
    def minDays(self, bloomDay, m, k):
        """
        :type bloomDay: List[int]
        :type m: int
        :type k: int
        :rtype: int
        """
        low=min(bloomDay)
        high=max(bloomDay)
        if m*k>len(bloomDay):
            return -1
        while low<high:
            mid=(low+high)//2
            count=0
            bou=0
            for i in bloomDay:
                if i<=mid:
                    count+=1
                else:
                    count=0
                if k==count:
                    bou+=1
                    count=0
            if bou>=m:
                high=mid
            elif low==high:
                return low
            else:
                low=mid+1
    
````
