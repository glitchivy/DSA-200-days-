# LeetCode 35 - Search Insert Position

## Problem

Given a sorted array of **distinct integers** `nums` and an integer `target`, return:

- The index if the target exists.
- Otherwise, return the index where it should be inserted to maintain the sorted order.

---

## Brute Force Approach

### Idea

Traverse the array from left to right.

- If the target is found, return its index.
- If an element greater than the target is encountered, return its index as the insertion position.
- If the entire array is traversed, insert the target at the end.

### Algorithm

1. Traverse the array.
2. If `nums[i] == target`, return `i`.
3. If `nums[i] > target`, return `i`.
4. If no such element exists, return `n`.

### Time Complexity

**O(n)**

### Space Complexity

**O(1)**

---

## Optimal Approach (Binary Search)

### Key Observation

The array is sorted.

Instead of searching linearly, use Binary Search to determine where the target belongs.

At the end of Binary Search:

- `high` points to the last element smaller than the target.
- `low` points to the first element greater than or equal to the target.

The insertion position is always **`low`**.

This is exactly the concept of a **Lower Bound**.

---

## Algorithm

1. Initialize:

```
low = 0
high = n - 1
```

2. While:

```
low <= high
```

3. Compute:

```
mid = (low + high) // 2
```

4. If the target is found:

```
return mid
```

5. If:

```
target > nums[mid]
```

Search the right half.

```
low = mid + 1
```

6. Otherwise search the left half.

```
high = mid - 1
```

7. When the loop ends, return:

```
low
```

---

## Dry Run

### Input

```
nums = [1,3,5,6]

target = 2
```

### Iteration 1

```
low = 0

high = 3

mid = 1

nums[mid] = 3
```

Since:

```
2 < 3
```

Move left.

```
high = 0
```

---

### Iteration 2

```
low = 0

high = 0

mid = 0

nums[mid] = 1
```

Since:

```
2 > 1
```

Move right.

```
low = 1
```

---

Loop ends.

```
low = 1

high = 0
```

Return:

```
1
```

Insert the target at index `1`.

---

## Why do we return `low`?

When Binary Search finishes:

- All elements before `low` are smaller than the target.
- `low` is the first position where the target can be placed while maintaining sorted order.

Equivalently:

```
low = first index where nums[i] >= target
```

This is the **Lower Bound** of the target.

---

Each iteration removes half of the remaining search space.

```

↓

```

Number of iterations:

**O(log n)**

**O(1)**

## Common Mistakes

- Returning `high` instead of `low`.

- Returning the last computed `mid`.

- Using:

```
while low < high
```

instead of:

```
while low <= high
```

- Forgetting that when the target is larger than every element, the answer is:

```
len(nums)
```

- Moving the pointers incorrectly, causing an infinite loop.

---

## Interview Takeaway

Search Insert Position is a direct application of the **Lower Bound** concept. The goal is to find the first position where the target can exist without violating the sorted order. Binary Search naturally leads to this answer because, after the search ends, `low` always points to the correct insertion index. Understanding this idea is the foundation for problems involving Lower Bound, Upper Bound, First/Last Occurrence, and many Binary Search on Answer problems.

````python
class Solution(object):
    def searchInsert(self, nums, target):
        """
        :type nums: List[int]
        :type target: int
        :rtype: int
        """
        low=0
        high=len(nums)-1
        while low<=high:
            mid=(low+high)//2
            if target==nums[mid]:
                return mid
            elif target>=nums[mid]:
                low=mid+1
            else:
                high=mid-1
        return low
````
