# LeetCode 34 - Find First and Last Position of Element in Sorted Array

## Problem

Given a sorted array of integers `nums` and an integer `target`, return the starting and ending positions of the target value.

If the target is not present, return:

```
[-1, -1]
```

The solution must run in **O(log n)** time.

---

## Brute Force Approach

### Idea

Traverse the array from left to right.

- Record the first occurrence of the target.
- Continue traversing until the last occurrence is found.
- If the target never appears, return `[-1, -1]`.

### Algorithm

1. Initialize first and last as `-1`.
2. Traverse the array.
3. When the target is found for the first time, store its index.
4. Keep updating the last occurrence whenever the target is found.
5. Return `[first, last]`.

### Time Complexity

**O(n)**

### Space Complexity

**O(1)**

---

## Optimal Approach (Lower Bound + Upper Bound)

### Key Observation

The array is sorted.

Instead of searching every occurrence, use Binary Search twice.

- **Lower Bound** gives the first index where:

```
nums[i] >= target
```

- **Upper Bound** gives the first index where:

```
nums[i] > target
```

Therefore,

```
First Occurrence = Lower Bound
Last Occurrence = Upper Bound - 1
```

---

## Algorithm

### Step 1

Find the **Lower Bound**.

It returns the first index where:

```
nums[i] >= target
```

---

### Step 2

Verify whether the target actually exists.

If

```
left == len(nums)
```

or

```
nums[left] != target
```

then return

```
[-1, -1]
```

---

### Step 3

Find the **Upper Bound**.

It returns the first index where:

```
nums[i] > target
```

---

### Step 4

Compute

```
right = upperBound - 1
```

Return

```
[left, right]
```

---

## Dry Run

### Input

```
nums = [5,7,7,8,8,10]

```

### Lower Bound

Find first element:

```

```

```

### Upper Bound

Find first element:

```

```

```
Index : 0 1 2 3 4 5

                  ^
```

```

```

Therefore,

```

```

```

---

## Why do we check

```
left == len(nums) or nums[left] != target
```

Lower Bound always returns the first element greater than or equal to the target.

However, the target may not actually exist.

Example:

```
nums = [5,7,8]

target = 6
```

Lower Bound returns:

```
1
```

because

```
nums[1] = 7
```

Although `7 >= 6`, it is **not** the target.

Hence this verification step is necessary.

## Common Mistakes

- Returning immediately when the target is found.

- Confusing Lower Bound (`>=`) with Upper Bound (`>`).

- Forgetting to subtract one from the Upper Bound.

- Forgetting to verify whether the target actually exists.

- Returning `[left, upper]` instead of `[left, upper - 1]`.

---

## Interview Takeaway

This problem is a direct application of **Lower Bound** and **Upper Bound**. Instead of searching for every occurrence, Binary Search is used twice. Lower Bound identifies the first occurrence, while Upper Bound identifies the position immediately after the last occurrence. Together they provide the required range in **O(log n)** time, making this a fundamental Binary Search pattern used in many interview problems.

````python
class Solution(object):
    def searchRange(self, nums, target):
        """
        :type nums: List[int]
        :type target: int
        :rtype: List[int]
        """
        def lower():
            low=0
            high=len(nums)-1
            ans=len(nums)
            while low<=high:
                mid=(low+high)//2
                if nums[mid]>=target:
                    ans= mid
                    high=mid-1
                else:
                    low=mid+1
            return ans
        def upper():
            low=0
            high=len(nums)-1
            ans=len(nums)
            while low<=high:
                mid=(low+high)//2
                if nums[mid]>target:
                    ans=mid
                    high=mid-1
                else:
                    low=mid+1
            return ans

        left=lower()
        if left==len(nums) or nums[left]!=target:
            return [-1,-1]
        right=upper()-1
        return [left,right]
````
