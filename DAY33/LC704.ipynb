# LeetCode 704 - Binary Search

## Problem

Given a **sorted** integer array `nums` and an integer `target`, return the **index** of `target` if it exists.

If the target is not present, return:

```
-1
```

---

## Brute Force Approach

### Idea

Traverse the array from left to right.

If an element equals the target, return its index.

If the traversal finishes without finding the target, return `-1`.

### Algorithm

1. Iterate through every element.
2. Compare each element with the target.
3. If found, return its index.
4. Otherwise, return `-1`.

### Time Complexity

**O(n)**

### Space Complexity

**O(1)**

---

## Optimal Approach (Binary Search)

### Key Observation

The array is **already sorted**.

Instead of checking every element, repeatedly eliminate **half of the search space**.

At every step:

- Calculate the middle element.
- Compare it with the target.
- Discard the half that cannot contain the target.

This reduces the search space by half after every comparison.

---

## Algorithm

1. Initialize:

```
low = 0
high = n - 1
```

2. While:

```
low <= high
```

3. Find the middle element:

```
mid = (low + high) // 2
```

4. If:

```
nums[mid] == target
```

Return `mid`.

5. If:

```
nums[mid] < target
```

Search the right half.

```
low = mid + 1
```

6. Otherwise search the left half.

```
high = mid - 1
```

7. If the loop finishes, return:

```
-1
```

---

## Dry Run

### Input

```
nums = [-1,0,3,5,9,12]

target = 9
```

### Iteration 1

```
low = 0

high = 5

mid = 2

nums[mid] = 3
```

Since:

```
3 < 9
```

Move right.

```
low = 3
```

---

### Iteration 2

```
low = 3

high = 5

mid = 4

nums[mid] = 9
```

Target found.

Return:

```
4
```

---

## Why do we use `low <= high`?

Suppose:

```
low = 4

high = 4
```

There is still **one element left** to check.

If we used:

```
low < high
```

the loop would terminate before checking the last remaining element.

Using:

```
low <= high
```

ensures every possible element is examined.

---

## Time Complexity

At every iteration, half of the search space is discarded.

```
n
↓

n/2
↓

n/4
↓

n/8
↓

...
```

Number of iterations:

**O(log n)**

---

## Space Complexity

Only three variables are used.

**O(1)**

## Common Mistakes

- Using:

```
while low < high
```

instead of:

```
while low <= high
```

- Forgetting to return `-1` when the target is not found.

- Updating:

```
low = mid
```

instead of:

```
low = mid + 1
```

which can cause an infinite loop.

- Updating:

```
high = mid
```

instead of:

```
high = mid - 1
```

- Calculating the middle index incorrectly.

---

## Interview Takeaway

Binary Search works because the array is sorted. At every iteration, compare the middle element with the target and discard the half that cannot contain the answer. This reduces the search space by half every time, giving an efficient **O(log n)** solution. This template forms the foundation for many advanced Binary Search problems such as Search Insert Position, Rotated Sorted Array, First and Last Position, Koko Eating Bananas, and Binary Search on Answer.

````python
class Solution(object):
    def search(self, nums, target):
        """
        :type nums: List[int]
        :type target: int
        :rtype: int
        """
        low=0
        n=len(nums)
        high=n-1
        while low<=high:
            mid=(low+high)//2
            if target == nums[mid]:
                return mid 
            elif target >nums[mid]:
                low=mid+1
            else:
                high=mid-1
        return -1
````
