# LeetCode 540 - Single Element in a Sorted Array

## Problem

You are given a **sorted array** where:

- Every element appears **exactly twice**
- Except **one element**, which appears only once.

Return the single element.

The solution must run in:

```
O(log n)
```

---

## Brute Force Approach

### Idea

Traverse the array and count the frequency of every element using a HashMap.

The element with frequency `1` is the answer.

### Time Complexity

```
O(n)
```

### Space Complexity

```
O(n)
```

---

## Better Approach (XOR)

### Key Observation

```
a ^ a = 0

a ^ 0 = a
```

Every duplicate pair cancels out.

Example

```
1 ^ 1 ^ 2 ^ 3 ^ 3 ^ 4 ^ 4

= 2
```

### Algorithm

1. Initialize

```
xor = 0
```

2. Traverse the array.

3. XOR every element.

4. Return the final XOR value.

### Time Complexity

```
O(n)
```

### Space Complexity

```
O(1)
```

---

# Optimal Approach (Binary Search)

## Key Observation

The array is sorted.

Before the single element,

every pair starts at an **even index**.

Example

```
Index : 0 1 2 3 4 5 6

Value : 1 1 2 2 3 4 4
```

Pairs

```
1 1

0 1
```

Starts at even.

```
2 2

2 3
```

Starts at even.

---

After the single element,

all pairs shift by one position.

```
Index : 0 1 2 3 4 5 6 7 8

Value : 1 1 2 3 3 4 4 8 8
```

Pairs

```
3 3

3 4
```

Starts at odd.

```
4 4

5 6
```

Starts at odd.

This change in parity is the key idea behind Binary Search.

---

## Algorithm

### Step 1

Initialize

```python
low = 0
high = len(nums) - 1
```

---

### Step 2

Repeat while

```python
low < high
```

Find

```python
mid = (low + high) // 2
```

---

### Step 3

Ensure `mid` is even.

If

```python
mid % 2 == 1
```

then

```python
mid -= 1
```

Now `mid` always points to the **first element of a pair**.

---

### Step 4

Compare

```python
nums[mid]

nums[mid+1]
```

---

### Case 1

If

```python
nums[mid] == nums[mid+1]
```

The pair is valid.

Everything before this pair follows the correct pattern.

Therefore,

the single element must lie on the **right**.

Move

```python
low = mid + 2
```

---

### Case 2

Otherwise,

the pair is broken.

The single element is at `mid` or before it.

Move

```python
high = mid
```

---

### Step 5

When

```python
low == high
```

Return

```python
nums[low]
```

# Why does Binary Search work?

Before the single element

```
Pair starts at

Even index
```

After the single element

```
Pair starts at

Odd index
```

Binary Search identifies the first location where this pairing pattern breaks.

# Key Observations

- Before the single element, every pair starts at an even index.
- After the single element, every pair starts at an odd index.
- Always make `mid` even before comparing pairs.
- If the pair is correct, search the right half.
- If the pair is broken, search the left half including `mid`.

---

# Common Mistakes

- Forgetting to make `mid` even.

- Using

```python
high = mid - 1
```

instead of

```python
high = mid
```

which may skip the answer.

- Using

```python
low <= high
```

instead of

```python
low < high
```

- Returning `nums[mid]` instead of `nums[low]`.

---

# Interview Takeaway

This problem is solved by observing the pairing pattern in a sorted array. Before the single element, every duplicate pair begins at an even index. After the single element, the pairs shift and begin at odd indices. By always making `mid` even and checking whether `nums[mid]` matches `nums[mid + 1]`, Binary Search can determine which side still contains the unique element, leading to an **O(log n)** solution.

````python
class Solution(object):
    def singleNonDuplicate(self, nums):
        """
        :type nums: List[int]
        :rtype: int
        """
        low=0
        high=len(nums)-1
        while low<high:
            mid=(low+high)//2
            if mid%2==1:
                mid-=1
            if nums[mid]==nums[mid+1]:
                low=mid+2
            else:
                high=mid
        return nums[low]
````
