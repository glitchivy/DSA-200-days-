# LeetCode 875 - Koko Eating Bananas

## Problem

Koko has `n` piles of bananas.

The `i-th` pile contains

```
piles[i]
```

bananas.

Koko eats bananas at a constant speed of

```
k bananas/hour
```

Rules:

- Every hour, she chooses one pile.
- She eats at most `k` bananas from that pile.
- If the pile has fewer than `k` bananas, she finishes the pile and waits until the next hour.

Given

```
piles
```

and

```
h
```

return the **minimum eating speed `k`** so that Koko finishes within `h` hours.

---

# Brute Force Approach

## Idea

Try every possible eating speed from

```
1
```

to

```
max(piles)
```

For each speed,

calculate the total hours required.

Return the first speed that satisfies

```
hours <= h
```

---

## Time Complexity

```
O(n × max(piles))
```

---

## Space Complexity

```
O(1)
```

---

# Optimal Approach (Binary Search on Answer)

## Key Observation

We are **not searching an array**.

We are searching the answer.

The answer is the **minimum eating speed**.

Possible speeds are

```
1

2

3

...

max(piles)
```

---

### Monotonic Property

As eating speed increases,

required hours decrease.

Example

```
Speed

1 2 3 4 5 6 7
```

```
Works?

❌ ❌ ❌ ✅ ✅ ✅ ✅
```

This monotonic behaviour allows Binary Search.

---

# Search Space

Smallest possible speed

```python
low = 1
```

Largest possible speed

```python
high = max(piles)
```

---

# Algorithm

## Step 1

Initialize

```python
low = 1
high = max(piles)
```

---

## Step 2

Repeat while

```python
low <= high
```

Find

```python
mid = (low + high) // 2
```

Here,

```
mid
```

represents the candidate eating speed.

---

## Step 3

Calculate how many hours Koko needs at speed

```
mid
```

Initialize

```python
hours = 0
```

Traverse every pile.

For each pile,

calculate

```python
hours += (pile + mid - 1) // mid
```

This gives the number of hours needed to finish that pile.

---

# Why this Formula?

Suppose

```
pile = 11

speed = 4
```

Normal integer division

```
11 // 4 = 2
```

Wrong.

Koko needs

```
3 hours
```

because

Hour 1

```
4 bananas
```

Hour 2

```
4 bananas
```

Hour 3

```
3 bananas
```

Therefore we need

```
Ceiling Division
```

Instead of

```python
math.ceil(pile / speed)
```

we use

```python
(pile + speed - 1) // speed
```

Example

```
(11+4-1)//4

=14//4

=3
```

---

# Step 4

If

```python
hours <= h
```

the current speed works.

Try a smaller speed.

```python
high = mid - 1
```

---

# Step 5

Otherwise,

the speed is too slow.

Increase the speed.

```python
low = mid + 1
```

---

# Step 6

When Binary Search finishes,

```
low
```

points to the first valid speed.

Return

```python
low
```

- Using

```python

```

instead of

```python

```

- Searching array indices instead of speeds.

- Using

```python

```

instead of ceiling division.

- Returning

```python

```

```python

```


# Interview Takeaway

This problem introduces **Binary Search on Answer**. Instead of searching an array, we search over all possible eating speeds. Since increasing the speed always decreases the required hours, the answers form a monotonic pattern. Binary Search efficiently finds the first valid speed while using ceiling division to correctly calculate the hours required for each pile.

````python
class Solution(object):
    def minEatingSpeed(self, piles, h):
        """
        :type piles: List[int]
        :type h: int
        :rtype: int
        """
        low=1
        high=max(piles)
        while low<=high:
            mid=(low+high)//2
            hours=0
            for i in piles:
                hours+=(i+mid-1)//mid
            if hours<=h:
                high=mid-1
            else:
                low=mid+1
        return low
````
