# LeetCode 153 - Find Minimum in Rotated Sorted Array

## Problem

Given a sorted array that has been rotated between `1` and `n` times, return the minimum element.

- All elements are **unique**.
- The solution must run in **O(log n)** time.

### Example

```
Input:
nums = [3,4,5,1,2]

Output:
1
```

---

## Brute Force Approach

### Idea

Traverse the entire array and keep track of the smallest element.

### Algorithm

1. Initialize the minimum as the first element.
2. Traverse every element.
3. Update the minimum whenever a smaller element is found.
4. Return the minimum.

### Time Complexity

```
O(n)
```

### Space Complexity

```
O(1)
```

---

# Optimal Approach (Binary Search)

## Key Observation

Although the entire array is rotated,

**one half is always sorted.**

Example:

```
3 4 5 | 1 2
```

Left half:

```
3 4 5
```

is sorted.

If one half is sorted, we already know the minimum element of that half.

We save it and continue searching the unsorted half.

---

## Algorithm

### Step 1

Initialize

```python
low = 0
high = len(nums) - 1
minimum = nums[0]
```

---

### Step 2

Repeat while

```python
low <= high
```

Find

```python
mid = (low + high) // 2
```

---

### Step 3

Check if the left half is sorted.

```python
nums[low] <= nums[mid]
```

If true,

the smallest element of the left half is

```python
nums[low]
```

Update

```python
minimum = min(minimum, nums[low])
```

Since the left half has already been considered,

search the right half.

```python
low = mid + 1
```

---

### Step 4

Otherwise,

the rotation point lies in the left half.

The current middle element may itself be the minimum.

Update

```python
minimum = min(minimum, nums[mid])
```

Search the left half.

```python
high = mid - 1
```

---

### Step 5

Return

```python
minimum
```

# Optimized Observation

If at any point

```python
nums[low] <= nums[high]
```

then the entire remaining search space is already sorted.

Therefore,

the minimum is simply

```python
nums[low]
```

You can return immediately.

This avoids unnecessary iterations.

---

# Time Complexity

Binary Search removes half of the search space every iteration.

```
O(log n)
```

---

# Space Complexity

Only constant extra space is used.

```
O(1)
```

One half is always sorted.

# Key Observations

- One half of the array is always sorted.
- If the left half is sorted, its minimum is `nums[low]`.
- If the left half is not sorted, the rotation point lies in the left half.
- Keep updating the minimum before discarding a sorted half.

---

# Common Mistakes

- Forgetting to update the minimum before moving pointers.

- Assuming the minimum is always at `mid`.

- Searching the wrong half after identifying the sorted half.

- Forgetting to return the stored minimum.

- Missing the optimization when the remaining array is already sorted.

---

# Interview Takeaway

This problem is a variation of Binary Search on a rotated sorted array. Since one half of the array is always sorted, we can safely record the smallest element from that sorted half before discarding it. The search then continues in the unsorted half where the rotation point lies. Repeating this process guarantees an **O(log n)** solution.

````python
class Solution(object):
    def findMin(self, nums):
        """
        :type nums: List[int]
        :rtype: int
        """
        low=0
        high=len(nums)-1
        minimum=nums[0]
        while low<=high:
            mid=(low+high)//2
            if nums[low]<=nums[mid]:
                minimum=min(nums[low],minimum)
                low=mid+1
            else:
                minimum=min(nums[mid],minimum)
                high=mid-1
        return minimum
````
