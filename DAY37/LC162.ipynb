# LeetCode 162 - Find Peak Element

## Problem

A **peak element** is an element that is **strictly greater** than its neighbors.

Given an array `nums`, return the **index** of **any** peak element.

You may assume:

```
nums[-1] = -∞
nums[n] = -∞
```

The solution must run in

```
O(log n)
```

---

## Brute Force Approach

### Idea

Traverse the array and check whether each element is greater than both of its neighbors.

The first peak found can be returned.

### Algorithm

1. Traverse every element.
2. Compare it with its left and right neighbors.
3. Return the index if it is a peak.

### Time Complexity

```
O(n)
```

### Space Complexity

```
O(1)
```

---

# Optimal Approach (Binary Search)

## Why Binary Search?

The array is **not sorted**, but we can still eliminate half of the search space.

The decision is based on the **slope**.

If

```
nums[mid] < nums[mid+1]
```

we are moving **uphill**.

A peak must exist on the **right**.

If

```
nums[mid] > nums[mid+1]
```

we are moving **downhill**.

A peak is either at `mid` or somewhere on the **left**.

Thus, every iteration removes half of the search space.

---

## Key Observation

### Uphill

```
1 2 3 4 5
      ^
     mid
```

Since

```
nums[mid] < nums[mid+1]
```

the slope is increasing.

The peak cannot be on the left.

Search the right half.

```
low = mid + 1
```

---

### Downhill

```
5 4 3 2
  ^
 mid
```

Since

```
nums[mid] > nums[mid+1]
```

the slope is decreasing.

`mid` itself may be the peak.

Search the left half including `mid`.

```
high = mid
```

---

## Algorithm

### Step 1

Initialize

```python
low = 0
high = len(nums) - 1
```

---

### Step 2

Repeat while

```python
low < high
```

Find

```python
mid = (low + high) // 2
```

---

### Step 3

If

```python
nums[mid] < nums[mid + 1]
```

move right.

```python
low = mid + 1
```

---

### Step 4

Otherwise,

move left including `mid`.

```python
high = mid
```

---

### Step 5

When

```python
low == high
```

both pointers point to a peak.

Return

```python
low
```

---

nums = [1,2,3,1]

# Primary Pattern

**Binary Search on Slope**

---

# Recognition Clues

Look for problems where:

- The array is **not sorted**.
- You need a local maximum or local minimum.
- The required complexity is **O(log n)**.
- The slope determines which half contains the answer.

Think

```
Slope

+

Binary Search
```

---

# Key Observations

- Binary Search is **not limited to sorted arrays**.
- It works whenever one half can be safely discarded.
- If the slope is increasing, search the right half.
- If the slope is decreasing, search the left half including `mid`.
- The remaining index after the search is a valid peak.

---

# Common Mistakes

- Using

```python
high = mid - 1
```

instead of

```python
high = mid
```

which may discard the peak.

- Returning `mid` instead of `low`.

- Using

```python
while low <= high
```

instead of

```python
while low < high
```

- Trying to find the global maximum instead of **any** peak.

---

# Interview Takeaway

This problem demonstrates that Binary Search is not restricted to sorted arrays. Instead of using sorted order, it relies on the direction of the slope. If the slope is increasing, a peak must exist to the right. If the slope is decreasing, a peak exists at `mid` or to the left. Repeatedly eliminating one half leads to an efficient **O(log n)** solution.

````python
class Solution(object):
    def findPeakElement(self, nums):
        """
        :type nums: List[int]
        :rtype: int
        """
        low=0
        high=len(nums)-1

        while low<high:
            mid=(low+high)//2
            if nums[mid]<nums[mid+1]:
                low=mid+1
            else:
                high=mid
        return low
````
