```python
# LC 128
class Solution(object):
    def longestConsecutive(self, nums):
        """
        :type nums: List[int]
        :rtype: int
        """
        longest=0
        set_nums=set(nums)
        start=0
        for i in set_nums:
            if i-1 not in set_nums:
                start=1
                counter=i
                while counter+1 in set_nums:
                    start+=1
                    counter+=1
            longest=max(longest,start)
        return longest
```
