# LC 560 - Subarray Sum Equals K

## Problem

Given an integer array `nums` and an integer `k`, return the total number of continuous subarrays whose sum equals `k`.

A subarray must be **contiguous**.

### Example

`nums = [1,1,1]`

`k = 2`

Valid subarrays:

- `[1,1]` at indices `0-1`
- `[1,1]` at indices `1-2`

Answer:

`2`

---

## Brute Force Approach

The direct approach is to consider every possible subarray.

For every starting index:

1. Start a running sum.
2. Expand the ending index.
3. Add each new element to the running sum.
4. If the sum equals `k`, increase the count.

This requires two loops.

### Complexity

Time: `O(n²)`

Space: `O(1)`

This is better than recalculating every subarray sum from scratch, but we can do better.

---

## Optimal Approach

Use:

**Prefix Sum + Hash Map**

The key idea is to maintain the cumulative sum as we traverse the array.

Let:

`prefix_sum = sum of elements from index 0 to current index`

Suppose the current prefix sum is:

`prefix_sum`

We want a previous prefix sum such that:

`current_prefix_sum - previous_prefix_sum = k`

Rearranging:

`previous_prefix_sum = current_prefix_sum - k`

Therefore, for every current prefix sum, we look for:

`prefix_sum - k`

in our hash map.

---

## Key Observation

Suppose:

`nums = [1,2,3]`

and:

`k = 5`

At index `2`:

`prefix_sum = 1 + 2 + 3 = 6`

We need a previous prefix sum of:

`6 - 5 = 1`

We already had prefix sum `1` after index `0`.

So:

`6 - 1 = 5`

which means the subarray:

`[2,3]`

has sum `5`.

This is the central equation:

`current_prefix - previous_prefix = k`

Therefore:

`previous_prefix = current_prefix - k`

---

## Why a Hash Map?

The hash map stores:

`prefix_sum → number of times it has appeared`

Example:

`{0:1, 1:2, 3:1}`

means:

- prefix sum `0` appeared once
- prefix sum `1` appeared twice
- prefix sum `3` appeared once

We store the **frequency**, not just whether the prefix sum exists.

This is important because the same prefix sum can occur multiple times, and each occurrence can produce a different valid subarray.

---

## Why Initialize `value = {0:1}`?

This is one of the most important details.

We start with:

`prefix_sum = 0`

and pretend that a prefix sum of `0` has already occurred once.

This handles subarrays that begin at index `0`.

Example:

`nums = [3]`

`k = 3`

After processing `3`:

`prefix_sum = 3`

Target:

`3 - 3 = 0`

Since `0` exists in the map:

`count += 1`

So `[3]` is correctly counted.

Without `{0:1}`, we would miss subarrays beginning at index `0`.

---

## Algorithm

1. Initialize:
   - `prefix_sum = 0`
   - `count = 0`
   - `value = {0:1}`
2. Traverse the array.
3. Add the current element to `prefix_sum`.
4. Calculate:

`target = prefix_sum - k`

5. If `target` exists in the map:
   - add its frequency to `count`.
6. Add the current `prefix_sum` to the map.
7. If it already exists, increase its frequency.
8. Return `count`.

## Important Insight: Why We Add the Frequency

Consider:

`nums = [0,0,0]`

`k = 0`

Prefix sums:

`0,0,0`

Initial map:

`{0:1}`

When we see the first `0`:

`target = 0`

There is already one `0`.

So we find 1 subarray.

Map becomes:

`{0:2}`

Second `0`:

There are now 2 previous prefix sums equal to `0`.

So:

`count += 2`

Third `0`:

There are 3 previous prefix sums equal to `0`.

So:

`count += 3`

Total:

`1 + 2 + 3 = 6`

And there are exactly 6 subarrays in `[0,0,0]`.

This proves why we need:

`value[prefix_sum] += 1`

instead of just storing a boolean.

## Primary Pattern

### Prefix Sum + Frequency Hash Map

This is one of the most important subarray patterns.

The structure is:

`Running Prefix Sum`

↓

`Calculate Required Previous Prefix`

↓

`Look It Up in Hash Map`

↓

`Add Its Frequency`

The key equation is:

`current_prefix - previous_prefix = k`

Therefore:

`previous_prefix = current_prefix - k`

---

## Pattern Recognition

When you see:

- subarray sum equals `k`
- count subarrays with sum `k`
- continuous subarray
- negative numbers may exist
- need to count how many subarrays satisfy a sum condition

think:

**Prefix Sum + Hash Map**

Immediately ask:

> What previous prefix sum would make the current subarray sum equal to `k`?

Answer:

`current_prefix - k`

---

## Why Sliding Window Is Not the General Solution

A common instinct is to use a sliding window.

That works well when all numbers are positive because expanding the window increases the sum and shrinking decreases it predictably.

But LC 560 allows negative numbers.

Example:

`[1,-1,1]`

The sum can increase or decrease unpredictably when the window moves.

Therefore, a standard sliding window does not reliably solve the general problem.

Prefix sums work with:

- positive numbers
- negative numbers
- zeros

This is a major pattern-recognition point.

---

## Generalization

The prefix-sum idea is much broader.

Whenever:

`subarray_sum = current_prefix - previous_prefix`

you can transform a subarray problem into a relationship between two prefix sums.

For LC 560:

`current_prefix - previous_prefix = k`

For other problems, the required relationship might involve:

- equality
- remainder
- difference
- frequency
- parity

The broader pattern is:

**Convert a subarray condition into a relationship between prefix states.**

---

## Common Mistakes

### 1. Forgetting `{0:1}`

Without:

`value = {0:1}`

you miss subarrays that start at index `0`.

---

### 2. Looking for `k` instead of `prefix_sum - k`

The required previous prefix is not:

`k`

It is:

`current_prefix - k`

because:

`current_prefix - previous_prefix = k`

---

### 3. Storing only whether a prefix exists

This is wrong for counting problems.

You need:

`prefix_sum → frequency`

because the same prefix sum can occur multiple times.

---

### 4. Updating the map before checking the target

The correct order is:

1. Update current prefix sum.
2. Search for `prefix_sum - k`.
3. Add its frequency to the answer.
4. Then store the current prefix sum.

If you insert the current prefix before checking, you can accidentally count an invalid zero-length subarray in certain cases.

---

### 5. Using sliding window automatically

Do not assume every subarray sum problem is a sliding-window problem.

If negative numbers are allowed, standard sliding window generally does not work.

---

## Complexity

Let `n` be the length of `nums`.

### Time Complexity

Each element is processed once.

Hash map lookup and insertion are `O(1)` average.

Therefore:

**O(n) average time**

### Space Complexity

The hash map can contain up to `n` different prefix sums.

Therefore:

**O(n) space**

## Interview Follow-Ups

### Q1. Why does `prefix_sum - k` work?

Suppose:

`previous_prefix = 5`

and:

`current_prefix = 12`

The subarray between them has sum:

`12 - 5 = 7`

So if `k = 7`, we need:

`previous_prefix = 12 - 7 = 5`

That is exactly what the hash map searches for.

---

### Q2. Why do we store frequencies?

Because multiple previous prefix sums can have the same value.

Every matching previous prefix creates a different subarray ending at the current index.

Therefore:

`count += frequency[prefix_sum - k]`

---

### Q3. Why does this work with negative numbers?

Because prefix sums do not rely on the sum increasing or decreasing.

We only use the mathematical relationship:

`current_prefix - previous_prefix = subarray_sum`

So negative values are completely valid.

---

### Q4. What if `k = 0`?

Then we search for:

`prefix_sum - 0 = prefix_sum`

So whenever the same prefix sum appears again, the elements between those two positions sum to zero.

The most important lesson from LC 560 is:

> **When you need to count subarrays with a particular sum, convert the subarray condition into a relationship between two prefix sums.**

The key equation is:

`current_prefix - previous_prefix = k`

Therefore:

`previous_prefix = current_prefix - k`

```python
class Solution(object):
    def subarraySum(self, nums, k):
        """
        :type nums: List[int]
        :type k: int
        :rtype: int
        """
        prefix_sum=0 
        n=len(nums)
        value={0:1}
        count=0
        for i in range(n):
            prefix_sum+=nums[i]
            target=prefix_sum-k 
            if target in value:
                count+= value[target]
            if prefix_sum in value:
                value[prefix_sum]+=1
            else:
                value[prefix_sum]=1
        return count 
```
