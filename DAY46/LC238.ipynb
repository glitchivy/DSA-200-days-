# LC 238 - Product of Array Except Self

## Problem

Given an integer array `nums`, return an array `res` such that:

`res[i] = product of every element in nums except nums[i]`

### Important Constraint

You must solve it:

- Without using division
- In `O(n)` time
- With `O(1)` extra space, excluding the output array

### Example

Input:

`nums = [1,2,3,4]`

Output:

`[24,12,8,6]`

Because:

- index 0 → `2 × 3 × 4 = 24`
- index 1 → `1 × 3 × 4 = 12`
- index 2 → `1 × 2 × 4 = 8`
- index 3 → `1 × 2 × 3 = 6`

---

## Brute Force Approach

For every index `i`:

1. Traverse the entire array.
2. Multiply every element except `nums[i]`.
3. Store the product.

This means we traverse the array once for every element.

### Complexity

Time: `O(n²)`

Space: `O(1)` extra space excluding the output.

This works, but it repeats a lot of multiplication.

---

## Better Approach

A common first optimization is to calculate:

- Product of everything to the left
- Product of everything to the right

Then:

`res[i] = left_product × right_product`

We can calculate these products using prefix and suffix arrays.

That gives:

Time: `O(n)`

Space: `O(n)`

But we can avoid storing separate left and right arrays.

---

## Optimal Approach

Use the output array itself to store the left products, then multiply the right products into it during a reverse traversal.

This gives:

**O(n) time + O(1) auxiliary space**

---

## Key Observation

For every index `i`:

`answer[i] = product of elements before i × product of elements after i`

Think:

`nums = [1,2,3,4]`

For index `2`:

`nums[2] = 3`

Elements before it:

`[1,2]`

Product:

`1 × 2 = 2`

Elements after it:

`[4]`

Product:

`4`

Therefore:

`answer[2] = 2 × 4 = 8`

So the problem becomes:

> Calculate the product on the left and the product on the right for every index.

---

## Why Two Traversals?

We cannot know both sides at the same time with only one simple running product.

So we process the array from both directions.

### Left-to-Right

Calculate:

**Product of everything before the current index**

### Right-to-Left

Calculate:

**Product of everything after the current index**

Then multiply the two together.

---

## Algorithm

### Pass 1: Left to Right

Initialize:

`left = 1`

For every index `i`:

1. Put the current `left` product into `res[i]`.
2. Update `left` by multiplying it with `nums[i]`.

Conceptually:

`res[i] = product of everything before i`

Then:

`left = left × nums[i]`

---

### Pass 2: Right to Left

Initialize:

`right = 1`

For every index from right to left:

1. Multiply `res[i]` by `right`.
2. Update `right` by multiplying it with `nums[i]`.

Conceptually:

`res[i] = left_product × right_product`

Then:

`right = right × nums[i]`

---

## Primary Pattern

### Prefix Product + Suffix Product

The main pattern is:

**Left Contribution + Right Contribution**

For every index:

`answer[i] = left_product × right_product`

The clever part is that we don't need separate arrays.

We use:

`res`

to store the left products and then multiply the right products into it.

---

## Pattern Recognition

When you see:

- product of all elements except current
- left and right contribution
- prefix/suffix relationship
- "except self"
- no division allowed

immediately ask:

> What information comes from the left?

and:

> What information comes from the right?

If the answer can be expressed as:

`left contribution × right contribution`

think:

**Prefix + Suffix**

---

## Why We Cannot Simply Multiply Everything

A tempting solution is:

`total_product = product of all elements`

Then:

`answer[i] = total_product / nums[i]`

But the problem explicitly says not to use division.

More importantly, division creates problems when the array contains zeros.

Example:

`[1,2,0,4]`

The total product is `0`, but the answer is still meaningful:

`[0,0,8,0]`

The prefix/suffix approach naturally handles zeros without any special division logic.

---

## Generalization

This pattern applies beyond multiplication.

For many array problems, each element's answer can be divided into:

`information before i`

and:

`information after i`

Then:

`answer[i] = left_information + right_information`

or:

`answer[i] = left_information × right_information`

depending on the problem.

This is the broader:

**Prefix/Suffix Contribution Pattern**

---

## Common Mistakes

### 1. Including `nums[i]` in its own product

For:

`nums = [1,2,3,4]`

At index `2`, the answer must be:

`1 × 2 × 4`

It must NOT contain `3`.

That is why:

`left`

represents everything before `i`

and:

`right`

represents everything after `i`.

---

### 2. Updating the running product too early

During the left-to-right pass:

First:

`res[i] = left`

Then:

`left *= nums[i]`

If you multiply first, you accidentally include `nums[i]` in its own left product.

The same idea applies to the right-to-left pass.

---

### 3. Confusing `left` and `right`

Your variables represent:

`left = product of elements before current index`

`right = product of elements after current index`

They are not the product of the entire left/right half of the array.

They change as we move through the array.

---

### 4. Creating separate left and right arrays

That works, but it uses `O(n)` extra space.

The optimal approach uses the output array itself for the left products.

---

## Complexity

Let `n` be the length of the array.

### Time Complexity

First traversal:

`O(n)`

Second traversal:

`O(n)`

Total:

**O(n)**

### Space Complexity

The output array `res` requires:

`O(n)`

But this is required output space.

Extra auxiliary variables:

`left`, `right`, indices

Therefore:

**O(1) auxiliary space**

This satisfies the problem's constant-space requirement.

---

## Interview Follow-Ups

### Q1. Why do we use two passes?

Because each answer depends on two directions:

`left of i`

and:

`right of i`

One pass naturally builds one side, while the second pass builds the other.

---

### Q2. Why initialize `left` and `right` to 1?

Because `1` is the multiplicative identity.

Before any elements have been multiplied:

`product = 1`

This allows the first and last positions to work naturally.

---

### Q3. What happens with zeros?

The prefix/suffix approach handles them naturally.

Example:

`[1,2,0,4]`

The answer becomes:

`[0,0,8,0]`

No special zero handling is required.

---

### Q4. Why is this considered O(1) extra space?

Because `res` is the required output array.

We are not creating separate prefix and suffix arrays.

Only constant-size variables are used besides the output.

```python
class Solution(object):
    def productExceptSelf(self, nums):
        """
        :type nums: List[int]
        :rtype: List[int]
        """
        n=len(nums)
        res=[1]*n 
        left=1
        right=1 
        for i in range(n):
            res[i]*=left
            left*=nums[i]
        for i in range(n-1,-1,-1):
            res[i]*=right
            right*=nums[i]
        return res
        
```
