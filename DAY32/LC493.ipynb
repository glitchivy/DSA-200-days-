# LeetCode 493 - Reverse Pairs

## Problem

Given an integer array `nums`, return the number of **reverse pairs**.

A reverse pair is defined as:

```
i < j and nums[i] > 2 * nums[j]
```

---

## Brute Force Approach

### Idea

Check every possible pair `(i, j)` where `i < j`.

If:

```
nums[i] > 2 * nums[j]
```

increment the answer.

### Algorithm

1. Iterate through every element.
2. For each element, check all elements after it.
3. Count every valid reverse pair.

### Time Complexity

- **O(n²)**

### Space Complexity

- **O(1)**

---

## Optimal Approach (Merge Sort)

### Key Observation

A reverse pair can belong to:

- Left half
- Right half
- One element in left half and one in right half (Cross Pair)

The recursive calls count reverse pairs inside the left and right halves.

The only remaining task is counting **cross reverse pairs** before merging the two sorted halves.

Since both halves are already sorted, we can count these pairs efficiently using two pointers.

---

## Algorithm

1. Divide the array into two halves.
2. Recursively count reverse pairs in the left half.
3. Recursively count reverse pairs in the right half.
4. Count reverse pairs where:
   - first element is in the left half
   - second element is in the right half
5. Merge the two sorted halves.
6. Return the total count.

---

## Counting Cross Reverse Pairs

Suppose:

```
Left  = [1,3,5]
Right = [1,2,4]
```

Initialize:

```
j = mid + 1
```

For every element in the left half:

Move `j` while:

```
nums[i] > 2 * nums[j]
```

Once the condition fails,

```
Reverse Pairs = j - (mid + 1)
```

Why?

Because every element before `j` satisfied the condition.

The pointer `j` never moves backward because both halves are sorted.

Thus the counting step runs in **O(n)**.

---

## Merge Step

After counting reverse pairs:

1. Merge the two sorted halves.
2. Store the merged elements in a temporary array.
3. Copy the temporary array back into the original array.

```
nums[left:right+1] = temp
```

This replaces only the current portion being merged.

## Common Mistakes

- Using `<` instead of `>`.
- Writing:

```
count += j - mid + 1
```

instead of

```
count += j - (mid + 1)
```

- Resetting `j` for every `i`.
- Forgetting to merge after counting.
- Forgetting to copy the merged array back into `nums`.
- Returning the count before merging.

---

## Interview Takeaway

This problem teaches the classic **Merge Sort + Counting** pattern.

Instead of checking every pair, recursively sort the array and count only the reverse pairs that span the two halves. Because both halves are sorted, a two-pointer technique counts all cross pairs in linear time, reducing the overall complexity from **O(n²)** to **O(n log n)**.

This same pattern is used in problems like:

- Count Inversions
- Count of Smaller Numbers After Self
- Count of Range Sum
- Reverse Pairs

````python
class Solution(object):
    def reversePairs(self, nums):
        """
        :type nums: List[int]
        :rtype: int
        """
        def merge_sort(left,right):
            if left>=right:
                return 0
            mid=(left+right)//2
            count=0
            count+=merge_sort(left,mid)
            count+=merge_sort(mid+1,right)
            j=mid+1
            for i in range(left,mid+1):
                while j<=right and nums[i]>2*nums[j]:
                    j+=1
                count+=j-(mid+1)
            i=left
            j=mid+1
            temp=[]
            while i <= mid and j <= right:
                if nums[i]<=nums[j]:
                    temp.append(nums[i])
                    i+=1
                else:
                    temp.append(nums[j])
                    j+=1
            while i<=mid:
                temp.append(nums[i])
                i+=1
            while j <= right:
                temp.append(nums[j])
                j += 1
            nums[left:right + 1] = temp
            return count 
        return merge_sort(0, len(nums) - 1)
````
