# LeetCode 152 - Maximum Product Subarray

## Problem

Given an integer array `nums`, find the **contiguous subarray** within the array that has the **largest product**, and return that product.

---

## Brute Force Approach

### Idea

Generate every possible contiguous subarray.

For each subarray:

- Calculate its product.
- Keep track of the maximum product obtained.

### Algorithm

1. Start every index as the beginning of a subarray.
2. Extend the subarray one element at a time.
3. Compute the product.
4. Update the maximum product.

### Time Complexity

- **O(n²)**

### Space Complexity

- **O(1)**

---

## Why Normal Kadane Doesn't Work

Kadane's Algorithm works for sums because:

- Adding a negative number always decreases the sum.

Products behave differently.

```
Positive × Negative = Negative

Negative × Negative = Positive
```

A negative product that looks bad now can become the maximum product later after multiplying by another negative number.

Example:

```
[-2, 3, -4]

(-2 × 3) = -6

(-6 × -4) = 24
```

If we discard `-6`, we lose the correct answer.

---

## Key Observation

At every index we must keep track of:

- Maximum product ending at the current index.
- Minimum product ending at the current index.

Why minimum?

Because multiplying a large negative product with another negative number produces a large positive product.

---

## Optimal Approach

Maintain three variables:

```
maxProd
minProd
ans
```

If the current element is negative:

```
Swap(maxProd, minProd)
```

Why?

Because:

```
Maximum × Negative → Minimum

Minimum × Negative → Maximum
```

The roles reverse whenever we multiply by a negative number.

---

## Algorithm

1. Initialize:

```
maxProd = nums[0]
minProd = nums[0]
ans = nums[0]
```

2. Traverse from index `1`.

3. If the current element is negative:

```
Swap(maxProd, minProd)
```

4. Update maximum product:

```
maxProd = max(nums[i], maxProd × nums[i])
```

5. Update minimum product:

```
minProd = min(nums[i], minProd × nums[i])
```

6. Update answer:

```
ans = max(ans, maxProd)
```

7. Return `ans`.

## Recognition Clues

Look for problems where:

- Maximum product is required.
- Array contains negative numbers.
- Product changes sign.
- Negative × Negative becomes Positive.
- Need to track both best and worst values.

Keywords:

- Maximum Product
- Contiguous Subarray
- Negative Numbers
- Dynamic Programming
- Kadane Variant

---

## Common Mistakes

- Applying normal Kadane's Algorithm.
- Tracking only the maximum product.
- Forgetting to swap `maxProd` and `minProd` when encountering a negative number.
- Updating `ans` using `minProd`.
- Initializing variables with `0` instead of `nums[0]`.

---

## Interview Takeaway

Unlike maximum subarray sum, product problems require tracking **both the maximum and minimum products** because multiplying by a negative number flips their roles. By maintaining these two values and swapping them whenever a negative number is encountered, we can compute the maximum product subarray in **O(n)** time using a modified version of Kadane's Algorithm.

````python
class Solution(object):
    def maxProduct(self, nums):
        """
        :type nums: List[int]
        :rtype: int
        """
        maxprod=nums[0]
        minprod=nums[0]
        ans=nums[0]
        for i in range(1,len(nums)):
            if nums[i]<0:
                maxprod,minprod=minprod,maxprod
            maxprod = max(nums[i], maxprod * nums[i])
            minprod = min(nums[i], minprod * nums[i])
            ans = max(ans, maxprod)
        return ans
        
````
