# LeetCode 1283 - Find the Smallest Divisor Given a Threshold

## Problem

You are given:

- An integer array `nums`
- An integer `threshold`

Choose a positive integer **divisor**.

For every element in `nums`:

```
ceil(nums[i] / divisor)
```

is calculated.

The sum of all these values must be

```
<= threshold
```

Return the **smallest divisor** that satisfies the condition.

---

# Example

```
nums = [1,2,5,9]

threshold = 6
```

Try divisor = 5

```
ceil(1/5) = 1

ceil(2/5) = 1

ceil(5/5) = 1

ceil(9/5) = 2
```

Total

```
1 + 1 + 1 + 2 = 5
```

Since

```
5 <= 6
```

divisor = 5 works.

---

# Brute Force Approach

## Idea

Try every divisor from

```
1
```

to

```
max(nums)
```

For each divisor,

calculate the total sum.

Return the first divisor whose sum is

```
<= threshold
```

---

## Time Complexity

```
O(n × max(nums))
```

---

## Space Complexity

```
O(1)
```

# Optimal Approach (Binary Search on Answer)

## Key Observation

We are **not searching the array**.

We are searching the answer.

The answer is

```
Smallest Divisor
```

Possible divisors lie between

```python
1
```

and

```python
max(nums)
```

---

## Monotonic Property

As the divisor increases,

each division result becomes smaller or remains the same.

Therefore,

the total sum decreases.

Example

```
Divisor

1 2 3 4 5 6 7 8 9
```

Valid?

```
❌ ❌ ❌ ❌ ✅ ✅ ✅ ✅ ✅
```

This monotonic pattern allows Binary Search.

---

# Search Space

```python
low = 1
high = max(nums)
```

---

# Algorithm

## Step 1

Initialize

```python
low = 1
high = max(nums)
```

---

## Step 2

Repeat while

```python
low <= high
```

Find

```python
mid = (low + high) // 2
```

Here,

```
mid
```

represents the candidate divisor.

---

## Step 3

Initialize

```python
total = 0
```

Traverse every number.

```python
for i in nums:
```

---

## Step 4

Calculate ceiling division.

```python
total += (i + mid - 1) // mid
```

This computes

```
⌈i / mid⌉
```

without using `math.ceil()`.

---

# Why Ceiling Division?

Suppose

```
i = 9

divisor = 4
```

Normal division

```
9 // 4 = 2
```

Wrong.

Actual value

```
ceil(9/4)

= 3
```

Formula

```python
(i + divisor - 1) // divisor
```

Example

```
(9+4-1)//4

=12//4

=3
```

---

## Step 5

After processing every number,

check

```python
total <= threshold
```

---

### If True

The divisor works.

Try finding a smaller divisor.

```python
high = mid - 1
```

---

### Otherwise

The divisor is too small.

Increase it.

```python
low = mid + 1
```

---

## Step 6

When Binary Search finishes,

return

```python
low
```

because it points to the first valid divisor.

# Key Observations

- We search the **divisor**, not the array.
- Larger divisor → smaller total sum.
- Smaller divisor → larger total sum.

Pattern

```
❌ ❌ ❌ ✅ ✅ ✅
```

We need the **first** valid divisor.

# Common Mistakes

- Using

```python
total += i // divisor
```

instead of ceiling division.

- Writing

```python
if total <= threshold
```

inside the `for` loop instead of after processing the entire array.

- Mixing Binary Search templates:

Wrong

```python
while low < high

high = mid - 1
```

Correct (Template 1)

```python
while low <= high

high = mid - 1
```

Correct (Template 2)

```python
while low < high

high = mid
```

- Returning `mid` instead of `low`.

# Interview Takeaway

This is a classic **Binary Search on Answer** problem. We search over all possible divisors instead of array indices. For each candidate divisor, we compute the sum of the ceiling divisions and check whether it satisfies the threshold. Since increasing the divisor always decreases (or maintains) the total sum, the answers form a monotonic pattern, allowing Binary Search to efficiently find the smallest valid divisor in **O(n × log(max(nums)))** time.

````python
class Solution(object):
    def smallestDivisor(self, nums, threshold):
        """
        :type nums: List[int]
        :type threshold: int
        :rtype: int
        """
        low=1
        high=max(nums)
        while low<=high:
            mid=(low+high)//2
            total=0
            for i in nums:
                total+=(i+mid-1)//mid
            if total<=threshold:
                high=mid-1
            else:
                low=mid+1
        return low

````
