# LeetCode 35 - Search Insert Position

## Problem

Given a sorted array of **distinct integers** `nums` and an integer `target`, return:

- The index if the target exists.
- Otherwise, return the index where it should be inserted to maintain the sorted order.

The solution should run in **O(log n)** time.

---

## Brute Force Approach

### Idea

Traverse the array from left to right.

- If the target is found, return its index.
- If an element greater than the target is found, return its index as the insertion position.
- If the target is greater than every element, return the length of the array.

### Algorithm

1. Traverse the array.
2. If `nums[i] == target`, return `i`.
3. If `nums[i] > target`, return `i`.
4. If the traversal completes, return `len(nums)`.

### Time Complexity

**O(n)**

### Space Complexity

**O(1)**

---

## Optimal Approach (Binary Search)

### Key Observation

The array is sorted.

Instead of checking every element, Binary Search eliminates half of the remaining search space in every iteration.

At the end of the search:

- `high` points to the last element smaller than the target.
- `low` points to the first element greater than or equal to the target.

Therefore, the insertion position is always:

```
low
```

This is exactly the **Lower Bound** of the target.

---

## Algorithm

1. Initialize:

```
low = 0
high = n - 1
```

2. While:

```
low <= high
```

3. Compute:

```
mid = (low + high) // 2
```

4. If the target is found:

```
return mid
```

5. If:

```
target > nums[mid]
```

Search the right half.

```
low = mid + 1
```

6. Otherwise search the left half.

```
high = mid - 1
```

7. If the loop finishes without finding the target:

```
return low
```

---

## Dry Run

### Input

```
nums = [1,3,5,6]

target = 2
```

### Iteration 1

```
low = 0

high = 3

mid = 1

nums[mid] = 3
```

Since

```
2 < 3
```

Move left.

```
high = 0
```

---

### Iteration 2

```
low = 0

high = 0

mid = 0

nums[mid] = 1
```

Since

```
2 > 1
```

Move right.

```
low = 1
```

---

Loop ends.

```
low = 1

high = 0
```

Return:

```
1
```

The target should be inserted at index `1`.

---

## Why do we return `low`?

When Binary Search finishes:

- Every index before `low` contains a value smaller than the target.
- `low` is the first position where the target can be inserted while keeping the array sorted.

In other words,

```
low = Lower Bound
```

where Lower Bound means:

```
First index where nums[i] >= target
```

## Common Mistakes

- Returning `high` instead of `low`.

- Returning the last computed `mid`.

- Using:

```
while low < high
```

instead of

```
while low <= high
```

- Forgetting that if the target is greater than every element, the answer is:

```
len(nums)
```

- Moving the pointers incorrectly and causing an infinite loop.

---

## Interview Takeaway

Search Insert Position is a direct application of the **Lower Bound** concept. Binary Search is used to locate the first position where the target can appear while maintaining sorted order. When the target is absent, the variable `low` naturally points to the correct insertion index. This Lower Bound template is widely used in interview problems involving first occurrence, last occurrence, insertion position, and Binary Search on Answer.

````python
class Solution(object):
    def searchInsert(self, nums, target):
        """
        :type nums: List[int]
        :type target: int
        :rtype: int
        """
        low=0
        high=len(nums)-1
        while low<=high:
            mid=(low+high)//2
            if target==nums[mid]:
                return mid
            elif target>nums[mid]:
                low=mid+1
            else:
                high=mid-1
        return low
````
