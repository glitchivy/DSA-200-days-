```python
# LC 118
class Solution(object):
    def generate(self, numRows):
        """
        :type numRows: int
        :rtype: List[List[int]]
        """
        final=[]
        final.append([1])
        for i in range(numRows-1):
            prev=final[-1]
            row=[1]
            for i in range(len(prev)-1):
                row.append(prev[i]+prev[i+1])
            row.append(1)
            final.append(row)
        return final

```

```python
# 229
class Solution(object):
    def majorityElement(self, nums):
        """
        :type nums: List[int]
        :rtype: List[int]
        """
        can1= None
        can2= None
        count1=0
        count2=0
        for i in nums:
            if i==can1:
                count1+=1
            elif i == can2:
                count2+=1
            elif count1==0:
                can1=i
                count1=1
            elif count2==0:
                can2=i
                count2=1
            else:
                count1-=1
                count2-=1
        count1=0
        count2=0
        ans=[]
        for i in nums:
            if i ==can1:
                count1+=1
            elif i==can2:
                count2+=1
        if count1>len(nums)//3:
            ans.append(can1)
        if count2>len(nums)//3:
            ans.append(can2)
        return ans
```
