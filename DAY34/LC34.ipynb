# LeetCode 34 - Find First and Last Position of Element in Sorted Array

## Problem

Given a sorted array of integers `nums` and an integer `target`, return the starting and ending position of the target in the array.

- If the target exists, return:

```
[first occurrence, last occurrence]
```

- If the target does not exist, return:

```
[-1, -1]
```

The solution must run in **O(log n)** time.

---

## Brute Force Approach

### Idea

Traverse the array from left to right.

- Record the first occurrence of the target.
- Continue traversing and update the last occurrence whenever the target is found.

### Algorithm

1. Initialize `first = -1` and `last = -1`.
2. Traverse the array.
3. When the target is found:
   - If `first == -1`, store the current index as `first`.
   - Update `last` to the current index.
4. Return `[first, last]`.

### Time Complexity

**O(n)**

### Space Complexity

**O(1)**

---

## Optimal Approach (Lower Bound + Upper Bound)

### Key Observation

Since the array is sorted, Binary Search can be used.

Instead of searching for the first and last occurrence directly, perform two Binary Searches.

### Lower Bound

Lower Bound returns:

```
First index where nums[i] >= target
```

This is the **first occurrence** of the target if it exists.

---

### Upper Bound

Upper Bound returns:

```
First index where nums[i] > target
```

Since it points to the first element greater than the target,

the last occurrence is:

```
upperBound - 1
```

---

## Algorithm

### Step 1

Find the Lower Bound.

```
left = lowerBound(target)
```

---

### Step 2

Verify whether the target exists.

If

```
left == len(nums)
```

or

```
nums[left] != target
```

return

```
[-1, -1]
```

---

### Step 3

Find the Upper Bound.

```
right = upperBound(target)
```

---

### Step 4

Return

```
[left, right - 1]
```

---

## Dry Run

### Input

```
nums = [5,7,7,8,8,10]

target = 8
```

### Lower Bound

Search for the first element

```
>= 8
```

Result:

```
3
```

---

### Upper Bound

Search for the first element

```
> 8
```

Result:

```
5
```

---

### Final Answer

```
left = 3

right = 5 - 1 = 4
```

Return

```
[3,4]
```

---

## Example (Target Not Present)

### Input

```
nums = [5,7,7,8,8,10]

target = 6
```

Lower Bound returns

```
1
```

because

```
nums[1] = 7
```

Since

```
7 != 6
```

the target is not present.

Return

```
[-1,-1]
```

---

## Why do we use `upperBound - 1`?

Upper Bound returns the index of the **first element greater than the target**.

Example:

```
5 7 7 8 8 10
        ^
```

Upper Bound points here:

```
5 7 7 8 8 10
            ^
```

Therefore,

```
Last Occurrence = Upper Bound - 1
```

**O(log n)**

## Common Mistakes

- Forgetting to verify whether the target actually exists.

- Returning

```
[left, upperBound]
```

instead of

```
[left, upperBound - 1]
```

- Using `>=` instead of `>` in the Upper Bound function.

- Moving the wrong pointer in Binary Search.

- Returning the Lower Bound even when the target is absent.

---

## Interview Takeaway

This problem is a classic application of **Lower Bound** and **Upper Bound**. The first Binary Search finds the first index where the target can appear (`>= target`), while the second finds the first index after the target (`> target`). After confirming that the target exists, the required range is simply `[lowerBound, upperBound - 1]`. Mastering this pattern makes many Binary Search interview questions much easier.

````python
class Solution(object):
    def searchRange(self, nums, target):
        """
        :type nums: List[int]
        :type target: int
        :rtype: List[int]
        """
        def lower():
            low=0
            high=len(nums)-1
            ans=len(nums)
            while low<=high:
                mid=(low+high)//2
                if nums[mid]>=target:
                    ans=mid
                    high=mid-1
                else:
                    low=mid+1
            return ans
        def higher():
            low=0
            high=len(nums)-1
            ans=len(nums)
            while low<=high:
                mid=(low+high)//2
                if nums[mid]>target:
                    ans=mid
                    high=mid-1
                else:
                    low=mid+1
            return ans
        left=lower()
        if left==len(nums) or nums[left]!=target:
            return [-1,-1]
        right=higher()
        return [left,right-1]
````
