# LC 11 - Container With Most Water

## 1. Problem

Given an integer array `height`, where each element represents the height of a vertical line, find two lines that together with the x-axis form a container that holds the maximum amount of water.

Return the maximum amount of water the container can store.

### Example

Input:

`height = [1,8,6,2,5,4,8,3,7]`

Output:

`49`

The maximum area is formed using heights:

`8` and `7`

with width:

`7`

Therefore:

`area = 7 × 7 = 49`

---

## 2. Core Idea

Use the **Two Pointer** approach.

Start with:

- `left` at the beginning
- `right` at the end

For every pair of lines:

`area = width × height`

where:

`width = right - left`

and:

`height = min(height[left], height[right])`

We calculate the area, update the maximum, and then move the pointer pointing to the **shorter line**.

### Memory Rule

**Calculate → Record maximum → Move shorter pointer**

---

## 3. Pointer Initialization

Initialize:

`left = 0`

`right = len(height) - 1`

This gives the maximum possible width.

Example:

`height = [1,8,6,2,5,4,8,3,7]`

Initially:

`left = 0`

`right = 8`

Width:

`8 - 0 = 8`

---

## 4. Area Formula

The area of the container is:

`Area = Width × Height`

Width:

`right - left`

Height:

`min(height[left], height[right])`

Therefore:

`area = (right - left) × min(height[left], height[right])`

### Important

The shorter line determines how much water can actually be held.

If:

`height[left] = 4`

and:

`height[right] = 7`

the container can only hold water up to height:

`4`

Therefore:

`area = width × 4`

---

## 5. Why Move the Shorter Pointer?

This is the most important insight in the problem.

Suppose:

`height[left] < height[right]`

The left line is shorter.

If we move `right` inward:

- Width decreases.
- The limiting height is still potentially the left height.
- Therefore, we cannot improve the area by keeping the same shorter left line.

So we move:

`left += 1`

### Similarly

If:

`height[right] < height[left]`

move:

`right -= 1`

### Memory Rule

**Shorter side is the bottleneck → move it**

---

## 6. Why Not Move the Taller Pointer?

Suppose:

`height[left] = 3`

`height[right] = 10`

Current area:

`width × 3`

If we move the right pointer:

- Width becomes smaller.
- The height cannot exceed `3` unless the left pointer changes.

So the area cannot improve by moving only the taller side.

Therefore:

**Move the shorter side.**

## 17. Code

class Solution(object):
    def maxArea(self, height):
        """
        :type height: List[int]
        :rtype: int
        """
        max_area = 0
        left = 0
        right = len(height) - 1

        while left < right:
            width = right - left
            h = min(height[left], height[right])
            area = width * h
            max_area = max(max_area, area)

            if height[left] < height[right]:
                left += 1
            else:
                right -= 1

        return max_area

## 19. Important Code Observation

The most important line is:

`if height[left] < height[right]:`

This determines which pointer moves.

The rule is:

**Move the shorter line.**

Why?

Because the shorter line limits the current container height.

Moving the taller line only decreases the width without removing the current height bottleneck.

This is what allows us to eliminate many impossible pairs without checking them individually.

## 20. Complexity + Pattern Recognition

### Brute Force Approach

Try every pair:

`i, j`

Calculate every possible area.

This requires:

`O(n²)`

time.

That is too slow for large arrays.

### Two Pointer Approach

Start with the widest possible container.

Then eliminate impossible candidates by moving the shorter side.

Time:

`O(n)`

Space:

`O(1)`

---

### Pattern

**Two Pointers + Greedy Elimination**

### Trigger

When you see:

- Array of heights
- Choose two positions
- Maximize an area/value
- Width depends on distance between pointers
- The smaller height is the limiting factor

Think:

**Two Pointers + Move the Bottleneck**

### General Pattern

`left = beginning`

`right = end`

Calculate current result.

Then:

**Move the pointer with the smaller limiting value.**
