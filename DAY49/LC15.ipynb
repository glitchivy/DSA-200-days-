# LC 15 - 3Sum

## 1. Problem

Given an integer array `nums`, return all unique triplets:

`[nums[i], nums[j], nums[k]]`

such that:

`nums[i] + nums[j] + nums[k] == 0`

The solution must not contain duplicate triplets.

### Example

Input:

`nums = [-1,0,1,2,-1,-4]`

Output:

`[[-1,-1,2],[-1,0,1]]`

The valid triplets are:

`-1 + -1 + 2 = 0`

`-1 + 0 + 1 = 0`

---

## 2. Core Idea

Use:

**Sorting + Two Pointers**

First sort the array.

Then:

1. Fix one number using `i`.
2. Use `left` and `right` to find the other two numbers.
3. Compare their sum with `0`.
4. Skip duplicates at every level.

### Memory Rule

**Sort → Fix `i` → Two Pointers → Find sum → Skip duplicates**

---

## 3. Why Do We Sort?

Sorting is the key that makes the two-pointer approach possible.

After sorting:

- If the sum is too small, move `left` forward.
- If the sum is too large, move `right` backward.
- Equal values become adjacent, making duplicate removal easy.

Example:

Before:

`[-1,0,1,2,-1,-4]`

After sorting:

`[-4,-1,-1,0,1,2]`

Now the values have a predictable order.

---

## 4. Three Pointer Roles

There are effectively three positions:

### `i`

The fixed first number of the triplet.

### `left`

Searches for the second number.

Starts at:

`i + 1`

### `right`

Searches for the third number.

Starts at:

`n - 1`

### Memory Rule

**`i` fixes → `left` increases → `right` decreases**

---

## 5. Pointer Initialization

After sorting:

`n = len(nums)`

For every possible first element:

`i`

we initialize:

`left = i + 1`

`right = n - 1`

This guarantees:

`i < left < right`

So the same element is never used twice.

---

## 6. Main Decision Logic

Calculate:

`target = nums[i] + nums[left] + nums[right]`

Then there are three possibilities.

### Case 1: `target > 0`

The sum is too large.

Because the array is sorted, we need a smaller value.

Move:

`right -= 1`

### Case 2: `target < 0`

The sum is too small.

We need a larger value.

Move:

`left += 1`

### Case 3: `target == 0`

We found a valid triplet.

Add it to the result.

Then move both pointers:

`left += 1`

`right -= 1`

---

## 7. Why Does `target > 0` Mean Move `right`?

The array is sorted.

If:

`nums[i] + nums[left] + nums[right] > 0`

the sum is too large.

Moving `left` forward would make the sum even larger or equal.

That is the wrong direction.

So we decrease the largest value:

`right -= 1`

### Memory Rule

**Sum too big → decrease `right`**

---

## 8. Why Does `target < 0` Mean Move `left`?

If:

`nums[i] + nums[left] + nums[right] < 0`

the sum is too small.

Moving `right` backward would make the sum smaller.

Again, wrong direction.

So we increase the smaller value:

`left += 1`

### Memory Rule

**Sum too small → increase `left`**

---

## 9. Duplicate Handling - `i`

This line is critical:

`if i > 0 and nums[i] == nums[i-1]:`

`continue`

Why?

Because if the current `nums[i]` is the same as the previous `nums[i]`, we would generate the same set of triplets again.

Example:

`[-1,-1,0,1,2]`

The first `-1` can be used as the fixed value.

The second `-1` would produce the same combinations.

Therefore:

**Skip duplicate fixed values.**

---

## 10. Duplicate Handling - `left`

After finding a valid triplet:

`left += 1`

Then:

`while left < right and nums[left] == nums[left-1]:`

move `left` forward.

Why?

Suppose we found:

`[-1,0,1]`

If the next value is another `0`, using it would produce the same triplet again.

So we skip duplicate second values.

---

## 11. Duplicate Handling - `right`

Similarly:

`right -= 1`

Then:

`while left < right and nums[right] == nums[right+1]:`

move `right` backward.

This prevents duplicate third values from producing the same triplet.

## 20. Code

class Solution(object):
    def threeSum(self, nums):
        """
        :type nums: List[int]
        :rtype: List[List[int]]
        """
        nums.sort()
        n = len(nums)
        res = []

        for i in range(n):
            if i > 0 and nums[i] == nums[i-1]:
                continue

            left = i + 1
            right = n - 1

            while left < right:
                target = nums[i] + nums[left] + nums[right]

                if target > 0:
                    right -= 1

                elif target < 0:
                    left += 1

                else:
                    res.append([nums[i], nums[left], nums[right]])

                    left += 1
                    right -= 1

                    while left < right and nums[left] == nums[left-1]:
                        left += 1

                    while left < right and nums[right] == nums[right+1]:
                        right -= 1

        return res

## 21. Final Summary + Status

### Problem

Find all unique triplets whose sum is `0`.

### Approach

**Sorting + Fixed Pointer + Two Pointers**

First:

`nums.sort()`

Then for every `i`:

`left = i + 1`

`right = n - 1`

Calculate:

`nums[i] + nums[left] + nums[right]`

If:

`sum > 0`

move:

`right--`

If:

`sum < 0`

move:

`left++`

If:

`sum == 0`

store the triplet and move both pointers.

### Duplicate Rule

There are **three duplicate checks**:

1. Skip duplicate `i` values.
2. After a valid triplet, skip duplicate `left` values.
3. After a valid triplet, skip duplicate `right` values.

### Complexity

Sorting:

`O(n log n)`

Two-pointer search:

`O(n²)`

Overall:

`O(n²)`

Extra space:

`O(1)` excluding the output array, depending on the sorting implementation.

### Pattern

**Sort + Fix One + Two Pointers + Duplicate Skipping**

### Most Important Insight

3Sum is essentially:

**1 fixed element + 2Sum using two pointers**

Once `i` is fixed, the remaining problem becomes:

`nums[left] + nums[right] = -nums[i]`

### Master Mental Model

**Sort → Fix `i` → `left/right` search → Too small? `left++` → Too big? `right--` → Found? Store + skip duplicates**

### One-Line Rule

**3Sum = Fix one number + solve the remaining 2Sum with two pointers**

```python
class Solution(object):
    def threeSum(self, nums):
        """
        :type nums: List[int]
        :rtype: List[List[int]]
        """
        nums.sort()
        n=len(nums)
        res=[]
        for i in range(n):
            if i>0 and nums[i]==nums[i-1]:
                continue
            left=i+1
            right=n-1
            while left<right:
                target=nums[i]+nums[left]+nums[right]
                if target>0:
                    right-=1
                elif target<0:
                    left+=1
                else:
                    res.append([nums[i],nums[left],nums[right]])
                    left+=1
                    right-=1
                    while left<right and nums[left]==nums[left-1]:
                        left+=1
                    while left<right and nums[right]==nums[right+1]:
                        right-=1
        return res
```
