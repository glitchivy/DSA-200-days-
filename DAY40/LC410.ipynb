# LeetCode 410 - Split Array Largest Sum

## Problem

Given an array `nums` and an integer `k`, split `nums` into exactly `k` non-empty contiguous subarrays.

The goal is to **minimize the largest subarray sum**.

---

## Example

```text
nums = [7,2,5,10,8]
k = 2
```

Possible splits:

```text
[7] | [2,5,10,8]
7   | 25
```

Largest sum = `25`

```text
[7,2] | [5,10,8]
9     | 23
```

Largest sum = `23`

```text
[7,2,5] | [10,8]
14      | 18
```

Largest sum = `18`

```text
[7,2,5,10] | [8]
24          | 8
```

Largest sum = `24`

Therefore:

```text
Answer = 18
```

---

## Brute Force Approach

### Idea

Try every possible way to split the array into `k` contiguous subarrays.

For every partition:

1. Calculate the sum of each subarray.
2. Find the largest subarray sum.
3. Keep the minimum largest sum.

This becomes very expensive because the number of possible partitions grows rapidly.

### Time Complexity

Exponential in the number of possible partitions.

### Space Complexity

`O(k)`

---

## Optimal Approach

### Binary Search on Answer + Greedy

Instead of trying every possible partition, binary search for the **minimum possible largest subarray sum**.

The answer lies between:

```python
low = max(nums)
high = sum(nums)
```

### Why `low = max(nums)`?

The largest element must belong to some subarray.

Therefore, the answer cannot be smaller than the largest element.

### Why `high = sum(nums)`?

If all elements are placed into one subarray, its sum is `sum(nums)`.

Therefore, the answer cannot be greater than the total sum.

---

## Feasibility Check

For every `mid`, ask:

> Can we split the array into at most `k` subarrays where every subarray has a sum `<= mid`?

We greedily build subarrays from left to right.

```python
curr_sum = 0
groups = 1

for num in nums:
    if curr_sum + num > mid:
        groups += 1
        curr_sum = num
    else:
        curr_sum += num
```

If adding the current number makes the sum greater than `mid`, start a new group.

---

## Example Feasibility Check

```text
nums = [7,2,5,10,8]
mid = 21
```

Process:

```text
7       → 7
7 + 2   → 9
9 + 5   → 14
14 + 10 → 24 > 21
```

Since `24 > 21`, start a new group:

```text
[7,2,5] | [10]
14      | 10
```

Then:

```text
10 + 8 = 18
```

Final:

```text
[7,2,5] | [10,8]
14      | 18
```

Number of groups:

```text
2
```

Since:

```text
2 <= k
```

`mid = 21` is feasible.

---

## Binary Search Logic

### Case 1: `groups > k`

We needed too many groups.

Therefore `mid` is too small.

```python
low = mid + 1
```

### Case 2: `groups <= k`

The current `mid` is feasible.

Try to find an even smaller feasible answer.

```python
high = mid
```

---

## Why `groups <= k`?

The problem asks for exactly `k` groups.

During the feasibility check, if we can make fewer than `k` groups, we can split existing groups further because all numbers are positive.

Therefore:

```text
groups <= k
```

means the candidate `mid` is feasible.

## Key Observations

### Search Range

```python
low = max(nums)
high = sum(nums)
```

### Feasibility

For a fixed `mid`, greedily create the minimum number of groups.

```python
if curr_sum + num > mid:
    groups += 1
    curr_sum = num
```

### Binary Search Direction

```python
if groups > k:
    low = mid + 1
else:
    high = mid
```

### Core Invariant

> `groups` represents the minimum number of contiguous subarrays required when no subarray can have a sum greater than `mid`.

### Greedy Principle

For a fixed `mid`, keep adding elements to the current subarray until adding the next element would exceed `mid`.

---

## Common Mistakes

- Binary searching array indices instead of the answer.
- Using `low = 0`.
- Using `high = len(nums) - 1`.
- Using `>=` instead of `>` when checking the sum.
- Starting `groups = 0` instead of `1`.
- Moving `low` and `high` in the wrong direction.
- Forgetting `return low`.
- Confusing `k` with the maximum allowed subarray sum.

---

## Interview Takeaway

This is a classic **Binary Search on Answer** problem.

The mental model is:

```text
Guess maximum allowed sum
        ↓
Greedily count required groups
        ↓
Is it feasible?
   ↓             ↓
  YES            NO
   ↓              ↓
Try smaller    Try larger
```

The goal is to find the **smallest feasible maximum subarray sum**.

### One-Line Revision

> **Binary search the smallest maximum allowed subarray sum, and use greedy partitioning to check whether that limit can be achieved with at most `k` groups.**


````python
class Solution(object):
    def splitArray(self, nums, k):
        """
        :type nums: List[int]
        :type k: int
        :rtype: int
        """
        low=max(nums)
        high=sum(nums)
        while low<high:
            mid=(low+high)//2
            curr_sum=0
            group=1
            for i in nums:
                if curr_sum+i>mid:
                    group+=1
                    curr_sum=i
                else:
                    curr_sum+=i
            if group>k:
                low=mid+1
            else:
                high=mid
        return low
            

        
````
