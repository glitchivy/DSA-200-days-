# LC 485 - Max Consecutive Ones

## Problem

Given a binary array `nums` containing only `0` and `1`, return the maximum number of consecutive `1`s.

### Example

Input:

`nums = [1,1,0,1,1,1]`

Consecutive groups of `1`s:

- `[1,1]` → 2
- `[1,1,1]` → 3

Output:

`3`

---

## Brute Force Approach

One way to think about the problem is:

1. Find where a group of `1`s starts.
2. Count how many consecutive `1`s follow it.
3. Keep track of the largest group.

My original solution tried to do this by detecting a `0 → 1` transition and then using another loop to count the streak.

The idea of finding a streak was reasonable, but this creates unnecessary repeated traversal.

### Problems in my original implementation

1. `range(1, len(nums))` skips index `0`, so a streak beginning at the first element can be missed.

2. The extra `while` loop is unnecessary because we can count the streak while already traversing the array.

3. If the streak reaches the end of the array, `val` can move outside the valid index range.

4. `return count` returns the last/current streak instead of the maximum streak. The required answer is `max_count`.

### Complexity of this style

Worst-case time can become `O(n²)` because the same elements can be traversed more than once.

---

## Optimal Approach

Use a single traversal with two variables:

- `current_count` = current consecutive streak of `1`s
- `max_count` = longest streak seen so far

For every element:

- If it is `1`, increase `current_count`.
- If it is `0`, reset `current_count` to `0`.
- After increasing the streak, update `max_count`.

The important idea is that we do not need to search for the beginning of every streak.

The traversal itself can maintain the streak.

---

## Key Observation

Whenever we see a `1`, the current streak continues:

`1 → current_count + 1`

Whenever we see a `0`, the streak is broken:

`0 → current_count = 0`

At the same time, we maintain the largest streak we have seen.

So the problem becomes:

`current streak + reset when broken + maximum streak`

---

## Algorithm

1. Initialize `current_count = 0`.
2. Initialize `max_count = 0`.
3. Traverse every element in `nums`.
4. If the element is `1`:
   - Increase `current_count`.
   - Update `max_count`.
5. If the element is `0`:
   - Reset `current_count` to `0`.
6. Return `max_count`.

---

## Dry Run

For:

`nums = [1,1,0,1,1,1]`

| Element | Current Count | Max Count |
|---|---:|---:|
| 1 | 1 | 1 |
| 1 | 2 | 2 |
| 0 | 0 | 2 |
| 1 | 1 | 2 |
| 1 | 2 | 2 |
| 1 | 3 | 3 |

Final answer:

`3`

---

## Primary Pattern

### Running Streak + Reset on Break + Global Maximum

This is the main pattern to remember.

Mental model:

`1 → increase streak`

`0 → reset streak`

`every time streak increases → update maximum`

---

## Generalization

This pattern is not limited to binary arrays.

For example:

### Longest consecutive 5s

If current element is `5`:

`current_count += 1`

Otherwise:

`current_count = 0`

### Longest increasing consecutive sequence

If the current element continues the increasing condition:

`current_count += 1`

Otherwise:

`current_count = 1`

The general idea is:

> Track the current valid run and reset or restart it whenever the condition breaks.

---

## Common Mistakes

### 1. Only checking `0 → 1`

A streak can begin at index `0`.

Example:

`[1,1,1,0]`

There is no previous `0`, but the answer is still `3`.

---

### 2. Forgetting to reset

For:

`[1,1,0,1]`

the two groups are separate.

After seeing `0`, the current streak must become `0`.

---

### 3. Returning the current streak

For:

`[1,1,1,0,1]`

at the end:

`current_count = 1`

but:

`max_count = 3`

The answer is `3`.

---

### 4. Using another loop unnecessarily

If you are already traversing the array, ask:

> Can I maintain the information I need during this traversal?

Here, the answer is yes.

---

## Complexity

Let `n` be the length of `nums`.

### Time Complexity

We visit every element exactly once.

`O(n)`

### Space Complexity

Only two variables are required.

`O(1)`

This is optimal.

---

## Interview Follow-Ups

### Why do we need two counters?

Because they represent different things:

`current_count` = the streak we are currently inside.

`max_count` = the best streak found anywhere so far.

---

### Why reset to 0?

Because a `0` completely breaks a consecutive sequence of `1`s.

The next `1` starts a new streak.

---

### Can we solve this with nested loops?

Yes, but it is unnecessary.

A one-pass solution is simpler and gives `O(n)` time.

## Key Takeaway

The important lesson is not simply "count the 1s."

The real pattern is:

> **When a problem asks for the longest consecutive run, maintain the current run and reset it whenever the condition breaks.**

Think:

**Current State → Condition Continues → Increase**

**Condition Breaks → Reset**

**Best State → Keep Maximum**

This pattern appears frequently in array and string problems.

```python
class Solution(object):
    def findMaxConsecutiveOnes(self, nums):
        """
        :type nums: List[int]
        :rtype: int
        """
        max_count=0
        count=0

        for i in nums:
            if i==1:
                count+=1
                max_count=max(max_count,count)
            else:
                count=0
        return max_count
            
```
