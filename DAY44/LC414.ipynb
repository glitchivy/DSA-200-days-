# LC 414 - Third Maximum Number

## Problem

Given an integer array `nums`, return the **third distinct maximum number** in the array.

If the third distinct maximum does not exist, return the maximum number.

### Important

The word **distinct** matters.

Example:

`nums = [2,2,3,1]`

Distinct values:

`[3,2,1]`

Third maximum:

`1`

If:

`nums = [1,2]`

There are only 2 distinct values, so return:

`2`

---

## Example

### Example 1

Input:

`[3,2,1]`

Output:

`1`

### Example 2

Input:

`[1,2]`

Output:

`2`

### Example 3

Input:

`[2,2,3,1]`

Output:

`1`

---

## Brute Force Approach

The simple approach is:

1. Remove duplicates using a set.
2. Sort the remaining values.
3. If there are at least 3 distinct values, return the third-largest.
4. Otherwise, return the largest.

### Complexity

Time: `O(n log n)`

Space: `O(n)`

This works, but sorting is unnecessary because we only care about the top 3 values.

---

## Optimal Approach

Use a set to remove duplicates, then maintain three variables:

- `first` = largest distinct value
- `second` = second-largest distinct value
- `third` = third-largest distinct value

For every current number:

### Case 1: `current > first`

The current number becomes the new largest.

Shift:

`third ← second`

`second ← first`

`first ← current`

### Case 2: `current > second`

The current number becomes the new second-largest.

Shift:

`third ← second`

`second ← current`

### Case 3: `current > third`

The current number becomes the new third-largest.

`third ← current`

---

## Key Observation

We do not need to sort the entire array.

We only care about:

`1st largest → 2nd largest → 3rd largest`

So we can maintain these three positions while traversing.

The important pattern is:

**Top K Tracking**

For this problem:

`K = 3`

---

## Algorithm

1. Convert `nums` into a set to remove duplicates.
2. If there are fewer than 3 distinct values, return the maximum.
3. Initialize:
   - `first = -∞`
   - `second = -∞`
   - `third = -∞`
4. Traverse every distinct number.
5. If current number is greater than `first`:
   - move `second` to `third`
   - move `first` to `second`
   - put current number in `first`
6. Else if current number is greater than `second`:
   - move `second` to `third`
   - put current number in `second`
7. Else if current number is greater than `third`:
   - put current number in `third`
8. Return `third`.

---

## Dry Run

For:

`nums = [2,2,3,1]`

After removing duplicates:

`{1,2,3}`

Start:

`first = -∞`

`second = -∞`

`third = -∞`

### Current = 1

`1 > first`

So:

`first = 1`

`second = -∞`

`third = -∞`

### Current = 2

`2 > first`

Shift:

`third = second`

`second = first`

`first = 2`

Now:

`first = 2`

`second = 1`

`third = -∞`

### Current = 3

`3 > first`

Shift:

`third = second`

`second = first`

`first = 3`

Now:

`first = 3`

`second = 2`

`third = 1`

Answer:

`1`

## Primary Pattern

### Top K Tracking

The main pattern is:

**Maintain the top K values while traversing the array.**

For this problem:

`K = 3`

Mental model:

`current value`

↓

`Does it belong in the top 3?`

↓

`Yes → shift the existing rankings`

↓

`Update first / second / third`

---

## Pattern Recognition

When you see:

- largest
- second largest
- third largest
- top K
- Kth largest
- Kth smallest

ask:

> **Do I really need to sort everything, or can I maintain the required extreme values while traversing?**

For a small fixed `K`, tracking the top values directly can be very effective.

---

## Generalization

The same idea works for other fixed K problems.

### Top 1

Track:

`first`

### Top 2

Track:

`first, second`

### Top 3

Track:

`first, second, third`

### General Top K

For larger K, maintaining many variables becomes inconvenient.

This is where data structures such as a **heap** can become useful.

The general pattern is:

**Traversal → Compare → Insert into ranking → Shift/update**

---

## Common Mistakes

### 1. Forgetting "distinct"

For:

`[5,5,4,3]`

The values are:

`5,5,4,3`

But the distinct values are:

`5,4,3`

Therefore the third maximum is:

`3`

This is why using a set first is useful.

---

### 2. Confusing the set with the current element

In my original code:

`num = set(nums)`

and:

`for i in num`

Here:

- `num` = entire set
- `i` = current element

So comparisons inside the loop must use `i`.

For example:

`if i > first`

not:

`if num > first`

This was the main implementation error in my solution.

---

### 3. Returning the set

The set is only used to remove duplicates.

The final answer is:

`third`

not the entire set.

---

### 4. Forgetting the middle cases

It is not enough to check:

`current > first`

A number can also belong between:

`first and second`

or:

`second and third`

So all three cases must be handled.

---

### 5. Forgetting to shift values

If a new maximum appears:

`old first → second`

`old second → third`

The existing rankings must move down.

---

## Complexity

Let `n` be the number of elements.

### Time

Creating the set:

`O(n)` average

Traversing the distinct values:

`O(n)` average

Total:

**O(n) average time**

### Space

The set stores distinct values:

**O(n) auxiliary space**

The three ranking variables themselves use:

**O(1)** space.

---

## Interview Follow-Ups

### Q1. Why use a set?

Because the problem asks for the third **distinct** maximum.

Duplicates cannot occupy multiple ranking positions.

---

### Q2. Why not sort?

Sorting takes:

`O(n log n)`

But we only need the top 3 distinct values.

Tracking them directly gives:

`O(n)` average time.

---

### Q3. Why use negative infinity?

We need an initial value smaller than any valid number.

`float("-inf")` guarantees that the first real value can replace it.

---

### Q4. What if there are fewer than 3 distinct values?

Return the maximum.

Example:

`[1,2]`

There is no third distinct maximum.

Answer:

`2`

## Key Takeaway

The main lesson is not simply how to find the third maximum.

The important pattern is:

> **When you only care about a small number of extreme values, you often do not need to sort the entire array. Maintain the top K values while traversing.**

For this problem:

`Remove duplicates`

↓

`Track first, second, third`

↓

`Shift rankings when necessary`

↓

`Return third`

Also remember the implementation lesson:

> **Always know exactly what each variable represents.**

`unique_nums` = collection

`num` = current element

`first / second / third` = ranking states

Clear variable roles prevent many implementation errors.

```python
class Solution(object):
    def thirdMax(self, nums):
        """
        :type nums: List[int]
        :rtype: int
        """
        num=set(nums)
        n=len(num)
        if n<3:
            return max(nums)
        first = float("-inf")
        second = float("-inf")
        third = float("-inf")
        for i in num:
            if i>first:
                third=second
                second=first
                first=i
            elif i > second:
                third = second
                second = i

            elif i > third:
                third =i
        return third
        
```
