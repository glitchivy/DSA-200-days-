# LC 1572 - Matrix Diagonal Sum

## Problem

Given a square matrix, return the sum of the elements on both diagonals.

The matrix is square, meaning:

`number of rows = number of columns`

### Example

Input:

`[[1,2,3],[4,5,6],[7,8,9]]`

The two diagonals are:

`1 + 5 + 9`

and:

`3 + 5 + 7`

The center `5` belongs to both diagonals, but it should only be counted **once**.

Therefore:

`1 + 5 + 9 + 3 + 7 = 25`

Output:

`25`

---

# Step 1: Understand the Main Diagonal

The main diagonal goes:

`top-left → bottom-right`

Example:

`[1] 2  3`
` 4 [5] 6`
` 7  8 [9]`

Coordinates:

`(0,0)`

`(1,1)`

`(2,2)`

Notice:

> Row index = Column index

Therefore:

`matrix[i][i]`

represents the main diagonal.

---

# Step 2: Understand the Secondary Diagonal

The secondary diagonal goes:

`top-right → bottom-left`

Example:

`1  2 [3]`
`4 [5] 6`
`[7] 8  9`

Coordinates:

`(0,2)`

`(1,1)`

`(2,0)`

For an `n × n` matrix, the column is:

`n - 1 - i`

Therefore:

`matrix[i][n - 1 - i]`

represents the secondary diagonal.

---

# Key Coordinate Patterns

## Main Diagonal

`(i, i)`

The row and column are equal.

## Secondary Diagonal

`(i, n - 1 - i)`

The row and column add up to:

`n - 1`

This is a very important matrix pattern.

---

# Primary Pattern

## 2D Array Traversal + Coordinate-Based Access + Accumulator

The general pattern is:

`Traverse rows`
→
`Calculate coordinates`
→
`Access required cells`
→
`Add to accumulator`

---

# Algorithm

1. Find the size of the matrix.
2. Create `total = 0`.
3. Traverse every row using one loop.
4. Add the main diagonal element:

`matrix[i][i]`

5. Add the secondary diagonal element:

`matrix[i][n - 1 - i]`

6. If the matrix has an odd size, the center was counted twice.
7. Subtract the center once.
8. Return `total`.

---

# Why Does the Center Get Counted Twice?

Consider:

`1 2 3`
`4 5 6`
`7 8 9`

Main diagonal:

`1, 5, 9`

Secondary diagonal:

`3, 5, 7`

The `5` appears in both.

If we simply add both:

`1 + 5 + 9 + 3 + 5 + 7 = 30`

But the correct answer is:

`25`

So:

`30 - 5 = 25`

---

# Detecting the Center

The matrix has a center only when its size is odd.

Check:

`n % 2 == 1`

The center index is:

`n // 2`

Therefore, the center element is:

`matrix[n // 2][n // 2]`

So when `n` is odd:

`total -= matrix[n // 2][n // 2]`

---

# Dry Run

Matrix:

`1 2 3`
`4 5 6`
`7 8 9`

`n = 3`

`total = 0`

### i = 0

Main:

`matrix[0][0] = 1`

Secondary:

`matrix[0][2] = 3`

Total:

`4`

### i = 1

Main:

`matrix[1][1] = 5`

Secondary:

`matrix[1][1] = 5`

The center has been counted twice.

`matrix[2][2] = 9`

`matrix[2][0] = 7`

`30`

`matrix[1][1] = 5`

Subtract:

`30 - 5 = 25`

Final answer:

`25`

# One-Loop Insight

A common beginner approach is to think:

`Loop through main diagonal`

then:

`Loop through secondary diagonal`

But we don't need two loops.

For every row `i`, we can calculate both diagonal positions:

`matrix[i][i]`

and:

`matrix[i][n - 1 - i]`

Therefore, one loop is enough.

This is an important optimization in **thinking**, even though both versions would still be `O(n)`.

---

Think:

When you see:

> "Find the opposite/secondary diagonal"

Think:

`matrix[i][n - 1 - i]`

When you see:

> "Sum both diagonals"

Think:

`main + secondary`

Then immediately ask:

> "Can the center be counted twice?"

If `n` is odd:

`Yes`

Therefore:

`subtract center once`

# Alternative Way to Understand the Secondary Diagonal

There is another useful way to recognize it.

For every cell on the secondary diagonal:

`row + column = n - 1`

For a `3 × 3` matrix:

`(0,2) → 0 + 2 = 2`

`(1,1) → 1 + 1 = 2`

`(2,0) → 2 + 0 = 2`

Since:

`n - 1 = 2`

This gives another mental model:

### Main diagonal

`row = column`

### Secondary diagonal

`row + column = n - 1`

These relationships are extremely useful in matrix problems.

---

# Common Mistakes

### 1. Forgetting `n - 1 - i`

Wrong:

`matrix[i][n - i]`

Correct:

`matrix[i][n - 1 - i]`

The `-1` is necessary because indices start from `0`.

---

### 2. Forgetting the center duplication

For odd-sized matrices, the center belongs to both diagonals.

If you add both diagonals directly, you count it twice.

---

### 3. Subtracting the center for an even-sized matrix

An even-sized matrix has no single center cell.

Therefore, only subtract when:

`n % 2 == 1`

---

### 4. Thinking two diagonals require two loops

They don't.

Both coordinates can be calculated during the same traversal.

---

### 5. Confusing the two diagonals

Main:

`(i, i)`

Secondary:

`(i, n - 1 - i)`

Memorize the **coordinate relationship**, not just the formula.

---

# Complexity

For an `n × n` matrix:

We traverse the matrix by rows once.

Each iteration performs constant-time operations.

Therefore:

**Time: O(n)**

**Auxiliary Space: O(1)**

We don't create another matrix or array.

---

# Important Complexity Insight

The matrix contains `n²` elements, but we do **not** need to visit all `n²` elements.

We only visit:

`n` elements from the main diagonal

and:

`n` elements from the secondary diagonal.

Therefore, the work is:

`O(n)`

not:

`O(n²)`.

This is an important matrix optimization pattern:

> **If the problem asks about a specific structure inside a matrix, don't automatically traverse the entire matrix.**

# Interview Follow-Ups

### Q1. Why is the solution O(n) instead of O(n²)?

Because we only visit the two diagonals rather than every cell.

### Q2. Why do we need `n - 1 - i`?

Because the secondary diagonal moves from the last column toward the first column as the row increases.

### Q3. When is the center duplicated?

When `n` is odd.

### Q4. How do you find the center?

`n // 2`

### Q5. How do you prevent double counting?

Either:

`subtract the center once`

or:

`don't add the secondary diagonal when i is the center`

### Q6. Can this be done with one loop?

Yes.

For each `i`, calculate both diagonal positions.

```python
class Solution(object):
    def diagonalSum(self, mat):
        """
        :type mat: List[List[int]]
        :rtype: int
        """
        n=len(mat)
        final_sum=0
        for i in range(n):
            #primary
            final_sum+=mat[i][i]
            final_sum+=mat[i][n-1-i]
        if n%2==1:
            final_sum-=mat[n//2][n//2]
        return final_sum
```
