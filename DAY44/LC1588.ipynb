# LC 1588 - Sum of All Odd Length Subarrays

## 1. Problem

Given an array `arr`, find the sum of **all subarrays whose length is odd**.

### Important

A **subarray** is a contiguous part of an array.

For example:

`arr = [1, 4, 2]`

All subarrays are:

- `[1]`
- `[4]`
- `[2]`
- `[1,4]`
- `[4,2]`
- `[1,4,2]`

Odd-length subarrays:

- `[1]` → length 1
- `[4]` → length 1
- `[2]` → length 1
- `[1,4,2]` → length 3

Answer:

`1 + 4 + 2 + 7 = 14`

---

# 2. What Is a Subarray?

A subarray must be:

- **contiguous**
- maintain the original order

For:

`[1, 2, 3]`

Valid subarrays:

- `[1]`
- `[2]`
- `[3]`
- `[1,2]`
- `[2,3]`
- `[1,2,3]`

`[1,3]` is NOT a subarray because we skipped `2`.

---

# 3. Brute Force Approach

The most direct way is:

1. Choose a starting index.
2. Choose an ending index.
3. Calculate the sum of that subarray.
4. Calculate its length.
5. If the length is odd, add the sum to the answer.

If we calculate every subarray sum from scratch, we need three levels of work:

- Start index
- End index
- Traverse the subarray to calculate its sum

### Complexity

Time: **O(n³)**

Space: **O(1)**

This works, but there is unnecessary repeated work.

---

# 4. Better Approach: Fixed Start + Expand End + Running Sum

We can remove one entire loop.

Instead of calculating the sum of every subarray from scratch, maintain a **running sum**.

For each starting index `i`:

```text
current_sum = 0

Move j from i to the end:
    current_sum += arr[j]

    current_sum now represents:
    arr[i ... j]

class Solution(object):
    def sumOddLengthSubarrays(self, arr):
        """
        :type arr: List[int]
        :rtype: int
        """
        final=0
        n=len(arr)
        for i in range(n):
            cs=0
            for j in range(i,n):
                cs+=arr[j]
                l=j-i+1
                if l %2==1:
                    final+=cs
        return final
