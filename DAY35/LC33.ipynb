# LeetCode 33 - Search in Rotated Sorted Array

## Problem

Given a rotated sorted array `nums` and an integer `target`, return the index of the target if it exists.

If the target is not present, return:

```
-1
```

The solution must run in **O(log n)** time.

---

## Brute Force Approach

### Idea

Traverse the array from left to right.

- If the target is found, return its index.
- If the traversal finishes without finding the target, return `-1`.

### Algorithm

1. Traverse every element.
2. Compare it with the target.
3. Return its index if found.
4. Otherwise return `-1`.

### Time Complexity

**O(n)**

### Space Complexity

**O(1)**

---

## Optimal Approach (Modified Binary Search)

### Key Observation

The entire array is not sorted because it has been rotated.

Example:

```
4 5 6 7 0 1 2
```

However, **at least one half is always sorted**.

During every Binary Search iteration:

- Either the left half is sorted.
- Or the right half is sorted.

After identifying the sorted half, check whether the target lies inside that half.

- If yes, search there.
- Otherwise, search the other half.

---

## Algorithm

### Step 1

Initialize:

```
low = 0
high = n - 1
```

---

### Step 2

Repeat while:

```
low <= high
```

Compute:

```
mid = (low + high) // 2
```

---

### Step 3

If:

```
nums[mid] == target
```

Return:

```
mid
```

---

### Step 4

Check whether the left half is sorted.

```
nums[low] <= nums[mid]
```

If true, the left half is sorted.

Now check whether the target lies inside it.

```
nums[low] <= target < nums[mid]
```

- If true, search the left half.

```
high = mid - 1
```

- Otherwise search the right half.

```
low = mid + 1
```

---

### Step 5

Otherwise, the right half is sorted.

Check whether the target lies inside it.

```
nums[mid] < target <= nums[high]
```

- If true, search the right half.

```
low = mid + 1
```

- Otherwise search the left half.

```
high = mid - 1
```

---

### Step 6

If the loop finishes without finding the target,

return

```
-1
```

---

## Time Complexity

Each Binary Search iteration removes half of the remaining search space.

```
O(log n)
```

---

## Space Complexity

Only constant extra space is used.

```
O(1)
```

---

## Primary Pattern

**Modified Binary Search**

---

## Recognition Clues

Look for problems where:

- The array is sorted but rotated.
- Search is required.
- The required complexity is **O(log n)**.
- A normal Binary Search does not work directly.

Think:

```
Modified Binary Search
```

---

## Key Observations

- The entire array is not sorted.
- At least one half is always sorted.
- First determine which half is sorted.
- Then check whether the target belongs to that half.
- Discard the half that cannot contain the target.

---

## Common Mistakes

- Using a normal Binary Search directly.

- Forgetting to check whether the target is found before determining the sorted half.

- Using incorrect comparison operators while checking the sorted half.

- Searching the wrong half after identifying the sorted portion.

- Forgetting that one half is always sorted.

---

## Interview Takeaway

This problem is a classic application of **Modified Binary Search**. Although the entire array is rotated, one of the two halves is always sorted. By identifying the sorted half in each iteration and checking whether the target lies within its range, half of the search space can be discarded every time. This preserves the **O(log n)** time complexity while correctly searching a rotated sorted array.

````python
class Solution(object):
    def search(self, nums, target):
        """
        :type nums: List[int]
        :type target: int
        :rtype: int
        """
        low=0
        high=len(nums)-1
        while low<=high:
            mid=(low+high)//2
            if nums[mid]==target:
                return mid
            if nums[low]<=nums[mid]:
                if nums[low]<=target<=nums[mid]:
                    high=mid-1
                else:
                    low=mid+1
            else:
                if nums[mid]<=target<=nums[high]:
                    low=mid+1
                else:
                    high=mid-1
        return -1
````
