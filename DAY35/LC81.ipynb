# LeetCode 81 - Search in Rotated Sorted Array II

## Problem

Given a rotated sorted array `nums` that **may contain duplicate elements**, determine whether a given `target` exists in the array.

Return:

```
True
```

if the target exists, otherwise return

```
False
```

---

## Brute Force Approach

### Idea

Traverse the array from left to right.

- If the target is found, return `True`.
- If the traversal finishes without finding the target, return `False`.

### Algorithm

1. Traverse every element.
2. Compare it with the target.
3. Return `True` if found.
4. Otherwise return `False`.

### Time Complexity

**O(n)**

### Space Complexity

**O(1)**

---

## Optimal Approach (Modified Binary Search)

### Key Observation

This problem is almost identical to **LeetCode 33**.

The only difference is that **duplicate values are allowed**.

Normally, one half of the array is always sorted.

However, duplicates can create ambiguity.

Example:

```
1 0 1 1 1
L   M   H
```

Here,

```
nums[low] == nums[mid] == nums[high]
```

Even though

```
nums[low] <= nums[mid]
```

the left half is **not actually sorted**.

Therefore, we cannot determine which half is sorted.

To remove the ambiguity, shrink the search space.

```
low += 1

high -= 1
```

Then continue the Binary Search.

---

## Algorithm

### Step 1

Initialize:

```
low = 0

high = n - 1
```

---

### Step 2

Repeat while:

```
low <= high
```

Compute:

```
mid = (low + high) // 2
```

---

### Step 3

If

```
nums[mid] == target
```

Return

```
True
```

---

### Step 4

Handle duplicates.

If

```
nums[low] == nums[mid] == nums[high]
```

then

```
low += 1

high -= 1
```

Continue to the next iteration.

---

### Step 5

Determine which half is sorted.

If

```
nums[low] <= nums[mid]
```

the left half is sorted.

Check whether

```
nums[low] <= target < nums[mid]
```

- If true

```
high = mid - 1
```

- Otherwise

```
low = mid + 1
```

---

### Step 6

Otherwise, the right half is sorted.

Check whether

```
nums[mid] < target <= nums[high]
```

- If true

```
low = mid + 1
```

- Otherwise

```
high = mid - 1
```

---

### Step 7

If the loop finishes without finding the target,

return

```
False
```

---

```
nums = [2,5,6,0,0,1,2]

target = 0
```

### Duplicate Example

```
nums = [1,0,1,1,1]

target = 0
```

Initially

```
low = 0

mid = 2

high = 4
```

Values

```
nums[low] = 1

nums[mid] = 1

nums[high] = 1
```

Since

```
nums[low] == nums[mid] == nums[high]
```

we cannot determine the sorted half.

Shrink the search space.

```
low += 1

high -= 1
```

Continue Binary Search until the ambiguity disappears.

## Time Complexity

### Average Case

```
O(log n)
```

### Worst Case

When the array contains many duplicate elements,

for example

```
[1,1,1,1,1,1]
```

the search space shrinks one element at a time.

Worst-case complexity becomes

```
O(n)
```

## Space Complexity

```
O(1)
```

## Common Mistakes

- Forgetting the duplicate handling condition.

- Checking the sorted half before removing duplicate ambiguity.

- Using incorrect comparison operators.

- Forgetting to use

```
continue
```

after shrinking both pointers.

- Assuming the worst-case time complexity is always `O(log n)`.

---

## Interview Takeaway

LeetCode 81 is an extension of **LeetCode 33**. The Binary Search logic remains the same except for one additional case. When the values at `low`, `mid`, and `high` are equal, it becomes impossible to determine which half is sorted. In that situation, shrink the search space by moving both pointers inward and continue searching. This modification correctly handles duplicate values while preserving **O(log n)** performance on average, although the worst-case complexity can degrade to **O(n)**.

````python
class Solution(object):
    def search(self, nums, target):
        """
        :type nums: List[int]
        :type target: int
        :rtype: bool
        """
        low=0
        high=len(nums)-1
        while low<=high:
            mid=(low+high)//2
            if nums[mid]==target:
                return True
            if nums[low]==nums[mid]==nums[high]:
                low+=1
                high-=1
                continue
            if nums[low]<=nums[mid]:
                if nums[low]<=target<nums[mid]:
                    high=mid-1
                else:
                    low=mid+1
            else:
                if nums[mid]<=target<=nums[high]:
                    low=mid+1
                else:
                    high=mid-1
        return False

````
