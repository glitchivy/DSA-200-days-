# LC 2114 - Maximum Number of Words Found in Sentences

## Problem

Given an array of strings `sentences`, where each string represents a sentence, return the maximum number of words found in any single sentence.

Words are separated by spaces.

### Example

Input:

`sentences = ["alice and bob love coding", "i love leetcode", "hello world"]`

Output:

`5`

Explanation:

- `"alice and bob love coding"` → 5 words
- `"i love leetcode"` → 3 words
- `"hello world"` → 2 words

Maximum = `5`

---

# First Attempt

The initial approach was:

`sentence → traverse every character → count`

This was incorrect because the problem asks for **words**, not characters.

For example:

`"alice and bob"`

contains:

- 13 characters
- 3 words

So character traversal does not solve the required problem.

---

# Correct Approach

## String Splitting + Counting + Maximum

For every sentence:

1. Split the sentence into individual words.
2. Count how many words were produced.
3. Keep track of the maximum word count.

Mental model:

`sentence`
→
`split into words`
→
`count words`
→
`update maximum`

---

# Algorithm

1. Initialize `value = 0`.
2. Traverse every sentence.
3. Split the sentence into words.
4. Find the number of words.
5. Update the maximum.
6. Return the maximum.

The important operations are:

`sentence.split()`

and:

`len(...)`

# Cleaner Solution

The manual counting loop:

`for word in words → count += 1`

is unnecessary once we have the list of words.

Instead:

`len(sentence.split())`

directly gives the number of words.

So the thought process becomes:

`sentence → split() → len() → max`

This is cleaner without changing the overall asymptotic time complexity.

---

# More Space-Efficient Approach

Because the problem guarantees that words are separated by spaces, we can use another observation:

> Number of words = number of spaces + 1

Example:

`"alice and bob"`

There are 2 spaces.

Therefore:

`2 + 1 = 3 words`

So we can count spaces instead of creating a list of words.

Conceptually:

`sentence.count(" ") + 1`

---

# Why This Uses Less Space

### Using `split()`

`"alice and bob"`

↓

`["alice", "and", "bob"]`

A new list is created.

Therefore, extra space can grow with the sentence.

### Using `count(" ")`

The sentence is scanned and the number of spaces is counted.

No list of words needs to be created.

Therefore, auxiliary space is constant.

---

# Complexity

Let `C` be the total number of characters across all sentences.

## Using `split()`

- **Time:** `O(C)`
- **Auxiliary Space:** `O(C)` in the worst case

The temporary list created by `split()` can contain many words.

## Using `count(" ")`

- **Time:** `O(C)`
- **Auxiliary Space:** `O(1)`

Both approaches are linear in the amount of input.

The space complexity is where they differ.

---

# Common Mistakes

### 1. Counting characters instead of words

A sentence contains both letters and spaces.

Characters ≠ words.

---

### 2. Forgetting spaces

If using the space-counting approach:

`words = spaces + 1`

For:

`"hello world"`

There is 1 space.

Therefore:

`1 + 1 = 2 words`

---

### 3. Using the space trick without checking the constraints

The `spaces + 1` method works here because the problem guarantees words are separated by spaces.

For arbitrary text, there could be:

- Multiple spaces
- Tabs
- Newlines
- Leading/trailing whitespace

In those cases, `split()` is safer.

---

### 4. Manually counting a list

After:

`words = sentence.split()`

you already know the number of words:

`len(words)`

There is no need for another loop.

Yes.

Because of the problem's constraints:

```python
class Solution(object):
    def mostWordsFound(self, sentences):
        """
        :type sentences: List[str]
        :rtype: int
        """
        value=0

        for sentence in sentences:
            count = len(sentence.split())
            value = max(value, count)

        return value
```
