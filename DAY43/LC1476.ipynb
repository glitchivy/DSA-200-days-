# LC 1476 - Subrectangle Queries

## Problem

You are given a 2D integer array `rectangle`.

You need to support two operations:

1. **Update a subrectangle**
   - Given two corners `(row1, col1)` and `(row2, col2)`, change every value inside that rectangle to `newValue`.

2. **Get a value**
   - Given `(row, col)`, return the current value at that position.

---

# Understanding the Grid

Consider:

`rectangle = [[1, 2, 1], [4, 3, 4], [3, 2, 1]]`

The coordinates are:

| Cell | Value |
|---|---:|
| `(0,0)` | 1 |
| `(0,1)` | 2 |
| `(0,2)` | 1 |
| `(1,0)` | 4 |
| `(1,1)` | 3 |
| `(1,2)` | 4 |
| `(2,0)` | 3 |
| `(2,1)` | 2 |
| `(2,2)` | 1 |

So:

`rectangle[row][col]`

accesses a specific cell.

---

# What Is a Subrectangle?

A subrectangle is a rectangular region inside the matrix.

It is defined by two opposite corners:

`(row1, col1)`

and

`(row2, col2)`

For example:

`row1 = 0`

`col1 = 1`

`row2 = 2`

`col2 = 2`

selects:

`(0,1), (0,2)`

`(1,1), (1,2)`

`(2,1), (2,2)`

So:

`1 2 1`
`4 3 4`
`3 2 1`

becomes:

`1 9 9`
`4 9 9`
`3 9 9`

if `newValue = 9`.

---

# Primary Pattern

## 2D Array Traversal + Coordinate-Based Update

The core pattern is:

`Rectangle boundaries`
→
`Nested loops`
→
`Visit every cell inside the boundaries`
→
`Modify the cell`

This is a direct extension of the 2D traversal pattern from LC 1672.

---

# Algorithm

## `__init__`

Store the given matrix inside the object:

`self.rectangle = rectangle`

Why?

Because the matrix needs to **remember modifications between function calls**.

---

## `updateSubrectangle`

Given:

`row1, col1, row2, col2, newValue`

1. Traverse rows from `row1` to `row2`.
2. For every row, traverse columns from `col1` to `col2`.
3. Replace every cell with `newValue`.

Mental model:

`for each selected row`
→
`for each selected column`
→
`modify that cell`

---

## `getValue`

Given:

`row, col`

Simply return:

`rectangle[row][col]`

No traversal is required.

---

# Why `row2 + 1` and `col2 + 1`?

Python's `range()` excludes the ending value.

If we want:

`0, 1, 2`

we need:

`range(0, 3)`

Therefore:

`range(row1, row2 + 1)`

includes `row2`.

Similarly:

`range(col1, col2 + 1)`

includes `col2`.

This is a very common source of off-by-one errors.

---

# Dry Run

Initial matrix:

`1 2 1`
`4 3 4`
`3 2 1`

Operation:

`updateSubrectangle(0, 1, 2, 2, 9)`

### Outer loop

Rows visited:

`0, 1, 2`

### Inner loop

For each row:

Columns visited:

`1, 2`

Therefore cells visited:

`(0,1), (0,2)`

`(1,1), (1,2)`

`(2,1), (2,2)`

Each becomes `9`.

Final matrix:

`1 9 9`
`4 9 9`
`3 9 9`

Now:

`getValue(1,2)`

returns:

`9`

---

# Important Python Concept: `self`

The class needs to remember the matrix.

When the object is created:

`obj = SubrectangleQueries(rectangle)`

Python calls:

`__init__`

and stores:

`self.rectangle = rectangle`

Later:

`obj.updateSubrectangle(...)`

modifies:

`self.rectangle`

Later still:

`obj.getValue(...)`

reads from that same modified matrix.

Therefore, `self` allows the object to maintain **state between method calls**.

---

# Common Mistake

## Defining methods inside another method

Incorrect structure:

`class`
→
`updateSubrectangle`
→
`__init__`
→
`another updateSubrectangle`

The three methods need to be at the same class level:

`class`
→ `__init__`

→ `updateSubrectangle`

→ `getValue`

This was the Python-specific issue in the first attempt.

The DSA logic itself was correct.

---

# Another Common Mistake

Forgetting that the rectangle boundaries are inclusive.

If:

`row1 = 1`

`row2 = 3`

you need:

`1, 2, 3`

not:

`1, 2`

Therefore:

`range(row1, row2 + 1)`

---

# Complexity

Suppose the selected rectangle has:

`R = row2 - row1 + 1`

rows and:

`C = col2 - col1 + 1`

columns.

Then the number of cells updated is:

`R × C`

Therefore:

### Update

**Time: O(R × C)**

### Get Value

**Time: O(1)**

Direct array access requires no traversal.

### Auxiliary Space

**O(1)**

We modify the existing matrix instead of creating another matrix.

---

# Why We Don't Need a Fancy Optimization

This is an important interview lesson.

Sometimes you see a 2D update problem and immediately think:

- Prefix sums
- Difference arrays
- Lazy updates
- Segment trees
- 2D Fenwick trees

Do not jump there automatically.

First check the constraints.

For LC 1476, the straightforward approach is sufficient.

The correct interview mindset is:

> **Use the simplest algorithm that satisfies the constraints.**

Optimization is useful only when the constraints require it.

# Pattern Recognition

When you see:

`row1, col1, row2, col2`

immediately visualize:

`top-left corner`
+
`bottom-right corner`

Then ask:

> "Do I need to process every cell inside this rectangle?"

If yes:

**Nested loops over the selected rows and columns.**

Template:

`for row in row1 → row2`
`    for col in col1 → col2`
`        process matrix[row][col]`

The operation inside the loop could be:

- Update
- Sum
- Count
- Search
- Compare
- Mark
- Modify

# Interview Follow-Ups

### Q1. Why do we need `self.rectangle`?

Because the matrix must retain its modified state across multiple method calls.

### Q2. Why is `getValue()` O(1)?

Because it directly accesses:

`rectangle[row][col]`

### Q3. Why is the update O(R × C)?

Because every cell inside the selected rectangle must potentially be changed.

### Q4. Can we update the entire matrix?

Yes, but that would do unnecessary work.

We only traverse:

`row1 → row2`

and:

`col1 → col2`

### Q5. Do we need another matrix?

No.

We can modify the original matrix in place.

### Q6. What is the main pattern?

**2D Coordinate-Based Traversal.**

```python
class SubrectangleQueries(object):

    def __init__(self, rectangle):
        self.rectangle = rectangle

    def updateSubrectangle(self, row1, col1, row2, col2, newValue):
        for i in range(row1, row2 + 1):
            for j in range(col1, col2 + 1):
                self.rectangle[i][j] = newValue

    def getValue(self, row, col):
        return self.rectangle[row][col]
```
