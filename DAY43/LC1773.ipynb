# LC 1773 - Count Items Matching a Rule

## Problem

You are given a list of items.

Each item contains exactly 3 values:

`[type, color, name]`

You are also given:

- `ruleKey` → tells you which property to check
- `ruleValue` → tells you what value that property should have

Return the number of items that match the given rule.

### Example

Input:

`items = [["phone","blue","pixel"],["computer","silver","lenovo"],["phone","gold","iphone"]]`

`ruleKey = "type"`

`ruleValue = "phone"`

Output:

`2`

Explanation:

- `["phone","blue","pixel"]` → type is phone → match
- `["computer","silver","lenovo"]` → type is computer → no match
- `["phone","gold","iphone"]` → type is phone → match

Answer = `2`

---

# Key Observation

Every item has a fixed structure:

| Index | Property |
|---:|---|
| `0` | type |
| `1` | color |
| `2` | name |

The `ruleKey` is a string, but the actual data is stored using indices.

Therefore, we need to convert:

`ruleKey → corresponding index`

We can use a dictionary:

`"type" → 0`

`"color" → 1`

`"name" → 2`

Then we can access the required property using:

`item[index[ruleKey]]`

---

# Primary Pattern

## Array Traversal + Mapping + Conditional Counting

The overall pattern is:

`Traverse`
→
`Find relevant property`
→
`Compare`
→
`Count`

This is a very common pattern when working with structured data.

---

# Algorithm

1. Create a dictionary mapping each rule key to its index.
2. Initialize `count = 0`.
3. Traverse every item.
4. Use `ruleKey` to determine which index to inspect.
5. Compare that value with `ruleValue`.
6. If they match, increment `count`.
7. Return `count`.

---

# Understanding the Important Expression

The most important part of this problem is:

`item[index[ruleKey]]`

Break it into two steps.

### Step 1: Find the index

Suppose:

`ruleKey = "color"`

Then:

`index[ruleKey]`

gives:

`1`

### Step 2: Access the item

If:

`item = ["phone", "blue", "pixel"]`

then:

`item[1]`

gives:

`"blue"`

Therefore:

`item[index[ruleKey]]`

means:

`item[1]`

which gives:

`"blue"`.

This is called **indirect indexing**.

# Common Mistakes

### 1. Checking the wrong index

Remember:

`type = 0`

`color = 1`

`name = 2`

---

### 2. Comparing the entire item

You don't want:

`item == ruleValue`

You only want the property specified by `ruleKey`.

---

### 3. Using `<=` or other comparisons

The problem asks for an exact match.

Use:

`item[index[ruleKey]] == ruleValue`

---

### 4. Forgetting to reset or initialize the counter

The counter should start at:

`count = 0`

and increase only when a match is found.

---

# Complexity

Let `n` be the number of items.

Each item is inspected exactly once.

- **Time:** `O(n)`
- **Auxiliary Space:** `O(1)`

The dictionary contains only 3 fixed mappings, so its size is constant.

The output is just a single integer.

# Interview Follow-Ups

### Q1. Why use a dictionary?

Because `ruleKey` is a string and we need to convert it into the corresponding array index.

### Q2. Can you solve it without a dictionary?

Yes.

You could use conditional statements:

`if ruleKey == "type"`

`elif ruleKey == "color"`

`else`

But the dictionary is cleaner and scales better if there are more properties.

### Q3. Why is the solution O(n)?

We visit every item exactly once.

The dictionary lookup and array access are constant-time operations.

### Q4. What pattern is this?

**Mapping + Array Traversal + Conditional Counting**

### Q5. What is the most important expression?

`item[index[ruleKey]]`

Because it demonstrates indirect indexing.

```python
class Solution(object):
    def countMatches(self, items, ruleKey, ruleValue):
        """
        :type items: List[List[str]]
        :type ruleKey: str
        :type ruleValue: str
        :rtype: int
        """
        index={
            'type':0,
            'color':1,
            'name':2
        }
        count=0
        for i in items:
            if i[index[ruleKey]]==ruleValue:
                count+=1
        return count
```
