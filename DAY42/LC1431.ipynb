# LeetCode 1431 - Kids With the Greatest Number of Candies

## Problem

You are given an integer array `candies`, where `candies[i]` represents the number of candies the `i`th kid has.

You are also given `extraCandies`.

For each kid, determine whether giving all `extraCandies` to that kid would make them have **the greatest number of candies** among all kids.

A kid is considered capable if:

`candies[i] + extraCandies >= maximum candies currently held`

Return a boolean array representing the answer for every kid.

---

## Example

Input:

`candies = [2, 3, 5, 1, 3]`  
`extraCandies = 3`

Maximum candies:

`5`

Check each kid:

| Kid | Candies | After Extra Candies | `>= 5` | Result |
|----:|--------:|--------------------:|:------:|:------:|
| 0 | 2 | 5 | Yes | `True` |
| 1 | 3 | 6 | Yes | `True` |
| 2 | 5 | 8 | Yes | `True` |
| 3 | 1 | 4 | No | `False` |
| 4 | 3 | 6 | Yes | `True` |

Output:

`[True, True, True, False, True]`

---

# Brute Force Approach

For every kid:

1. Add `extraCandies` to their current candies.
2. Compare their new total with every other kid.
3. Determine whether they can reach or exceed the maximum.

This repeatedly searches for the greatest value.

### Complexity

- **Time:** `O(n²)`
- **Space:** `O(n)` for output

This is unnecessary because the maximum of the original array can be found once.

---

# Optimal Approach

## Key Observation

The only value we actually need for comparison is the **maximum number of candies currently held by any kid**.

First calculate:

`max(candies)`

Then for every kid check:

`candies[i] + extraCandies >= maxCandies`

If true:

`True`

Otherwise:

`False`

There is no need to recalculate the maximum for every kid.

---

## Algorithm

1. Find the maximum value in `candies`.
2. Traverse the array.
3. For each kid, calculate their potential number of candies after receiving `extraCandies`.
4. Compare it with the global maximum.
5. Store the boolean result.
6. Return the result.

---

# Dry Run

Input:

`candies = [2, 3, 5, 1, 3]`  
`extraCandies = 3`

### Step 1: Find Maximum

`maxCandies = 5`

### Step 2: Evaluate Each Kid

```text
2 + 3 = 5  → 5 >= 5 → True
3 + 3 = 6  → 6 >= 5 → True
5 + 3 = 8  → 8 >= 5 → True
1 + 3 = 4  → 4 >= 5 → False
3 + 3 = 6  → 6 >= 5 → True

````python
class Solution(object):
    def kidsWithCandies(self, candies, extraCandies):
        """
        :type candies: List[int]
        :type extraCandies: int
        :rtype: List[bool]
        """
        max_can=max(candies)
        result=[]
        
        for i in candies:
            if i+extraCandies>=max_can:
                result.append(True)
            else:
                result.append(False)
        return result
````
