# LC 1295 - Find Numbers with Even Number of Digits

## Problem

Given an array `nums`, return the number of integers that contain an even number of digits.

### Example

Input:
`nums = [12, 345, 2, 6, 7896]`

Output:
`2`

Explanation:
- `12` → 2 digits → even → count
- `345` → 3 digits → odd
- `2` → 1 digit → odd
- `6` → 1 digit → odd
- `7896` → 4 digits → even → count

Answer = `2`

---

# Approach 1: String Conversion

## Idea

Convert every number into a string and check the length.

Mental process:

`number → string → count digits → check even`

For example:

`7896 → "7896" → length = 4 → even`

### Complexity

- Time: `O(n × d)`, where `d` is the number of digits
- Auxiliary Space: `O(d)` for the temporary string

Under the problem's bounded constraints, this behaves effectively like `O(n)`.

---

# Approach 2: Mathematical Digit Counting

This is the approach used in the final solution.

## Key Observation

Every time we integer-divide a positive number by `10`, we remove its last digit.

Example:

`7896`

`7896 → 789 → 78 → 7 → 0`

Number of divisions = number of digits.

Therefore, we can count how many times we divide by `10`.

---

## Algorithm

For every number:

1. Set `sub = 0`.
2. While the number is greater than `0`:
   - Divide the number by `10`.
   - Increment `sub`.
3. `sub` now represents the number of digits.
4. If `sub % 2 == 0`, increment the answer.
5. Return the answer.

---

# Dry Run

Suppose:

`nums = [12, 345, 7896]`

### Number: 12

`12 → 1 → 0`

Number of divisions = `2`

`2 % 2 == 0`

So count = `1`.

### Number: 345

`345 → 34 → 3 → 0`

Number of divisions = `3`

`3 % 2 != 0`

Don't count.

### Number: 7896

`7896 → 789 → 78 → 7 → 0`

Number of divisions = `4`

`4 % 2 == 0`

Count it.

Final answer = `2`.

# Secondary Pattern

## Repeated Reduction / Digit Manipulation

The inner `while` loop repeatedly transforms one number until it reaches a termination condition.

General mental model:

`current value → transform → transform → transform → termination`

This pattern appears frequently in problems involving:

- Digit counting
- Digit sums
- Reversing numbers
- Extracting digits
- Checking numeric properties

# Common Mistakes

### 1. Forgetting that each number needs its own digit counter

The digit count must reset for every number.

Conceptually:

`number 1 → sub = 0`

`number 2 → sub = 0`

`number 3 → sub = 0`

---

### 2. Checking the original number after modifying it

The repeated division changes the local variable.

For example:

`7896 → 789 → 78 → 7 → 0`

So don't expect the variable to still contain `7896` afterward.

---

### 3. Thinking nested loops automatically mean O(n²)

They don't.

The outer loop processes `n` numbers.

The inner loop runs once per digit.

Therefore:

`O(n × d)`

where `d` is the maximum number of digits.

Since `d` is bounded by the problem constraints, this is effectively `O(n)`.

---

# Alternative: Range Checking

There is also a constraint-based trick.

Numbers with an even number of digits fall into ranges:

- 2 digits → `10-99`
- 4 digits → `1000-9999`
- 6 digits → `100000-999999`

So you could check whether a number falls into one of those ranges.

However, this is **not the best general pattern to learn** because it depends on the allowed constraints.

If the problem suddenly allowed 8-digit or 10-digit numbers, the conditions would need to be extended.

Therefore:

**Range checking = clever constraint trick**

**Repeated division = reusable DSA technique**

---

# Complexity

For the mathematical solution:

- **Time:** `O(n × d)`
- **Auxiliary Space:** `O(1)`

Where:
- `n` = number of elements
- `d` = number of digits in a number

With bounded integer sizes, we generally treat this as:

**Time: O(n)**

**Space: O(1)**

# Interview Follow-Ups

### Q1. Can you solve it without converting numbers to strings?

Yes.

Use repeated division by `10`.

### Q2. Why does dividing by 10 remove one digit?

Integer division discards the remainder.

Example:

`7896 // 10 = 789`

The last digit `6` is removed.

### Q3. Why does the number of divisions equal the number of digits?

Each division removes exactly one digit until the number becomes `0`.

### Q4. Can this be done without modifying the original array?

Yes.

The solution only needs a counter and temporary variables.

### Q5. Can you do it using logarithms?

For positive integers, digit count can be calculated using:

`floor(log10(num)) + 1`

But this introduces floating-point considerations and is unnecessary here.

```python
class Solution(object):
    def findNumbers(self, nums):
        """
        :type nums: List[int]
        :rtype: int
        """
        count=0

        for i in nums:
            sub=0
            while i>0:
                i=i//10
                sub+=1
            if sub%2==0:
                count+=1
        return count
```

```python
class Solution(object):
    def findNumbers(self, nums):
        """
        :type nums: List[int]
        :rtype: int
        """
        count = 0

        for num in nums:
            if (10 <= num <= 99) or (1000 <= num <= 9999) or (100000 <= num <= 999999):
                count += 1

        return count
```
