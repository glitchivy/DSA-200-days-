# LeetCode 1480 - Running Sum of 1d Array

## Problem

Given an integer array `nums`, return the running sum of the array.

The running sum at index `i` is:

\[
runningSum[i] = nums[0] + nums[1] + \dots + nums[i]
\]

### Example

Input: `[1, 2, 3, 4]`

Output: `[1, 3, 6, 10]`

---

## Brute Force Approach

### Idea

For every index `i`:

1. Start from index `0`.
2. Calculate the sum from `0` to `i`.
3. Store the result.

Example:

- Index 0 → `1`
- Index 1 → `1 + 2 = 3`
- Index 2 → `1 + 2 + 3 = 6`
- Index 3 → `1 + 2 + 3 + 4 = 10`

### Problem With This Approach

The same elements are repeatedly added.

For example, `nums[0]` is used in the calculation for every index.

### Complexity

- Time: `O(n²)`
- Space: `O(n)` for output

---

## Optimal Approach

### Key Observation

The current running sum can be calculated using the previous running sum.

`runningSum[i] = runningSum[i - 1] + nums[i]`

Instead of recalculating the entire prefix, carry the previous result forward.

### Core Logic

Previous State + Current Element = Current State

For this problem:

Previous Running Sum + Current Element = Current Running Sum

### Algorithm

1. Initialize a cumulative sum.
2. Traverse the array once.
3. Add the current element to the cumulative sum.
4. Store the updated sum.
5. Return the result.

---

## Dry Run

Input: `nums = [1, 2, 3, 4]`

| Index | Element | Previous Sum | Current Sum |
|------:|--------:|-------------:|------------:|
| 0 | 1 | 0 | 1 |
| 1 | 2 | 1 | 3 |
| 2 | 3 | 3 | 6 |
| 3 | 4 | 6 | 10 |

Output: `[1, 3, 6, 10]`

---

## In-Place Optimization

If modifying the input array is allowed, store the running sum directly inside `nums`.

Before: `[1, 2, 3, 4]`

After: `[1, 3, 6, 10]`

- Time: `O(n)`
- Extra Space: `O(1)`

---

# Primary Pattern

## Array Traversal + Accumulator

The deeper pattern is:

> Maintain state while traversing the array.

General structure:

`state = initial value`

For every element:

`update state using current element`

Then:

`use/store state`

For this problem, the state is the running sum.

---

## Pattern Recognition

Think **Accumulator** when:

- The problem asks for a cumulative result.
- The current result depends on previous elements.
- The previous result can be reused.
- Recalculating previous elements would repeat work.
- The array can be processed sequentially.

### Trigger Question

> "Can I carry information from the previous iteration into the current iteration?"

If yes, consider an accumulator.

---

## Generalization

The accumulator does not have to be a sum.

The same pattern can maintain:

- Running sum
- Running maximum
- Running minimum
- Count
- Product
- Best value so far
- Other cumulative state

General idea:

`Previous State + Current Information → Updated State`

---

## Common Mistakes

### 1. Recalculating the Prefix

Calculating the sum from the beginning for every index leads to `O(n²)`.

**Fix:** Reuse the previous cumulative result.

### 2. Forgetting the Current Element

Current State = Previous State + Current Element

### 3. Using Unnecessary Extra Space

If the input can be modified, a separate output array may be unnecessary.

---

## Complexity

### Separate Output Array

- Time: `O(n)`
- Extra Space: `O(n)`

### In-Place

- Time: `O(n)`
- Extra Space: `O(1)`

---

## Interview Follow-Ups

### Can you solve it in O(n)?

Yes. Traverse once while maintaining cumulative state.

### Can you do it in O(1) extra space?

Yes, if modifying the input array is allowed.

### Why is brute force inefficient?

Because it repeatedly recalculates sums that were already computed.

### What broader pattern does this demonstrate?

Maintaining state while traversing an array.

```python
class Solution(object):
    def runningSum(self, nums):
        """
        :type nums: List[int]
        :rtype: List[int]
        """
        sum_num=0
        arr=[]
        for i in nums:
            sum_num+=i
            arr.append(sum_num)
        return arr
```
