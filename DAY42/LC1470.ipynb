# LeetCode 1470 - Shuffle the Array

## Problem

You are given an array `nums` containing `2n` elements in the form:

`[x1, x2, ..., xn, y1, y2, ..., yn]`

Return the array in the shuffled form:

`[x1, y1, x2, y2, ..., xn, yn]`

### Example

Input:

`nums = [2, 5, 1, 3, 4, 7]`  
`n = 3`

Split the array logically:

`x = [2, 5, 1]`  
`y = [3, 4, 7]`

Interleave them:

`[2, 3, 5, 4, 1, 7]`

Output:

`[2, 3, 5, 4, 1, 7]`

---

# Brute Force Approach

A straightforward approach is to repeatedly find the next required element and construct the shuffled result.

However, this can involve unnecessary searching or shifting of elements.

The important observation is that the positions of all required elements are already known.

---

# Optimal Approach

## Key Observation

The array contains two logical halves:

`[x1, x2, ..., xn | y1, y2, ..., yn]`

For every index `i`:

`xᵢ → nums[i]`

`yᵢ → nums[n + i]`

The output needs:

`xᵢ → yᵢ`

Therefore, for every `i` from `0` to `n-1`:

1. Take `nums[i]`
2. Take `nums[n + i]`
3. Add them to the result

---

## Index Mapping

This is the most important part of the problem.

For:

`nums = [2, 5, 1, 3, 4, 7]`

and `n = 3`:

| i | `nums[i]` | `nums[n+i]` | Output |
|---:|---:|---:|---|
| 0 | 2 | 3 | 2, 3 |
| 1 | 5 | 4 | 5, 4 |
| 2 | 1 | 7 | 1, 7 |

Final result:

`[2, 3, 5, 4, 1, 7]`

---

# Algorithm

1. Create an empty result array.
2. Traverse `i` from `0` to `n-1`.
3. Add `nums[i]` to the result.
4. Add `nums[n+i]` to the result.
5. Return the result.

# Primary Pattern

## Array Traversal + Index Mapping / Interleaving

This problem is not about sorting.

The array already contains the elements in the correct relative order. We only need to access them using the correct indices.

The key relationship is:

`xᵢ → nums[i]`

`yᵢ → nums[n+i]`

Output positions:

`xᵢ → position 2i`

`yᵢ → position 2i+1`

---

# Pattern Recognition

Think **Index Mapping** when:

- The input has multiple logical sections.
- The required output rearranges elements according to their positions.
- There is a predictable relationship between input and output indices.
- No comparison between element values is required.
- The order of elements needs to be preserved.

### Trigger Question

> "Can I derive the source index directly from the current output position?"

If yes, look for an **index mapping** solution.

---

# Why This Is Not Merge Sort

Merge sort also deals with multiple sections of an array, but its purpose is completely different.

### Merge Sort

`Divide → Sort → Compare → Merge`

It compares values to determine their order.

### LC 1470

`Split logically → Map indices → Interleave`

It does not compare values or sort anything.

Using merge sort here would add unnecessary complexity.

---

# In-Place Consideration

An in-place `O(1)` auxiliary-space solution is theoretically possible, but it requires carefully rearranging elements without overwriting values that have not yet been processed.

For this problem, the simple output-array solution is preferable unless the interviewer explicitly requires:

> "Do this in-place with O(1) extra space."

The important thing is not to optimize space blindly when the simpler `O(n)` solution already has optimal time complexity.

---

# Complexity

- **Time:** `O(n)` because every element is processed exactly once.
- **Extra Space:** `O(n)` for the output array.

`O(n)` time is optimal because the final answer contains `2n` elements and must be produced.

---

# Common Mistakes

### 1. Mixing up the two halves

Remember:

`nums[i]` → first half

`nums[n+i]` → second half

### 2. Using the wrong offset

The second half starts at index `n`, not `n+1`.

Correct:

`n + i`

### 3. Sorting the array

Sorting destroys the required relative order.

The problem is about **position**, not value.

### 4. Overengineering with in-place manipulation

Unless `O(1)` space is explicitly required, the output-array solution is simpler and easier to reason about.

---

# Interview Follow-Ups

### Can you solve it in O(n)?

Yes. Each pair is processed once.

### Can you do better than O(n) time?

No. The output itself contains `2n` elements.

### Can you do it with O(1) extra space?

Yes, with a more complicated in-place rearrangement approach.

### What is the key index relationship?

`nums[i]` and `nums[n+i]` form the `i`th pair.

### What pattern does this problem teach?

Index mapping and structured array traversal.

# Key Takeaway

The main insight is:

`[x1, x2, ..., xn | y1, y2, ..., yn]`

can be viewed as two logical arrays:

`x = nums[0:n]`

`y = nums[n:2n]`

For every `i`:

`output = xᵢ, yᵢ`

or equivalently:

`nums[i], nums[n+i]`

The problem is therefore an **index mapping problem**, not a sorting problem.

```python
class Solution(object):
    def shuffle(self, nums, n):
        """
        :type nums: List[int]
        :type n: int
        :rtype: List[int]
        """
        arr=[]
        for i in range(n):
            
            arr.append(nums[i])
            arr.append(nums[n+i])
        return arr
```
