# LeetCode 1920 - Build Array from Permutation

## Problem

You are given a zero-based permutation `nums` of length `n`.

Construct an array `ans` such that:

`ans[i] = nums[nums[i]]`

Return `ans`.

### Example

Input:

`nums = [0, 2, 1, 5, 3, 4]`

Output:

`[0, 1, 2, 4, 5, 3]`

---

## Brute Force Approach

The direct approach is already efficient because every output element can be calculated independently.

For every index `i`:

1. Read `nums[i]`.
2. Treat that value as an index.
3. Access `nums[nums[i]]`.
4. Store the result.

There is no need for nested traversal because the permutation guarantees that `nums[i]` is a valid index.

### Complexity

- **Time:** `O(n)`
- **Space:** `O(n)` for the output array

There is no meaningful `O(n²)` brute-force solution needed here. The direct traversal is already optimal in time.

---

# Optimal Approach

## Key Observation

The value at `nums[i]` tells us **which index to access next**.

Therefore:

`ans[i] = nums[nums[i]]`

Break it into two steps:

`nums[i]` → gives an index

`nums[index]` → gives the required value

So:

`nums[i] → index → nums[index]`

---

## Algorithm

1. Initialize an empty result array.
2. Traverse every index `i` from `0` to `n-1`.
3. Read `nums[i]`.
4. Use `nums[i]` as an index into `nums`.
5. Store `nums[nums[i]]` in the result.
6. Return the result.

---

# Dry Run

Input:

`nums = [0, 2, 1, 5, 3, 4]`

| Index `i` | `nums[i]` | Access | Result |
|----------:|-----------:|--------|-------:|
| 0 | 0 | `nums[0]` | 0 |
| 1 | 2 | `nums[2]` | 1 |
| 2 | 1 | `nums[1]` | 2 |
| 3 | 5 | `nums[5]` | 4 |
| 4 | 3 | `nums[3]` | 5 |
| 5 | 4 | `nums[4]` | 3 |

Output:

`[0, 1, 2, 4, 5, 3]`

# Primary Pattern

## Indirect Indexing / Array as a Mapping

The important idea is that the array is being used as a **mapping from one index to another value**.

Instead of:

`i → nums[i]`

we perform:

`i → nums[i] → nums[nums[i]]`

The first lookup tells us **where to look**.

The second lookup gives us **the answer**.

---

# Pattern Recognition

Think **Indirect Indexing** when:

- Array values are guaranteed to be valid indices.
- A value in the array tells you where another value can be found.
- The problem contains expressions such as:
  - `nums[nums[i]]`
  - `arr[arr[i]]`
  - `arr[index]` where `index` comes from another array element.
- The array represents a relationship or mapping between indices.

### Trigger Question

> "Is the current element telling me which index I should access next?"

If yes, think **indirect indexing**.

---

# Important Distinction

### Normal Array Traversal

`nums[i]`

Meaning:

> "Give me the value at index `i`."

### Indirect Indexing

`nums[nums[i]]`

Meaning:

> "First give me the value at index `i`. Then use that value as another index."

This distinction is extremely important for recognizing permutation and mapping problems.

---

# Why the Permutation Matters

A permutation guarantees that every `nums[i]` is a valid index from:

`0` to `n-1`

Therefore:

`nums[nums[i]]`

will always access a valid position.

Without this guarantee, indirect indexing could cause an out-of-bounds access.

---

# Complexity

- **Time:** `O(n)`
- **Extra Space:** `O(n)` for the output array

### Why `O(n)` Time?

Every index is processed exactly once.

### Can Time Be Better?

No.

The output contains `n` elements, so producing the complete answer already requires `O(n)` work.

---

# In-Place Consideration

The straightforward solution uses a separate output array.

Trying to modify `nums` directly is more complicated because changing an element can destroy information needed for later calculations.

For the standard problem, the separate result array is the clean and practical approach.

---

# Common Mistakes

### 1. Confusing `nums[i]` with `nums[nums[i]]`

`nums[i]` gives the mapped index.

`nums[nums[i]]` gives the final answer.

### 2. Using the value directly as the answer

The value `nums[i]` is not necessarily the answer.

It tells us **where to look**.

### 3. Thinking nested indexing means nested loops

`nums[nums[i]]` contains two array accesses, but it is still `O(1)` work for each `i`.

Therefore the entire traversal is:

`O(n)`

### 4. Ignoring the permutation constraint

The fact that every value is a valid index is what makes the second lookup safe.

---

# Interview Follow-Ups

### What is the main idea?

Use each array value as an index to perform another lookup.

### Why is the solution O(n)?

There is one constant-time lookup operation for each of the `n` elements.

### Why does `nums[nums[i]]` work?

Because `nums[i]` is guaranteed to be a valid index.

### Can you do better than O(n) time?

No, because `n` output elements must be produced.

### What pattern does this problem demonstrate?

Indirect indexing and using an array as a mapping.

```python
class Solution(object):
    def buildArray(self, nums):
        """
        :type nums: List[int]
        :rtype: List[int]
        """
        arr=[]
        n=len(nums)
        for i in range(n):
            arr.append(nums[nums[i]])
        return arr
```
