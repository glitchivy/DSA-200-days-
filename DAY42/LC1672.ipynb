# LeetCode 1672 - Richest Customer Wealth

## Problem

You are given a 2D integer array `accounts`.

`accounts[i][j]` represents the amount of money the `i`th customer has in the `j`th bank.

The **wealth of a customer** is the total amount of money they have across all banks.

Return the wealth of the **richest customer**.

### Example

Input:

`accounts = [[1,2,3],[3,2,1]]`

Customer 0:

`1 + 2 + 3 = 6`

Customer 1:

`3 + 2 + 1 = 6`

Output:

`6`

---

# Brute Force Approach

For every customer:

1. Calculate their total wealth.
2. Compare their wealth with every other customer.
3. Keep track of the richest customer.

The comparison between customers is unnecessary because we can maintain the maximum while calculating each customer's wealth.

### Complexity

- **Time:** `O(n × k)`
- **Space:** `O(1)`

Even though this still has `O(n × k)` time, the approach can be described more simply as calculating every customer's wealth and maintaining the maximum.

---

# Optimal Approach

## Key Observation

We only need two pieces of information:

1. The **current customer's total wealth**
2. The **maximum wealth found so far**

For each row:

`current_sum = sum of all values in that row`

Then:

`maximum_wealth = max(maximum_wealth, current_sum)`

There is no need to store the wealth of every customer.

---

## Algorithm

1. Initialize `maximum_wealth = 0`.
2. Traverse each customer.
3. Initialize `current_sum = 0`.
4. Traverse every bank belonging to that customer.
5. Add the bank's money to `current_sum`.
6. Compare `current_sum` with `maximum_wealth`.
7. Keep the larger value.
8. Return `maximum_wealth`.

---

# Dry Run

Input:

`accounts = [[1,2,3],[3,2,1],[4,1,2]]`

### Customer 0

`1 + 2 + 3 = 6`

`maximum_wealth = 6`

### Customer 1

`3 + 2 + 1 = 6`

`maximum_wealth = max(6,6) = 6`

### Customer 2

`4 + 1 + 2 = 7`

`maximum_wealth = max(6,7) = 7`

Output:

`7`

---

# Primary Pattern

## 2D Array Traversal + Accumulator + Global Maximum

This problem combines patterns learned previously.

```text
2D Array Traversal
        ↓
Accumulator
        ↓
Calculate each row's total
        ↓
Global Maximum
        ↓
Keep the best result

````python
class Solution(object):
    def maximumWealth(self, accounts):
        """
        :type accounts: List[List[int]]
        :rtype: int
        """
        n=len(accounts)
        k=len(accounts[0])
        msum=0
        for i in range(n):
            sum_max=0
            for j in range(k):
                sum_max+=accounts[i][j]
            msum=max(msum,sum_max)
        return msum
````
