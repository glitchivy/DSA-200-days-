# LC 1365 - How Many Numbers Are Smaller Than the Current Number

## Problem

Given an integer array `nums`, for each element `nums[i]`, count how many numbers in the array are strictly smaller than it.

Return the resulting array.

### Example

Input:
`nums = [8, 1, 2, 2, 3]`

Output:
`[4, 0, 1, 1, 3]`

Explanation:

- `8` → `1, 2, 2, 3` are smaller → `4`
- `1` → nothing is smaller → `0`
- `2` → only `1` is smaller → `1`
- `2` → only `1` is smaller → `1`
- `3` → `1, 2, 2` are smaller → `3`

---

# Approach 1: Brute Force

This is the approach I used.

## Idea

For every number:

1. Traverse the entire array.
2. Compare every other number with the current number.
3. If the other number is smaller, increment a counter.
4. Store the counter in the result.

Mental model:

`Pick one element`
→
`Compare with every element`
→
`Count smaller values`
→
`Move to next element`

---

# Algorithm

For each `i` in `nums`:

1. Set `count = 0`.
2. Traverse `nums` again.
3. If `nums[j] < nums[i]`, increment `count`.
4. Append `count` to the result.
5. Return the result.

---

# Dry Run

Input:

`nums = [8, 1, 2, 2, 3]`

### Current = 8

Compare with everything:

- `8 < 8` → No
- `1 < 8` → Yes
- `2 < 8` → Yes
- `2 < 8` → Yes
- `3 < 8` → Yes

Count = `4`

Result:

`[4]`

### Current = 1

- `8 < 1` → No
- `1 < 1` → No
- `2 < 1` → No
- `2 < 1` → No
- `3 < 1` → No

Count = `0`

Result:

`[4, 0]`

### Current = 2

Only `1` is smaller.

Count = `1`

Result:

`[4, 0, 1]`

The second `2` also gives `1`.

The `3` has three smaller numbers:

`1, 2, 2`

Final:

`[4, 0, 1, 1, 3]`

# Primary Pattern

## Nested Array Traversal + Comparison Counting

The pattern is:

`For each element`
→
`scan the array`
→
`check a relationship`
→
`count matches`

This is a very common brute-force pattern.

You should immediately recognize it when a problem asks:

- How many elements are smaller?
- How many elements are greater?
- How many elements satisfy a condition relative to the current element?
- How many pairs satisfy some relationship?

---

# Pattern Recognition

The trigger question is:

> "For each element, do I need to compare it against every other element?"

If yes, the first solution you should consider is:

`Outer loop = current element`

`Inner loop = candidates to compare against`

This naturally produces:

`O(n²)`

---

# Important Observation

The brute-force solution does a lot of repeated work.

For example, when checking:

`8`

we repeatedly ask:

`Is 1 smaller than 8?`

`Is 2 smaller than 8?`

When checking another number, we again scan the same array.

So the natural optimization question is:

> "Can I preprocess the array so I don't have to compare every pair?"

That leads to the sorting approach.

---

# Approach 2: Sorting + Mapping

Sort the array:

`[8, 1, 2, 2, 3]`

becomes:

`[1, 2, 2, 3, 8]`

Now notice:

- `1` has `0` elements before it
- `2` has `1` element before its first occurrence
- `3` has `3` elements before it
- `8` has `4` elements before it

Therefore:

`1 → 0`

`2 → 1`

`3 → 3`

`8 → 4`

Then use this mapping to build the answer in the original order.

### Complexity

- Sorting: `O(n log n)`
- Mapping: `O(n)`
- Result construction: `O(n)`
- Total: **O(n log n)**

---

# Approach 3: Frequency Counting

LC 1365 has a useful constraint:

`0 <= nums[i] <= 100`

This means there are only **101 possible values**.

Instead of sorting or comparing every pair, we can count how frequently each value occurs.

For example:

`nums = [8, 1, 2, 2, 3]`

Frequency information:

`1 → 1`

`2 → 2`

`3 → 1`

`8 → 1`

Then we can calculate how many numbers are smaller than each value.

This produces an **O(n)** solution.

This is an important pattern called:

## Frequency Array / Counting Array

# Common Mistakes

### 1. Using `<=` instead of `<`

The problem asks for numbers **strictly smaller**.

For:

`[2, 2]`

Neither `2` has a smaller number.

Answer:

`[0, 0]`

So the condition must be:

`other < current`

not:

`other <= current`

---

### 2. Forgetting duplicate values

Duplicates do not count as smaller.

Example:

`[5, 5, 5]`

Answer:

`[0, 0, 0]`

---

### 3. Accidentally comparing an element only with elements after it

You need to compare against the **entire array**, not just elements to the right.

---

# Complexity of Your Solution

For every element, you scan all `n` elements.

Therefore:

**Time: O(n²)**

The returned result requires `O(n)` space.

If we only consider auxiliary space apart from the output:

**Auxiliary Space: O(1)**

# Interview Follow-Ups

### Q1. Can you improve O(n²)?

Yes.

Possible approaches:

- Sorting + mapping → `O(n log n)`
- Frequency array using the constraints → `O(n)`

### Q2. Why can we use a frequency array?

Because the values are limited to `0-100`.

### Q3. Why is sorting useful?

In sorted order, the position of the first occurrence of a value tells us how many values are strictly smaller.

### Q4. What pattern does the brute force use?

Nested traversal with a relative comparison.

### Q5. What should you ask when you see O(n²)?

Ask:

> "Am I repeating comparisons or calculations that I could preprocess?"

This question often leads to the optimized solution.

```python
class Solution(object):
    def smallerNumbersThanCurrent(self, nums):
        """
        :type nums: List[int]
        :rtype: List[int]
        """
        result = []

        for i in nums:
            count = 0
            for j in nums:
                if j < i:
                    count += 1
            result.append(count)

        return result
```
